# Historical & Related Work

The history of NER is tied to the evolution of NLP itself:

### Rule-based systems (1980s–1990s)
- Early systems relied on **hand-crafted linguistic rules** and **gazetteers** (dictionaries of names).  
- Example: If a word follows *“Mr.”*, it is likely a **Person**.  
- **Limitation:** Not scalable across domains or languages.  

### Statistical Models (1990s–2000s)
- With more annotated data, researchers shifted to **probabilistic models** like:  
  - Hidden Markov Models (HMMs)  
  - Maximum Entropy Models  
  - Conditional Random Fields (CRFs)  
- CRFs became a **gold standard** for NER, as they model label dependencies.  

### Neural Models (2010s–present)
- With the rise of deep learning, **Recurrent Neural Networks (RNNs)**, especially **Long Short-Term Memory (LSTM)** models, improved performance by capturing **contextual dependencies** in text.  
- Later, **BiLSTM-CRF** became the state-of-the-art (combining sequence modeling with structured prediction).  
- Today, **transformers** (BERT, RoBERTa, GPT-based models) dominate NER, but LSTM remains important for **foundational understanding** and **resource-constrained setups**.  

---

# Why Data Analysis Matters in Classical NLP

Before deep learning, classical NLP models heavily depended on:

- **Feature engineering** (POS tags, capitalization, prefixes/suffixes).  
- **Statistical assumptions** (Markov property in HMMs, linear dependencies in CRFs).  
- **Data preprocessing** (removing noise, balancing entity types).  

Thus, **data analysis** was crucial:  

- Understanding class distribution helped avoid bias.  
- Identifying rare entity types guided feature design.  
- Detecting errors in annotation improved robustness.  

Even in neural methods, exploratory analysis remains vital for **token distributions, sequence lengths, and vocabulary coverage**.

````python
def sentence_word_counts(df, col="tokens"):
    """
    Return a list with the number of words in each sentence.
    """
    return [len(tokens) for tokens in df[col]]

def plot_sentence_length_distribution(counts, title="Sentence Length Distribution"):
    """
    Plot histogram of sentence lengths.
    """
    plt.figure(figsize=(10, 6))
    plt.hist(counts, bins=50, alpha=0.7, color='skyblue', edgecolor='black')
    plt.xlabel('Sentence Length (number of tokens)')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.grid(axis='y', alpha=0.75)
    plt.show()
    
    # Print statistics
    print(f"Mean sentence length: {np.mean(counts):.2f}")
    print(f"Median sentence length: {np.median(counts):.2f}")
    print(f"Std sentence length: {np.std(counts):.2f}")
    print(f"Max sentence length: {np.max(counts)}")
    print(f"Min sentence length: {np.min(counts)}")
````

````python
def word_lengths(df, col="tokens"):
    """
    Return a list with the lengths of all words in the dataset.
    """
    all_lengths = []
    for tokens in df[col]:
        all_lengths.extend([len(token) for token in tokens])
    return all_lengths
````

````python
def plot_word_length_distribution(lengths, title="Word Length Distribution"):
    """
    Plot histogram of word lengths.
    """
    plt.figure(figsize=(10, 6))
    plt.hist(lengths, bins=30, alpha=0.7, color='lightcoral', edgecolor='black')
    plt.xlabel('Word Length (number of characters)')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.grid(axis='y', alpha=0.75)
    plt.show()
    
    # Print statistics
    print(f"Mean word length: {np.mean(lengths):.2f}")
    print(f"Median word length: {np.median(lengths):.2f}")
    print(f"Std word length: {np.std(lengths):.2f}")
    print(f"Max word length: {np.max(lengths)}")
    print(f"Min word length: {np.min(lengths)}")
````

###1.3 NER Tag Distribution


````python
def extract_ner_tags(df, col="ner_tags"):
    """
    Return a flat list of all NER tags in the dataset.
    """
    all_tags = []
    for tags in df[col]:
        all_tags.extend(tags)
    return all_tags

def plot_ner_tag_distribution(tags, label_list, title="NER Tag Distribution", top_n=None):
    """
    Plot distribution of NER tags (flattened).
    """
    # Convert tag indices to names
    tag_names = [label_list[tag] for tag in tags]
    tag_counts = Counter(tag_names)
    
    if top_n:
        # Get top N most common tags
        most_common = tag_counts.most_common(top_n)
        labels, counts = zip(*most_common)
    else:
        labels, counts = zip(*tag_counts.most_common())
    
    plt.figure(figsize=(12, 6))
    bars = plt.bar(labels, counts, color='lightgreen', alpha=0.7, edgecolor='black')
    plt.xlabel('NER Tags')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', alpha=0.75)
    
    # Add value labels on bars
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                f'{int(height)}', ha='center', va='bottom')
    
    plt.tight_layout()
    plt.show()
````

````python
def tag_popularity(tags, label_list, top_k=5):
    """
    Return most and least frequent tags.
    """
    tag_names = [label_list[tag] for tag in tags]
    tag_counts = Counter(tag_names)
    
    most_common = tag_counts.most_common(top_k)
    least_common = tag_counts.most_common()[-top_k:]
    
    print("Most frequent tags:")
    for tag, count in most_common:
        print(f"  {tag}: {count} ({count/len(tags)*100:.2f}%)")
    
    print("\nLeast frequent tags:")
    for tag, count in least_common:
        print(f"  {tag}: {count} ({count/len(tags)*100:.2f}%)")
    
    return most_common, least_common
````

###1.4 Top Words per Entity Type


````python
def top_words_for_tag(df, token_col="tokens", tag_col="ner_tags", target_tag_id=None, target_tag_name=None, top_n=10):
    """
    Return the most frequent words assigned a given tag (by index or name).
    """
    if target_tag_name and not target_tag_id:
        target_tag_id = LABEL_LIST.index(target_tag_name)
    
    word_counter = Counter()
    for tokens, tags in zip(df[token_col], df[tag_col]):
        for token, tag in zip(tokens, tags):
            if tag == target_tag_id:
                word_counter[token] += 1
    
    return word_counter.most_common(top_n)
````

````python
def plot_top_words_for_tag(tag_name, top_words, title_prefix="Top Words for"):
    """
    Plot the top words associated with a given tag.
    """
    if not top_words:
        print(f"No words found for tag: {tag_name}")
        return
    
    words, counts = zip(*top_words)
    
    plt.figure(figsize=(10, 6))
    bars = plt.barh(words, counts, color='skyblue', alpha=0.7, edgecolor='black')
    plt.xlabel('Frequency')
    plt.ylabel('Words')
    plt.title(f'{title_prefix} {tag_name}')
    plt.gca().invert_yaxis()  # Display highest frequency at top
    plt.grid(axis='x', alpha=0.75)
    
    # Add value labels on bars
    for bar in bars:
        width = bar.get_width()
        plt.text(width, bar.get_y() + bar.get_height()/2.,
                f'{int(width)}', ha='left', va='center')
    
    plt.tight_layout()
    plt.show()
````

### 1.5 N-Gram Tag Distributions

N-gram tag frequency distributions at unigram, bigram, and trigram levels.

**(i).** Unigram Tag Frequencies:

Unigram distributions show the overall frequency of individual tags.
  * This distribution typically follows Zipf's law.
  (Do you remember this concept form you past assignments and does the plots you get are following the law?) .
  
  * The unigram frequencies serve as the baseline emission probabilities in HMM models, indicating how likely each word is to be tagged with a specific tag.

    * For example, in English corpora, words like articles and prepositions dominate the frequency distribution, while content words like specific nouns appear less frequently. This pattern reflects the fundamental structure of natural language where a small number of elements are used repeatedly.

**(ii).** Bigram Tag Frequencies:

Bigram tag sequences capture the transition patterns between consecutive tags, which form the core of HMM transition probabilities.


**(iii).** Trigram Tag Frequencies:

Trigram patterns provide deeper contextual understanding by capturing longer-range dependencies between three consecutive tags.

````python
def generate_ngrams(seq, n):
    """
    Yield n-grams from a sequence.
    """
    for i in range(len(seq) - n + 1):
        yield tuple(seq[i:i + n])

def tag_ngram_distribution(df_tags, n, label_list):
    """
    Return n-gram tag frequency Counter.
    """
    ngram_counter = Counter()
    for tags in df_tags:
        tag_names = [label_list[tag] for tag in tags]
        ngrams = list(generate_ngrams(tag_names, n))
        ngram_counter.update(ngrams)
    return ngram_counter
````

````python
def plot_ngram_distribution(counter, title, top_n=10):
    """
    Plot top-N n-grams as bar chart.
    """
    if not counter:
        print(f"No n-grams found for {title}")
        return
    
    most_common = counter.most_common(top_n)
    ngrams, counts = zip(*most_common)
    
    # Convert n-grams to strings for plotting
    ngram_labels = [' '.join(ngram) for ngram in ngrams]
    
    plt.figure(figsize=(12, 6))
    bars = plt.bar(ngram_labels, counts, color='salmon', alpha=0.7, edgecolor='black')
    plt.xlabel(f'{len(ngrams[0])}-grams')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', alpha=0.75)
    
    # Add value labels on bars
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                f'{int(height)}', ha='center', va='bottom')
    
    plt.tight_layout()
    plt.show()
````

````python
### 1.6 Execution Block

print("=== Dataset Summary ===")
print(f"Training set size: {len(df_train)} sentences")
print(f"Validation set size: {len(df_val)} sentences")
print(f"Test set size: {len(df_test)} sentences")

# Calculate total tokens
total_tokens_train = sum(len(tokens) for tokens in df_train['tokens'])
total_tokens_val = sum(len(tokens) for tokens in df_val['tokens'])
total_tokens_test = sum(len(tokens) for tokens in df_test['tokens'])

print(f"Total tokens in training set: {total_tokens_train}")
print(f"Total tokens in validation set: {total_tokens_val}")
print(f"Total tokens in test set: {total_tokens_test}")

# 1. Sentence Length Distribution
print("\n=== Sentence Length Analysis (Training Set) ===")
train_sent_lengths = sentence_word_counts(df_train)
plot_sentence_length_distribution(train_sent_lengths, "Training Set - Sentence Length Distribution")

# 2. Word Length Distribution
print("\n=== Word Length Analysis (Training Set) ===")
train_word_lengths = word_lengths(df_train)
plot_word_length_distribution(train_word_lengths, "Training Set - Word Length Distribution")

# 3. Tag Distribution
print("\n=== NER Tag Distribution (Training Set) ===")
train_tags = extract_ner_tags(df_train)
plot_ner_tag_distribution(train_tags, LABEL_LIST, "Training Set - NER Tag Distribution", top_n=15)

# Most/Least Frequent Tags
print("\n=== Tag Popularity (Training Set) ===")
most_common, least_common = tag_popularity(train_tags, LABEL_LIST, top_k=5)
print(f"Most Common Tag: {most_common}")
print(f"Least Common Tag: {least_common}")

# 4. Top Words per Tag
print("\n=== Top Words per Entity Type (Training Set) ===")
# Analyze top words for some important entity types
important_tags = ['B-PERSON', 'B-LOCATION', 'B-ORGANIZATION', 'B-TIMEX', 'O']

for tag in important_tags:
    print(f"\n--- {tag} ---")
    top_words = top_words_for_tag(df_train, target_tag_name=tag, top_n=10)
    if top_words:
        print("Top words:")
        for word, count in top_words:
            print(f"  {word}: {count}")
        plot_top_words_for_tag(tag, top_words)

# 5. Tag N-Gram Distributions
print("\n=== N-Gram Tag Distributions (Training Set) ===")

# Unigram
print("\n--- Unigram Tag Frequencies ---")
unigram_counter = tag_ngram_distribution(df_train['ner_tags'], 1, LABEL_LIST)
plot_ngram_distribution(unigram_counter, "Top Unigram Tag Patterns", top_n=15)

# Check Zipf's Law for unigrams
print("\n--- Zipf's Law Analysis for Unigrams ---")
unigram_freqs = [freq for _, freq in unigram_counter.most_common()]
ranks = range(1, len(unigram_freqs) + 1)

plt.figure(figsize=(10, 6))
plt.loglog(ranks, unigram_freqs, 'bo-', alpha=0.7)
plt.xlabel('Rank (log)')
plt.ylabel('Frequency (log)')
plt.title("Zipf's Law: Rank vs Frequency for NER Tags")
plt.grid(True, alpha=0.3)
plt.show()

# Bigram
print("\n--- Bigram Tag Frequencies ---")
bigram_counter = tag_ngram_distribution(df_train['ner_tags'], 2, LABEL_LIST)
plot_ngram_distribution(bigram_counter, "Top Bigram Tag Patterns", top_n=15)

# Trigram
print("\n--- Trigram Tag Frequencies ---")
trigram_counter = tag_ngram_distribution(df_train['ner_tags'], 3, LABEL_LIST)
plot_ngram_distribution(trigram_counter, "Top Trigram Tag Patterns", top_n=15)

# Additional analysis: Entity length distribution
print("\n=== Entity Length Analysis ===")
def analyze_entity_lengths(df, label_list):
    entity_lengths = defaultdict(list)
    
    for tags in df['ner_tags']:
        tag_names = [label_list[tag] for tag in tags]
        
        current_entity = None
        current_length = 0
        
        for tag in tag_names:
            if tag.startswith('B-'):
                # Save previous entity
                if current_entity and current_length > 0:
                    entity_lengths[current_entity].append(current_length)
                # Start new entity
                current_entity = tag[2:]  # Remove 'B-' prefix
                current_length = 1
            elif tag.startswith('I-') and current_entity == tag[2:]:
                # Continue current entity
                current_length += 1
            else:
                # Save previous entity and reset
                if current_entity and current_length > 0:
                    entity_lengths[current_entity].append(current_length)
                current_entity = None
                current_length = 0
        
        # Don't forget the last entity in the sequence
        if current_entity and current_length > 0:
            entity_lengths[current_entity].append(current_length)
    
    return entity_lengths

entity_lengths = analyze_entity_lengths(df_train, LABEL_LIST)

print("Average entity lengths by type:")
for entity_type, lengths in entity_lengths.items():
    if lengths:  # Only show if we have entities of this type
        avg_length = np.mean(lengths)
        print(f"  {entity_type}: {avg_length:.2f} (based on {len(lengths)} entities)")

print("\n=== Analysis Complete ===")
````

### **Task 2: HMM Training (20 Marks)**
---

With the data prepared, you'll now train the Hidden Markov Model. This involves calculating its three core probability distributions from the training corpus using **Maximum Likelihood Estimation**

A **Hidden Markov Model (HMM)** is a statistical model used to analyze sequences of data. It assumes that the system you are observing has a sequence of **hidden states** that you cannot see, which in turn generate a sequence of **observations** that you *can* see.

Think of it like trying to guess the weather (**hidden state**: Sunny or Rainy) just by observing whether someone carries an umbrella (**observation**: Yes or No). The HMM learns two key things:
1.  The probability of the weather changing (e.g., from Sunny to Rainy).
2.  The probability of seeing someone with an umbrella given the weather (e.g., high for Rainy, low for Sunny).


In NER, the goal is to label each word in a sentence with a tag like `PERSON`, `LOCATION`, or `O`. HMM is a natural fit for this task.

* **Observations**: The words in the sentence (e.g., "Ram", "went", "to", "Delhi").
* **Hidden States**: The sequence of NER tags you want to predict (e.g., `B-PER`, `O`, `O`, `B-LOC`).


The HMM is trained on a labeled corpus to learn pi, A, and O which will be defined shortly.

Once trained, the HMM uses the **Viterbi algorithm** to find the most probable sequence of hidden states (NER tags) for a new, unseen sentence (the observations).

#### **Conceptual & Mathematical Details**

An HMM is defined by the following three probability distributions (pi, A, O):

**1. Initial Probability Vector ($\pi$)**:
This vector stores the probability of a sentence starting with a particular NER tag. The $i$-th element of $\pi$, denoted $\pi_i$, is the probability that the first word of a sentence is assigned `tag_i`.

$$\pi_i = P(\text{tag}_i \text{ at start}) = \frac{\text{Count}(\text{sentences starting with tag}_i)}{\text{Total number of sentences}}$$

**2. Transition Probability Matrix ($A$)**:
This matrix captures the relationship between consecutive NER tags. An element $A_{ij}$ stores the probability of transitioning from `tag_i` to `tag_j`. This is crucial for learning patterns like `B-PER` is often followed by `I-PER`.

$$A_{ij} = P(\text{tag}_j | \text{tag}_i) = \frac{\text{Count}(\text{tag}_i \text{ is followed by tag}_j)}{\text{Count}(\text{tag}_i)}$$

**3. Emission Probability Matrix ($O$)**:
This matrix stores the probability of a given NER tag "emitting" a specific word. An element $O_{jk}$ stores the probability that `tag_j` is associated with `word_k`. For example, the word "दिल्ली" will have a high probability of being emitted by the `B-LOC` tag.

$$O_{jk} = P(\text{word}_k | \text{tag}_j) = \frac{\text{Count}(\text{word}_k \text{ appears with tag}_j)}{\text{Count}(\text{tag}_j)}$$

### **Task 3: HMM Training (with Laplace Smoothing) (10 Marks)**
---

A model trained on simple counts suffers from the **zero-probability problem**: it assumes events not seen in training are impossible. For example, if it only saw "Sachin" tagged as a Person (`B-PER`), it would assign a zero probability to "Rahul" also being a Person, causing it to fail on new data.

**Laplace smoothing** (or add-1 smoothing) fixes this.

* **The Idea**: We pretend to have seen every possible outcome one extra time by **adding one** to all our counts before calculating probabilities.

* **Why it Helps**: This ensures no probability is ever exactly zero. The probability of an unseen event like `"Rahul" | B-PER` becomes very small instead of impossible, making the model more **robust** and allowing it to **generalize** to new data.


#### **The Algorithmic Difference**
This idea is implemented by adding 1 to our raw counts in the numerator and adding the number of possible outcomes to the denominator. Here's how it formally changes our calculations:

* **Initial Probability ($\pi$) with Smoothing:**
    $$
    \pi_i = \frac{\text{Count(sentences starting with tag\_i)} + 1}{\text{Total number of sentences} + N_{\text{tags}}}
    $$
    (Where $N_{\text{tags}}$ is the total number of unique tags).

* **Transition Probability ($A$) with Smoothing:**
    $$
    A_{ij} = P(\text{tag\_j} \mid \text{tag\_i}) = \frac{\text{Count(tag\_i} \rightarrow \text{tag\_j}) + 1}{\text{Count(tag\_i)} + N_{\text{tags}}}
    $$
    (Where $N_{\text{tags}}$ is the total number of unique tags).

* **Emission Probability ($O$) with Smoothing:**
    $$
    O_{jk} = P(\text{word\_k} \mid \text{tag\_j}) = \frac{\text{Count}(\text{word\_k}, \text{tag\_j}) + 1}{\text{Count}(\text{tag\_j}) + V}
    $$
    (Where $V$ is the total number of unique words in the vocabulary).

### **Task 4: Viterbi Decoding (15 Marks)**
---

Now that you have your trained HMM parameters (pi, A, O), the final step is to use them to predict the most likely tag sequence for a new sentence. This process is known as **decoding**.


#### **Decoding Approaches**

**1. Brute-Force Decoding**
The most obvious approach is to generate every possible tag sequence for the sentence, calculate the probability of each one, and pick the sequence with the highest score.

* **Problem**: For a sentence with **T** words and **N** possible tags, there are **N<sup>T</sup>** possible sequences.
* **Time Complexity**: This method has an **exponential** time complexity of **O(N<sup>T</sup>)**, which is computationally infeasible.

**2. Viterbi Decoding**
To decode efficiently, we use the **Viterbi algorithm**. It is a **dynamic programming** approach that cleverly finds the single best path without having to check every possibility.

* **Time Complexity**: Viterbi has a **polynomial** time complexity of **O(T \* N<sup>2</sup>)**, making it dramatically faster and the standard method for HMM decoding.


#### **The Viterbi Algorithm in Detail**

The algorithm works by moving through the sentence one word at a time, building two matrices to keep track of the best path found so far.

![viterbi](https://i.postimg.cc/Dyq4KdqD/Screenshot-2025-09-15-at-12-15-52-AM.png)
(source: https://princeton-nlp.github.io/cos484/lectures/L6_sp25.pdf)
(for the derivation and further intuition, please check the above references.)

**The `M` and `B` Matrices**

* **Probability Matrix (`M`)**: `M[i, j]` stores the probability of the **most probable sequence of tags ending with tag `j` at the `i`-th token**. It essentially keeps track of the "score" of the best path to each tag at each position.
* **Backpointer Matrix (`B`)**: `B[i, j]` is the **tag of the `(i-1)`-th token in the most probable sequence ending with tag `j` at the `i`-th token**. It acts like a trail of breadcrumbs, remembering the path taken to reach each state, which is crucial for reconstruction.

**The Process**

The algorithm proceeds in three main steps:

![algo](https://i.postimg.cc/VkCLm5mH/Screenshot-2025-09-16-at-4-54-46-PM.png)
(source: https://web.stanford.edu/~jurafsky/slp3/A.pdf)

**1. Initialization**
For the first word (at index `i=0`), we calculate the initial probability for each possible tag `j`. This is the product of the start probability and the emission probability of the first word given that tag.
<!-- $$
M[0, j] = \pi_j \times \text{emission_prob}[j, \text{word}_0]
$$ -->
$$
M[0, j] = \pi_j \times \text{O}[j, \text{word}_0]
$$

**2. Recursion**
For each subsequent word `i` and each possible tag `j`, we find the path from the previous step that has the highest probability of leading to it. We consider all possible previous tags `k`. We store the maximum probability in `M` and the "winning" previous tag's index in `B`.
<!-- $$
M[i, j] = \max_{k} (M[i-1, k] \times \text{transition_prob}[k, j]) \times \text{emission_prob}[j, \text{word}_i]
$$ -->
$$
M[i, j] = \max_{k} (M[i-1, k] \times \text{A}[k, j]) \times \text{O}[j, \text{word}_i]
$$

<!-- $$
B[i, j] = \arg\max_{k} (M[i-1, k] \times \text{transition_prob}[k, j]) \times \text{emission_prob}[j, \text{word}_i]
$$ -->

$$
B[i, j] = \arg\max_{k} (M[i-1, k] \times \text{A}[k, j]) \times \text{O}[j, \text{word}_i]
$$

**3. Backtracking**
After filling the matrices, the algorithm finishes by:
* Finding the highest probability in the **last column** of the `M` matrix to identify the final tag of the best sequence.
* Using the **`B` matrix** to trace the path backward from this final tag to the beginning, revealing the complete optimal sequence.

### **Task 6: ROC AUC Curve Evaluation (20 Marks)**

While metrics like F1-score evaluate the quality of the single best tag sequence predicted by the Viterbi algorithm, they don't tell the whole story. They are based on "hard" decisions (this word *is* `B-PER`, this one *is not*). A **Receiver Operating Characteristic (ROC)** curve provides a more nuanced evaluation by showing how well the model can distinguish between classes across all possible decision thresholds.

---

#### **Conceptual Background**

* **What is an ROC Curve?** An ROC curve plots the **True Positive Rate (TPR)** against the **False Positive Rate (FPR)** at various threshold settings.
    * **TPR (Recall):** $TPR = \frac{TP}{TP + FN}$. It answers: "Of all the actual positive samples, how many did we correctly identify?"
    * **FPR:** $FPR = \frac{FP}{FP + TN}$. It answers: "Of all the actual negative samples, how many did we incorrectly identify as positive?"
    * A model that is good at discriminating will have a curve that bows towards the top-left corner, indicating a high TPR for a low FPR. A random model follows the diagonal line (y=x).

* **Area Under the Curve (AUC):** The AUC is the area under the ROC curve. It provides a single number to summarize the model's performance.
    * **AUC = 1.0:** Perfect classifier.
    * **AUC = 0.5:** Random classifier (no discriminative ability).
    * **AUC < 0.5:** Worse than random.



#### **Why is the Forward-Backward Algorithm Needed?**

To plot an ROC curve, we need a continuous **score** or probability for each class, not just a single, final prediction.
* The **Viterbi algorithm** (Task 4) is a "hard-decision" algorithm. It finds the single most likely sequence of tags. It doesn't tell you the probability of a specific tag at a specific position, considering the *entire* sentence.


* The **Forward–Backward algorithm**, in contrast, can compute the **posterior probability**, which is exactly what we need. It calculates the probability of a tag at a specific time step \(t\), given the *entire* observation sequence. This probability, denoted as \(\gamma_t(j)\), is calculated as:

$$
\gamma_t(j)
= P\bigl(\text{tag}_t = j \mid \text{all words in sentence}\bigr)
= \frac{\alpha_t(j)\,\beta_t(j)}{\sum_{i=1}^{N} \alpha_T(i)}
$$

Where \(\alpha_t(j)\) is the forward probability, \(\beta_t(j)\) is the backward probability,  
and the denominator is the total probability of the observed sequence.  

This posterior probability serves as the perfect score to generate the ROC curve. By varying the threshold against this score (e.g., “predict `B-PER` if its posterior probability is > 0.6”), we can compute the TPR and FPR to trace the curve.

````python
import numpy as np
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
import os
from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc, accuracy_score, classification_report, confusion_matrix
import warnings
warnings.filterwarnings('ignore')

### Task 1: Data Pre-processing

def load_file(file_path):
    """
    Parses a CoNLL file.
    Returns a list of sentences, where each sentence is a list of (token, tag) tuples.
    """
    sentences = []
    current_sentence = []

    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:  # Empty line indicates sentence boundary
                if current_sentence:
                    sentences.append(current_sentence)
                    current_sentence = []
            else:
                parts = line.split()
                if len(parts) >= 2:
                    token = parts[0]
                    tag = parts[-1]  # Use last element as tag (handles different formats)
                    current_sentence.append((token, tag))

        # Add the last sentence if file doesn't end with newline
        if current_sentence:
            sentences.append(current_sentence)

    return sentences

def create_mappings(sentences):
    """
    Creates word-to-id and tag-to-id and vice-versa mappings from the data.
    """
    # Count word frequencies
    word_freq = Counter()
    tag_set = set()

    for sentence in sentences:
        for token, tag in sentence:
            word_freq[token] += 1
            tag_set.add(tag)

    # Sort words by frequency (descending)
    sorted_words = sorted(word_freq.items(), key=lambda x: x[1], reverse=True)

    # Create mappings
    word_to_id = {'<UNK>': 0}  # UNK token at index 0
    id_to_word = {0: '<UNK>'}

    # Assign IDs to words by frequency
    for idx, (word, freq) in enumerate(sorted_words, 1):
        word_to_id[word] = idx
        id_to_word[idx] = word

    # Create tag mappings
    tag_list = sorted(list(tag_set))
    tag_to_id = {tag: idx for idx, tag in enumerate(tag_list)}
    id_to_tag = {idx: tag for idx, tag in enumerate(tag_list)}

    return word_to_id, id_to_word, tag_to_id, id_to_tag

def prepare_corpus(sentences, word_to_id, tag_to_id):
    """
    Converts sentences into a list of dictionaries with wordids and tagids
    """
    corpus = []

    for sentence in sentences:
        word_ids = []
        tag_ids = []

        for token, tag in sentence:
            # Map word to ID, use UNK if not found
            word_id = word_to_id.get(token, word_to_id['<UNK>'])
            tag_id = tag_to_id[tag]

            word_ids.append(word_id)
            tag_ids.append(tag_id)

        corpus.append({
            'words': word_ids,
            'tags': tag_ids
        })

    return corpus

### Task 2: HMM Training (without smoothing)

def train_hmm(train_corpus, num_tags, num_words):
    """
    Trains a Hidden Markov Model by computing the initial state vector pi,
    transition probability matrix A, and emission probability matrix O
    from a training corpus.
    """
    # Initialize count arrays
    initial_counts = np.zeros(num_tags)
    transition_counts = np.zeros((num_tags, num_tags))
    emission_counts = np.zeros((num_tags, num_words))

    # Count from training data
    for sentence in train_corpus:
        tags = sentence['tags']
        words = sentence['words']

        # Skip empty sentences
        if not tags:
            continue

        # Initial counts
        initial_counts[tags[0]] += 1

        # Transition counts
        for i in range(len(tags) - 1):
            current_tag = tags[i]
            next_tag = tags[i + 1]
            transition_counts[current_tag, next_tag] += 1

        # Emission counts
        for word, tag in zip(words, tags):
            emission_counts[tag, word] += 1

    # Convert counts to probabilities (with handling for zero counts)
    # Initial probabilities
    initial_total = np.sum(initial_counts)
    initial_prob = initial_counts / initial_total if initial_total > 0 else np.ones(num_tags) / num_tags

    # Transition probabilities
    transition_prob = np.zeros((num_tags, num_tags))
    for i in range(num_tags):
        row_total = np.sum(transition_counts[i])
        if row_total > 0:
            transition_prob[i] = transition_counts[i] / row_total
        else:
            transition_prob[i] = np.ones(num_tags) / num_tags

    # Emission probabilities
    emission_prob = np.zeros((num_tags, num_words))
    for i in range(num_tags):
        row_total = np.sum(emission_counts[i])
        if row_total > 0:
            emission_prob[i] = emission_counts[i] / row_total
        else:
            emission_prob[i] = np.ones(num_words) / num_words

    return initial_prob, transition_prob, emission_prob

### Task 3: HMM Training (with Laplace smoothing)

def train_hmm_laplace(train_corpus, num_tags, num_words):
    """
    Trains a Hidden Markov Model with Laplace smoothing.
    """
    # Initialize count arrays
    initial_counts = np.zeros(num_tags)
    transition_counts = np.zeros((num_tags, num_tags))
    emission_counts = np.zeros((num_tags, num_words))

    # Count from training data
    for sentence in train_corpus:
        tags = sentence['tags']
        words = sentence['words']

        # Skip empty sentences
        if not tags:
            continue

        # Initial counts
        initial_counts[tags[0]] += 1

        # Transition counts
        for i in range(len(tags) - 1):
            current_tag = tags[i]
            next_tag = tags[i + 1]
            transition_counts[current_tag, next_tag] += 1

        # Emission counts
        for word, tag in zip(words, tags):
            emission_counts[tag, word] += 1

    # Apply Laplace smoothing: add 1 to all counts
    initial_counts_smoothed = initial_counts + 1
    transition_counts_smoothed = transition_counts + 1
    emission_counts_smoothed = emission_counts + 1

    # Convert smoothed counts to probabilities
    # Initial probabilities
    initial_total = np.sum(initial_counts_smoothed)
    initial_prob = initial_counts_smoothed / initial_total

    # Transition probabilities
    transition_prob = np.zeros((num_tags, num_tags))
    for i in range(num_tags):
        row_total = np.sum(transition_counts_smoothed[i])
        transition_prob[i] = transition_counts_smoothed[i] / row_total

    # Emission probabilities
    emission_prob = np.zeros((num_tags, num_words))
    for i in range(num_tags):
        row_total = np.sum(emission_counts_smoothed[i])
        emission_prob[i] = emission_counts_smoothed[i] / row_total

    return initial_prob, transition_prob, emission_prob

### Task 4: Viterbi Decoding

def viterbi_decode(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags):
    """
    Decodes the most likely sequence of tags for a sentence using the Viterbi algorithm.
    """
    T = len(sentence_word_ids)  # Sequence length
    if T == 0:
        return []

    # Initialize matrices
    M = np.zeros((T, num_tags))  # Probability matrix
    B = np.zeros((T, num_tags), dtype=int)  # Backpointer matrix

    # Step 1: Initialization
    for j in range(num_tags):
        M[0, j] = initial_prob[j] * emission_prob[j, sentence_word_ids[0]]
        B[0, j] = -1  # No previous state for first word

    # Step 2: Recursion
    for t in range(1, T):
        for j in range(num_tags):
            # Find the best previous state
            best_prob = -1
            best_state = -1

            for k in range(num_tags):
                prob = M[t-1, k] * transition_prob[k, j] * emission_prob[j, sentence_word_ids[t]]
                if prob > best_prob:
                    best_prob = prob
                    best_state = k

            M[t, j] = best_prob
            B[t, j] = best_state

    # Step 3: Backtracking
    # Find the best final state
    best_final_prob = -1
    best_final_state = -1
    for j in range(num_tags):
        if M[T-1, j] > best_final_prob:
            best_final_prob = M[T-1, j]
            best_final_state = j

    # Backtrack to get the sequence
    tag_sequence = [best_final_state]
    current_state = best_final_state

    for t in range(T-1, 0, -1):
        current_state = B[t, current_state]
        tag_sequence.insert(0, current_state)

    return tag_sequence

### Task 5: Model Evaluation

def get_confusion_matrix(y_true, y_pred, labels):
    """
    Compute confusion matrix from true and predicted labels.
    """
    num_labels = len(labels)
    cm = np.zeros((num_labels, num_labels), dtype=int)

    for true, pred in zip(y_true, y_pred):
        cm[true, pred] += 1

    return cm

def get_classification_report(y_true, y_pred, id_to_tag):
    """
    Compute classification report from true and predicted labels.
    """
    labels = sorted(id_to_tag.keys())
    num_labels = len(labels)
    cm = get_confusion_matrix(y_true, y_pred, labels)

    # Initialize results
    results = {
        'precision': np.zeros(num_labels),
        'recall': np.zeros(num_labels),
        'f1': np.zeros(num_labels),
        'support': np.zeros(num_labels, dtype=int)
    }

    # Calculate per-class metrics
    for i, label in enumerate(labels):
        tp = cm[i, i]
        fp = np.sum(cm[:, i]) - tp
        fn = np.sum(cm[i, :]) - tp
        tn = np.sum(cm) - tp - fp - fn

        results['support'][i] = np.sum(cm[i, :])

        # Precision
        if tp + fp > 0:
            results['precision'][i] = tp / (tp + fp)
        else:
            results['precision'][i] = 0.0

        # Recall
        if tp + fn > 0:
            results['recall'][i] = tp / (tp + fn)
        else:
            results['recall'][i] = 0.0

        # F1-score
        if results['precision'][i] + results['recall'][i] > 0:
            results['f1'][i] = 2 * (results['precision'][i] * results['recall'][i]) / (results['precision'][i] + results['recall'][i])
        else:
            results['f1'][i] = 0.0

    # Calculate accuracy
    accuracy = np.sum(np.diag(cm)) / np.sum(cm) if np.sum(cm) > 0 else 0

    # Calculate macro, micro, and weighted averages
    macro_precision = np.mean(results['precision'])
    macro_recall = np.mean(results['recall'])
    macro_f1 = np.mean(results['f1'])

    # Micro averages
    total_tp = np.sum(np.diag(cm))
    total_fp = np.sum(cm) - total_tp
    total_fn = total_fp  # In multi-class single-label, FP = FN for micro average

    micro_precision = total_tp / (total_tp + total_fp) if (total_tp + total_fp) > 0 else 0
    micro_recall = total_tp / (total_tp + total_fn) if (total_tp + total_fn) > 0 else 0
    micro_f1 = 2 * micro_precision * micro_recall / (micro_precision + micro_recall) if (micro_precision + micro_recall) > 0 else 0

    # Weighted averages
    weighted_precision = np.average(results['precision'], weights=results['support'])
    weighted_recall = np.average(results['recall'], weights=results['support'])
    weighted_f1 = np.average(results['f1'], weights=results['support'])

    # Format report
    report = f"{'':<15} {'precision':<10} {'recall':<10} {'f1-score':<10} {'support':<10}\n"
    report += "-" * 55 + "\n"

    for i, label in enumerate(labels):
        tag_name = id_to_tag[label]
        report += f"{tag_name:<15} {results['precision'][i]:<10.4f} {results['recall'][i]:<10.4f} {results['f1'][i]:<10.4f} {results['support'][i]:<10}\n"

    report += "-" * 55 + "\n"
    report += f"{'accuracy':<15} {'':<10} {'':<10} {accuracy:<10.4f} {len(y_true):<10}\n"
    report += f"{'macro avg':<15} {macro_precision:<10.4f} {macro_recall:<10.4f} {macro_f1:<10.4f} {np.sum(results['support']):<10}\n"
    report += f"{'weighted avg':<15} {weighted_precision:<10.4f} {weighted_recall:<10.4f} {weighted_f1:<10.4f} {np.sum(results['support']):<10}\n"
    report += f"{'micro avg':<15} {micro_precision:<10.4f} {micro_recall:<10.4f} {micro_f1:<10.4f} {np.sum(results['support']):<10}\n"

    return report, cm

def evaluate_model(corpus, initial_prob, transition_prob, emission_prob, num_tags, id_to_tag, dataset_name):
    """
    Evaluate HMM model on a corpus.
    """
    y_true = []
    y_pred = []

    print(f"\nEvaluating on {dataset_name} set...")

    for sentence in corpus:
        word_ids = sentence['words']
        true_tags = sentence['tags']

        # Skip empty sentences
        if not word_ids:
            continue

        predicted_tags = viterbi_decode(word_ids, initial_prob, transition_prob, emission_prob, num_tags)

        y_true.extend(true_tags)
        y_pred.extend(predicted_tags)

    # Generate classification report
    report, cm = get_classification_report(y_true, y_pred, id_to_tag)
    print(report)

    # Plot confusion matrix
    plt.figure(figsize=(12, 10))
    tag_names = [id_to_tag[i] for i in range(num_tags)]

    # Normalize confusion matrix for better visualization
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
    cm_normalized = np.nan_to_num(cm_normalized)

    sns.heatmap(cm_normalized, annot=True, fmt='.2f', cmap='Blues',
                xticklabels=tag_names, yticklabels=tag_names)
    plt.title(f'Confusion Matrix - {dataset_name} Set (Normalized)')
    plt.xlabel('Predicted Labels')
    plt.ylabel('True Labels')
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()

    return y_true, y_pred

### Task 6: ROC AUC Curve Evaluation

def _forward_pass(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags):
    """
    Perform forward pass to compute alpha probabilities.
    """
    T = len(sentence_word_ids)
    alpha = np.zeros((T, num_tags))
    
    # Initialization
    for j in range(num_tags):
        alpha[0, j] = initial_prob[j] * emission_prob[j, sentence_word_ids[0]]
    
    # Recursion
    for t in range(1, T):
        for j in range(num_tags):
            alpha[t, j] = 0
            for i in range(num_tags):
                alpha[t, j] += alpha[t-1, i] * transition_prob[i, j] * emission_prob[j, sentence_word_ids[t]]
    
    return alpha

def _backward_pass(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags):
    """
    Perform backward pass to compute beta probabilities.
    """
    T = len(sentence_word_ids)
    beta = np.zeros((T, num_tags))
    
    # Initialization
    beta[T-1, :] = 1.0
    
    # Recursion backwards
    for t in range(T-2, -1, -1):
        for i in range(num_tags):
            beta[t, i] = 0
            for j in range(num_tags):
                beta[t, i] += beta[t+1, j] * transition_prob[i, j] * emission_prob[j, sentence_word_ids[t+1]]
    
    return beta

def get_posterior_scores(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags):
    """
    Compute posterior probabilities using forward-backward algorithm.
    """
    alpha = _forward_pass(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags)
    beta = _backward_pass(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags)
    
    T = len(sentence_word_ids)
    gamma = np.zeros((T, num_tags))
    
    # Compute posterior probabilities
    for t in range(T):
        denominator = np.sum(alpha[t, :] * beta[t, :])
        if denominator > 0:
            gamma[t, :] = (alpha[t, :] * beta[t, :]) / denominator
        else:
            gamma[t, :] = np.ones(num_tags) / num_tags
    
    return gamma

def roc_curve_from_scratch(y_true_binary, y_scores):
    """
    Compute ROC curve from scratch.
    """
    # Sort by scores descending
    indices = np.argsort(y_scores)[::-1]
    y_true_binary = y_true_binary[indices]
    y_scores = y_scores[indices]
    
    # Initialize
    tpr = [0]
    fpr = [0]
    thresholds = [y_scores[0] + 1]  # Start with threshold above max score
    
    # Current counts
    tp = fp = 0
    fn = np.sum(y_true_binary)
    tn = len(y_true_binary) - fn
    
    # Iterate through sorted scores
    for i in range(len(y_scores)):
        if i > 0 and y_scores[i] != y_scores[i-1]:
            tpr.append(tp / (tp + fn) if (tp + fn) > 0 else 0)
            fpr.append(fp / (fp + tn) if (fp + tn) > 0 else 0)
            thresholds.append(y_scores[i-1])
        
        if y_true_binary[i] == 1:
            tp += 1
            fn -= 1
        else:
            fp += 1
            tn -= 1
    
    # Add final point
    tpr.append(1.0)
    fpr.append(1.0)
    thresholds.append(y_scores[-1] - 1e-5)
    
    return np.array(fpr), np.array(tpr), np.array(thresholds)

def auc_from_scratch(fpr, tpr):
    """
    Compute AUC from scratch using trapezoidal rule.
    """
    # Sort by FPR
    indices = np.argsort(fpr)
    fpr_sorted = fpr[indices]
    tpr_sorted = tpr[indices]
    
    # Compute AUC using trapezoidal rule
    auc_score = 0.0
    for i in range(1, len(fpr_sorted)):
        auc_score += (fpr_sorted[i] - fpr_sorted[i-1]) * (tpr_sorted[i] + tpr_sorted[i-1]) / 2.0
    
    return auc_score

def plot_one_vs_rest_roc(y_true, y_scores, id_to_tag_map):
    """
    Computes and plots One-vs-Rest ROC curves for each class.
    """
    num_tags = len(id_to_tag_map)
    
    # Binarize the labels
    y_true_binarized = label_binarize(y_true, classes=range(num_tags))
    
    # Compute ROC curve and ROC area for each class
    fpr = dict()
    tpr = dict()
    roc_auc = dict()
    
    for i in range(num_tags):
        fpr[i], tpr[i], _ = roc_curve_from_scratch(y_true_binarized[:, i], y_scores[:, i])
        roc_auc[i] = auc_from_scratch(fpr[i], tpr[i])
    
    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    colors = plt.cm.rainbow(np.linspace(0, 1, num_tags))
    
    for i, color in zip(range(num_tags), colors):
        plt.plot(fpr[i], tpr[i], color=color, lw=2,
                 label='{0} (AUC = {1:0.2f})'.format(id_to_tag_map[i], roc_auc[i]))
    
    plt.plot([0, 1], [0, 1], 'k--', lw=2)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('One-vs-Rest ROC Curves')
    plt.legend(loc="lower right")
    plt.grid(True)
    plt.show()
    
    # Print per-tag AUC scores
    print('Per tag AUC scores:')
    per_tag_aucs = {id_to_tag_map[i]: roc_auc[i] for i in range(num_tags)}
    for tag, auc_score in per_tag_aucs.items():
        print(f"{tag}: {auc_score:.4f}")
    
    return per_tag_aucs

def evaluate_and_plot_roc(corpus, initial_prob, transition_prob, emission_prob, num_tags, id_to_tag):
    """
    Main function to evaluate and plot ROC curves.
    """
    y_true = []
    y_scores_list = []
    
    print("Computing posterior scores for ROC analysis...")
    
    for sentence in corpus:
        word_ids = sentence['words']
        true_tags = sentence['tags']
        
        if not word_ids:
            continue
            
        # Get posterior scores
        posterior_scores = get_posterior_scores(word_ids, initial_prob, transition_prob, emission_prob, num_tags)
        
        y_true.extend(true_tags)
        y_scores_list.append(posterior_scores)
    
    # Concatenate all scores
    y_scores = np.vstack(y_scores_list)
    
    # Plot ROC curves
    per_tag_aucs = plot_one_vs_rest_roc(y_true, y_scores, id_to_tag)
    
    return y_true, y_scores, per_tag_aucs

### Task 7: Evaluating Performance Without the 'O' Tag

def evaluate_model_sklearn(corpus, initial_prob, transition_prob, emission_prob, num_tags, id_to_tag, dataset_name, exclude_o_tag=True):
    """
    Evaluate model using sklearn metrics, optionally excluding 'O' tag.
    """
    y_true_all = []
    y_pred_all = []
    
    print(f"\nEvaluating on {dataset_name} set (exclude O: {exclude_o_tag})...")
    
    for sentence in corpus:
        word_ids = sentence['words']
        true_tags = sentence['tags']
        
        if not word_ids:
            continue
            
        predicted_tags = viterbi_decode(word_ids, initial_prob, transition_prob, emission_prob, num_tags)
        
        y_true_all.extend(true_tags)
        y_pred_all.extend(predicted_tags)
    
    # Convert to numpy arrays
    y_true_all = np.array(y_true_all)
    y_pred_all = np.array(y_pred_all)
    
    if exclude_o_tag:
        # Find 'O' tag ID
        o_tag_id = None
        for tag_id, tag_name in id_to_tag.items():
            if tag_name == 'O':
                o_tag_id = tag_id
                break
        
        if o_tag_id is not None:
            # Filter out 'O' tags
            mask = (y_true_all != o_tag_id) & (y_pred_all != o_tag_id)
            y_true = y_true_all[mask]
            y_pred = y_pred_all[mask]
            labels_to_include = [tag_id for tag_id in id_to_tag.keys() if tag_id != o_tag_id]
        else:
            y_true = y_true_all
            y_pred = y_pred_all
            labels_to_include = list(id_to_tag.keys())
    else:
        y_true = y_true_all
        y_pred = y_pred_all
        labels_to_include = list(id_to_tag.keys())
    
    # Calculate metrics
    accuracy = accuracy_score(y_true, y_pred)
    report = classification_report(y_true, y_pred, labels=labels_to_include, 
                                  target_names=[id_to_tag[i] for i in labels_to_include])
    cm = confusion_matrix(y_true, y_pred, labels=labels_to_include)
    
    print(f"Accuracy: {accuracy:.4f}")
    print("\nClassification Report:")
    print(report)
    
    # Plot confusion matrix
    plt.figure(figsize=(12, 10))
    tag_names = [id_to_tag[i] for i in labels_to_include]
    
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
    cm_normalized = np.nan_to_num(cm_normalized)
    
    sns.heatmap(cm_normalized, annot=True, fmt='.2f', cmap='Blues',
                xticklabels=tag_names, yticklabels=tag_names)
    plt.title(f'Confusion Matrix - {dataset_name} (exclude O: {exclude_o_tag})')
    plt.xlabel('Predicted Labels')
    plt.ylabel('True Labels')
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    
    return accuracy, report, cm

# Main execution
def main():
    """
    Main execution function for HMM NER system.
    """
    print("\n" + "="*60)
    print("HINDI NER USING HMM WITH VITERBI DECODING")
    print("="*60)
    
    # Convert datasets to CoNLL format and save
    output_dir = "conll_output"
    os.makedirs(output_dir, exist_ok=True)
    
    def to_conll(dataset, split_name, sep=" "):
        """Convert dataset to CoNLL format"""
        out_file = os.path.join(output_dir, f"{split_name}.conll")
    
        with open(out_file, "w", encoding="utf-8") as f:
            for example in dataset:
                tokens = example["tokens"]
                labels = example["ner_tags"]
    
                for token, label_id in zip(tokens, labels):
                    label_str = LABEL_LIST[label_id]
                    f.write(f"{token}{sep}{label_str}\n")
                f.write("\n")
    
    # Convert all splits
    to_conll(train, "train")
    to_conll(val, "val")
    to_conll(test, "test")
    
    # Load CoNLL files
    print("Loading CoNLL files...")
    train_data = load_file(os.path.join(output_dir, "train.conll"))
    val_data = load_file(os.path.join(output_dir, "val.conll"))
    test_data = load_file(os.path.join(output_dir, "test.conll"))
    
    print(f"Training sentences: {len(train_data)}")
    print(f"Validation sentences: {len(val_data)}")
    print(f"Test sentences: {len(test_data)}")
    
    # Create mappings
    print("\nCreating mappings...")
    word_to_id, id_to_word, tag_to_id, id_to_tag = create_mappings(train_data)
    
    num_words = len(word_to_id)
    num_tags = len(tag_to_id)
    
    print(f"Vocabulary size: {num_words}")
    print(f"Number of tags: {num_tags}")
    print(f"Tags: {list(tag_to_id.keys())}")
    
    # Prepare corpus
    print("\nPreparing corpus...")
    train_corpus = prepare_corpus(train_data, word_to_id, tag_to_id)
    val_corpus = prepare_corpus(val_data, word_to_id, tag_to_id)
    test_corpus = prepare_corpus(test_data, word_to_id, tag_to_id)
    
    # Train HMM without smoothing
    print("\nTraining HMM without smoothing...")
    initial_prob, transition_prob, emission_prob = train_hmm(train_corpus, num_tags, num_words)
    
    # Train HMM with Laplace smoothing
    print("Training HMM with Laplace smoothing...")
    initial_prob_laplace, transition_prob_laplace, emission_prob_laplace = train_hmm_laplace(train_corpus, num_tags, num_words)
    
    # Evaluate both models
    print("\n" + "="*40)
    print("EVALUATING HMM WITHOUT SMOOTHING")
    print("="*40)
    
    # Train set evaluation
    y_true_train, y_pred_train = evaluate_model(train_corpus, initial_prob, transition_prob, emission_prob, num_tags, id_to_tag, "Train")
    
    # Validation set evaluation
    y_true_val, y_pred_val = evaluate_model(val_corpus, initial_prob, transition_prob, emission_prob, num_tags, id_to_tag, "Validation")
    
    # Test set evaluation
    y_true_test, y_pred_test = evaluate_model(test_corpus, initial_prob, transition_prob, emission_prob, num_tags, id_to_tag, "Test")
    
    print("\n" + "="*40)
    print("EVALUATING HMM WITH LAPLACE SMOOTHING")
    print("="*40)
    
    # Train set evaluation
    y_true_train_laplace, y_pred_train_laplace = evaluate_model(train_corpus, initial_prob_laplace, transition_prob_laplace, emission_prob_laplace, num_tags, id_to_tag, "Train")
    
    # Validation set evaluation
    y_true_val_laplace, y_pred_val_laplace = evaluate_model(val_corpus, initial_prob_laplace, transition_prob_laplace, emission_prob_laplace, num_tags, id_to_tag, "Validation")
    
    # Test set evaluation
    y_true_test_laplace, y_pred_test_laplace = evaluate_model(test_corpus, initial_prob_laplace, transition_prob_laplace, emission_prob_laplace, num_tags, id_to_tag, "Test")
    
    # Task 6: ROC AUC Evaluation (using Laplace smoothed model)
    print("\n" + "="*40)
    print("ROC AUC EVALUATION (WITH LAPLACE SMOOTHING)")
    print("="*40)
    
    y_true_roc, y_scores_roc, per_tag_aucs = evaluate_and_plot_roc(
        val_corpus, initial_prob_laplace, transition_prob_laplace, emission_prob_laplace, num_tags, id_to_tag
    )
    
    # Task 7: Evaluation without 'O' tag
    print("\n" + "="*40)
    print("EVALUATION WITHOUT 'O' TAG")
    print("="*40)
    
    # With 'O' tag
    print("\n--- WITH 'O' TAG ---")
    accuracy_with_o, report_with_o, cm_with_o = evaluate_model_sklearn(
        val_corpus, initial_prob_laplace, transition_prob_laplace, emission_prob_laplace, 
        num_tags, id_to_tag, "Validation", exclude_o_tag=False
    )
    
    # Without 'O' tag
    print("\n--- WITHOUT 'O' TAG ---")
    accuracy_without_o, report_without_o, cm_without_o = evaluate_model_sklearn(
        val_corpus, initial_prob_laplace, transition_prob_laplace, emission_prob_laplace,
        num_tags, id_to_tag, "Validation", exclude_o_tag=True
    )
    
    print("\n" + "="*60)
    print("HMM EVALUATION COMPLETE")
    print("="*60)
    
    # Return all important objects for analysis
    return {
        'word_to_id': word_to_id,
        'id_to_word': id_to_word,
        'tag_to_id': tag_to_id,
        'id_to_tag': id_to_tag,
        'initial_prob': initial_prob,
        'transition_prob': transition_prob,
        'emission_prob': emission_prob,
        'initial_prob_laplace': initial_prob_laplace,
        'transition_prob_laplace': transition_prob_laplace,
        'emission_prob_laplace': emission_prob_laplace,
        'per_tag_aucs': per_tag_aucs
    }

# Run the main function
if __name__ == "__main__":
    results = main()
````

## Introduction to the Expectation-Maximization (EM) Algorithm

The **Expectation-Maximization (EM)** algorithm is a general framework for finding maximum likelihood estimates of parameters in probabilistic models when some variables are unobserved or "hidden".It is especially useful when direct computation of the likelihood function is difficult because the data includes latent variables.

**EM works in two main iterative steps:**

- **E-step (Expectation):** Compute an expectation of the log-likelihood with respect to the current estimate of the distribution for the latent variables.
- **M-step (Maximization):** Maximize this expected log-likelihood to update the parameters.

This process is repeated until convergence. EM guarantees that the likelihood will never decrease with each iteration.

---

## Baum-Welch Algorithm: EM for HMMs

The **Baum-Welch algorithm** is a specific case of the EM algorithm for training **Hidden Markov Models (HMMs)** on observation sequences when the “true” hidden states are unknown (on problem statements that are of unsupervised/ semi-supervised settings).

**Goal:**  
Given observed data sequences(words/events), estimate the optimal HMM parameters [Initial Distribution $(\pi)$, Transition Matrix $(A)$, and Emission Matrix $(O)$ ] to maximize the likelihood of the observed data.

**Why Baum-Welch?**  
In many practical cases (like speech, text, sequence tagging), we only observe the sequence of emitted symbols (e.g., words), not the underlying sequence of states (e.g., POS tags, NER tags). We want the HMM parameters that best explain the observed data.


---

## The Baum-Welch Algorithm: Step-by-Step

### 1. Initialization

Start with an initial guess for all HMM parameters:

- Initial state probabilities $(\pi)$
- Transition probabilities $(A)$
- Emission probabilities $(O)$

These can be random, uniform, or based on some prior knowledge.

### 2. E-step: Expected Sufficient Statistics

Given the current model parameters, run **Forward-Backward** algorithm over all training sequences:

- Forward: $\alpha_t(i)$ — Compute the probability of observing the  partial sequence up to time $t$, ending in state $i$.  
- Backward: $\beta_t(i)$ — Compute the probability of remaining sequence started from state $i$ at time $t$.

From these, compute:

- $\gamma_t(i)$ — expected probability of being in state $i$ at time $t$ given the observed sequence and current parameters.

- $\xi_t(i,j)$ — expected probability of transitioning from state $i$ at $t$ to state $j$ at $t+1$ given the observed sequence and current parameters.

### 3. M-step: Update Model Parameters

Update each HMM parameter to maximize the expected log-likelihood:

* **Initial probabilities $(\pi)$:**

$$ \pi_i = \gamma_0(i) $$ (average probability of starting in state
$i$)

* **Transition matrix $(A)$:**

$$
A_{i,j} = \frac{\sum_{t=0}^{T-2} \xi_t(i,j)}{\sum_{t=0}^{T-2} \gamma_t(i)}
$$

(expected number of transitions from $i$ to $j$ divided by expected number in $i$)

* **Emisssion Matrix $(O)$**:

$$
O_{j,k} = \frac{\sum_{t: o_t = k} \gamma_t(j)}{\sum_{t=0}^{T-1} \gamma_t(j)}
$$
(expected number of times state $j$ emitted symbol $k$, normalized)

### 4. Iterate Until Convergence

Repeat the E-step and M-step until the model parameters converge (i.e., the change in likelihood is below a small threshold, or a max number of iterations is reached), until log-likelihood stabilizes.

---

## Formal Statement of the Problem

Given:

- $N$: Number of hidden states (tags)

- $M$: Number of observable symbols (words)
- $O = (o_1, o_2, ..., o_T)$: Observed sequence  

Estimate HMM parameters $\lambda = (\pi, A, B)$ to maximize:

$$
P(O | \lambda)
$$

Where:

- $\pi_i$: Prob. of starting in state $i$  
- $A_{i,j}$: Prob. of transitioning $i \rightarrow j$  
- $B_{j,k}$: Prob. of emitting symbol $k$ in state $j$


---

## EM (Baum-Welch) Algorithm Details

### 1. Initialization

Start with random or heuristic values for $\pi, A, B$.

### 2. E-step: Forward-Backward Algorithm

**Forward**

- Base: $\alpha_1(i) = \pi_i B_{i,o_1}$
- Recursive:  
$$
\alpha_{t+1}(j) = \left[ \sum_{i=1}^{N} \alpha_t(i) A_{i,j} \right] B_{j,o_{t+1}}
$$

**Backward**

- Base: $\beta_T(i) = 1$
- Recursive:  
$$
\beta_t(i) = \sum_{j=1}^{N} A_{i,j} B_{j,o_{t+1}} \beta_{t+1}(j)
$$

**Posterior Probabilities**

$$
\gamma_t(i) = \frac{\alpha_t(i) \beta_t(i)}{\sum_{k=1}^{N} \alpha_T(k)}
$$

$$
\xi_t(i,j) = \frac{\alpha_t(i) A_{i,j} B_{j,o_{t+1}} \beta_{t+1}(j)}{\sum_{k=1}^{N} \alpha_T(k)}
$$

### 3. M-step: Re-estimate Parameters

$$
\pi_i^{new} = \gamma_1(i)
$$

$$
A_{i,j}^{new} = \frac{\sum_{t=1}^{T-1} \xi_t(i,j)}{\sum_{t=1}^{T-1} \gamma_t(i)}
$$

$$
B_{j,k}^{new} = \frac{\sum_{t: o_t = k} \gamma_t(j)}{\sum_{t=1}^{T} \gamma_t(j)}
$$

### 4. Convergence

Repeat E/M steps until:

$$
\log P(O | \lambda) = \log \left( \sum_{i=1}^{N} \alpha_T(i) \right)
$$


Key Quantities:

- $\pi_i$: Prob. of starting in state $i$  
- $A_{i,j}$: Prob. of transitioning $i \rightarrow j$  
- $B_{j,k}$: Prob. of emitting symbol $k$ in state $j$

- $\alpha_t(i)$: Prob. of observing initial segment and being in state $i$ at time $t$

- $\beta_t(i)$: Prob. of observing remaining sequence given state $i$ at time $t$

- $\gamma_t(i)$: Posterior prob. of being in state $i$ at time $t$

- $\xi_t(i,j)$: Posterior prob. of transitioning $i \rightarrow j$ at $t \rightarrow t+1$

````python
import numpy as np
from collections import defaultdict, Counter
import logging

# Set up logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

class HMMBaumWelch:
    """
    HMM with Baum-Welch (EM) algorithm for unsupervised training
    """

    def __init__(self, num_states, num_observations, random_seed=42):
        self.num_states = num_states
        self.num_observations = num_observations
        self.random_seed = random_seed
        np.random.seed(random_seed)

        # Initialize parameters
        self.pi = None  # Initial state probabilities
        self.A = None   # Transition matrix
        self.B = None   # Emission matrix

        # Training history
        self.log_likelihoods = []

    def initialize_parameters(self, method='random'):
        """
        Initialize HMM parameters with optimized operations
        """
        if method == 'random':
            # Vectorized initialization
            self.pi = np.random.dirichlet(np.ones(self.num_states))
            
            # Use Dirichlet distribution for proper normalization
            self.A = np.random.dirichlet(np.ones(self.num_states), size=self.num_states)
            self.B = np.random.dirichlet(np.ones(self.num_observations), size=self.num_states)

        elif method == 'uniform':
            # Vectorized uniform initialization
            self.pi = np.full(self.num_states, 1.0 / self.num_states)
            self.A = np.full((self.num_states, self.num_states), 1.0 / self.num_states)
            self.B = np.full((self.num_states, self.num_observations), 1.0 / self.num_observations)

        else:
            raise ValueError("Method must be 'random' or 'uniform'")

        logger.info(f"Initialized HMM with {self.num_states} states and {self.num_observations} observations")

    def forward_algorithm(self, observations):
        """
        Optimized forward algorithm with vectorization
        """
        T = len(observations)
        alpha = np.zeros((T, self.num_states))
        scales = np.zeros(T)

        # Vectorized initialization
        alpha[0] = self.pi * self.B[:, observations[0]]
        scales[0] = 1.0 / np.sum(alpha[0])
        alpha[0] *= scales[0]

        # Vectorized recursion using broadcasting
        for t in range(1, T):
            # Matrix multiplication equivalent: alpha[t-1] @ A * B[:, observations[t]]
            alpha[t] = np.dot(alpha[t-1], self.A) * self.B[:, observations[t]]
            
            # Scale and store scale factor
            scale_t = np.sum(alpha[t])
            if scale_t > 0:
                scales[t] = 1.0 / scale_t
                alpha[t] *= scales[t]
            else:
                scales[t] = 1.0
                alpha[t] = np.ones(self.num_states) / self.num_states

        log_likelihood = -np.sum(np.log(scales[scales > 0]))  # Avoid log(0)

        return alpha, scales, log_likelihood

    def backward_algorithm(self, observations, scales):
        """
        Optimized backward algorithm with vectorization
        """
        T = len(observations)
        beta = np.zeros((T, self.num_states))

        # Vectorized initialization
        beta[T-1] = scales[T-1]

        # Vectorized recursion
        for t in range(T-2, -1, -1):
            # Vectorized computation: sum over j of A[i,j] * B[j,obs[t+1]] * beta[t+1,j]
            beta[t] = np.dot(self.A, self.B[:, observations[t+1]] * beta[t+1])
            beta[t] *= scales[t]

        return beta

    def compute_posteriors(self, observations, alpha, beta):
        """
        Vectorized computation of gamma and xi
        """
        T = len(observations)
        gamma = np.zeros((T, self.num_states))
        xi = np.zeros((T-1, self.num_states, self.num_states))

        # Vectorized gamma computation
        gamma = alpha * beta
        row_sums = np.sum(gamma, axis=1, keepdims=True)
        gamma = np.divide(gamma, row_sums, where=row_sums > 0)

        # Vectorized xi computation using broadcasting
        if T > 1:
            # Precompute common term for efficiency
            B_obs = self.B[:, observations[1:T]]
            beta_shifted = beta[1:T]
            
            # Use einsum for efficient tensor operations
            xi = np.einsum('ti,ij,jt->tij', alpha[:-1], self.A, B_obs * beta_shifted.T)
            
            # Normalize xi
            xi_sums = np.sum(xi, axis=(1, 2), keepdims=True)
            xi = np.divide(xi, xi_sums, where=xi_sums > 0)

        return gamma, xi

    def e_step(self, observations):
        """
        Optimized expectation step
        """
        alpha, scales, log_likelihood = self.forward_algorithm(observations)
        beta = self.backward_algorithm(observations, scales)
        gamma, xi = self.compute_posteriors(observations, alpha, beta)

        return gamma, xi, log_likelihood

    def m_step(self, all_observations, all_gamma, all_xi):
        """
        Vectorized maximization step
        """
        num_sequences = len(all_observations)
        epsilon = 1e-10

        # Initialize accumulators
        pi_numerator = np.zeros(self.num_states)
        A_numerator = np.zeros((self.num_states, self.num_states))
        A_denominator = np.zeros(self.num_states)
        B_numerator = np.zeros((self.num_states, self.num_observations))
        B_denominator = np.zeros(self.num_states)

        # Process all sequences efficiently
        for seq_idx in range(num_sequences):
            observations = all_observations[seq_idx]
            gamma = all_gamma[seq_idx]
            xi = all_xi[seq_idx]
            T = len(observations)

            # Accumulate initial probabilities
            pi_numerator += gamma[0]

            # Vectorized transition probability accumulation
            if T > 1:
                A_numerator += np.sum(xi, axis=0)
                A_denominator += np.sum(gamma[:-1], axis=0)

            # Vectorized emission probability accumulation using advanced indexing
            np.add.at(B_numerator, (slice(None), observations), gamma.T)
            B_denominator += np.sum(gamma, axis=0)

        # Update parameters with vectorized operations
        self.pi = (pi_numerator + epsilon) / (np.sum(pi_numerator) + self.num_states * epsilon)

        # Update transition matrix with broadcasting
        A_denom = A_denominator[:, np.newaxis] + self.num_states * epsilon
        self.A = (A_numerator + epsilon) / np.where(A_denom > 0, A_denom, 1.0)

        # Update emission matrix with broadcasting
        B_denom = B_denominator[:, np.newaxis] + self.num_observations * epsilon
        self.B = (B_numerator + epsilon) / np.where(B_denom > 0, B_denom, 1.0)

        # Ensure proper normalization
        self.pi /= np.sum(self.pi)
        self.A /= np.sum(self.A, axis=1, keepdims=True)
        self.B /= np.sum(self.B, axis=1, keepdims=True)

    def baum_welch(self, all_observations, max_iter=50, tol=1e-6, verbose=True):
        """
        Optimized Baum-Welch algorithm with early stopping
        """
        self.initialize_parameters(method='random')

        prev_log_likelihood = -float('inf')
        best_params = (self.pi.copy(), self.A.copy(), self.B.copy())
        best_likelihood = -float('inf')

        for iteration in range(max_iter):
            # E-step: Process sequences in batch
            all_gamma, all_xi = [], []
            total_log_likelihood = 0.0

            for observations in all_observations:
                if len(observations) == 0:
                    continue
                gamma, xi, log_likelihood = self.e_step(observations)
                all_gamma.append(gamma)
                all_xi.append(xi)
                total_log_likelihood += log_likelihood

            # M-step: Update parameters
            self.m_step(all_observations, all_gamma, all_xi)

            # Store and track best likelihood
            self.log_likelihoods.append(total_log_likelihood)
            
            if total_log_likelihood > best_likelihood:
                best_likelihood = total_log_likelihood
                best_params = (self.pi.copy(), self.A.copy(), self.B.copy())

            # Check convergence
            likelihood_change = total_log_likelihood - prev_log_likelihood

            if verbose and iteration % 5 == 0:
                logger.info(f"Iteration {iteration}: Log Likelihood = {total_log_likelihood:.4f}, "
                          f"Change = {likelihood_change:.6f}")

            if abs(likelihood_change) < tol and iteration > 5:
                logger.info(f"Converged after {iteration} iterations")
                # Restore best parameters
                self.pi, self.A, self.B = best_params
                break

            prev_log_likelihood = total_log_likelihood

        logger.info(f"Training completed after {iteration + 1} iterations")
        return self.pi, self.A, self.B

    def viterbi_decode(self, observations):
        """
        Optimized Viterbi algorithm with log probabilities
        """
        T = len(observations)
        if T == 0:
            return np.array([], dtype=int)

        # Use log probabilities to avoid underflow
        log_pi = np.log(self.pi + 1e-20)
        log_A = np.log(self.A + 1e-20)
        log_B = np.log(self.B + 1e-20)

        delta = np.zeros((T, self.num_states))
        psi = np.zeros((T, self.num_states), dtype=int)

        # Initialization with log probabilities
        delta[0] = log_pi + log_B[:, observations[0]]

        # Vectorized recursion
        for t in range(1, T):
            # Matrix operation: delta[t-1] + log_A + log_B[:, observations[t]]
            temp = delta[t-1, :, np.newaxis] + log_A + log_B[:, observations[t]]
            delta[t] = np.max(temp, axis=0)
            psi[t] = np.argmax(temp, axis=0)

        # Backtracking
        states = np.zeros(T, dtype=int)
        states[T-1] = np.argmax(delta[T-1])

        for t in range(T-2, -1, -1):
            states[t] = psi[t+1, states[t+1]]

        return states

# Optimized helper functions
def prepare_unsupervised_data(sentences, word_to_id, max_seq_length=100):
    """
    Optimized data preparation with list comprehension
    """
    sequences = []
    unk_id = word_to_id.get('<UNK>', 0)
    
    for sentence in sentences:
        if len(sentence) > max_seq_length:
            continue
        # Use list comprehension with precomputed mapping
        word_ids = [word_to_id.get(word, unk_id) for word in sentence]
        sequences.append(word_ids)

    logger.info(f"Prepared {len(sequences)} sequences for unsupervised training")
    return sequences

def initialize_from_supervised(train_corpus, num_tags, num_words):
    """
    Optimized initialization from supervised training
    """
    initial_prob, transition_prob, emission_prob = train_hmm(train_corpus, num_tags, num_words)

    hmm_bw = HMMBaumWelch(num_tags, num_words)
    hmm_bw.pi = initial_prob
    hmm_bw.A = transition_prob
    hmm_bw.B = emission_prob

    return hmm_bw

# Optimized evaluation functions
def evaluate_baum_welch(hmm_bw, test_corpus, id_to_tag, dataset_name="Test"):
    """
    Optimized evaluation with batch processing
    """
    y_true = []
    y_pred = []

    logger.info(f"Evaluating Baum-Welch HMM on {dataset_name} set...")

    for sentence in test_corpus:
        word_ids = sentence['words']
        true_tags = sentence['tags']

        if not word_ids:
            continue

        predicted_tags = hmm_bw.viterbi_decode(word_ids)
        y_true.extend(true_tags)
        y_pred.extend(predicted_tags)

    report, cm = get_classification_report(y_true, y_pred, id_to_tag)
    print(f"\nBaum-Welch Results - {dataset_name} Set:")
    print(report)

    return y_true, y_pred

def plot_likelihood_convergence(log_likelihoods):
    """
    Plot convergence with error handling
    """
    try:
        import matplotlib.pyplot as plt
        
        plt.figure(figsize=(10, 6))
        plt.plot(range(1, len(log_likelihoods) + 1), log_likelihoods, 'b-', linewidth=2)
        plt.xlabel('Iteration')
        plt.ylabel('Log Likelihood')
        plt.title('Baum-Welch Training: Log Likelihood Convergence')
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
    except ImportError:
        print("Matplotlib not available, skipping convergence plot")
    except Exception as e:
        print(f"Error plotting convergence: {e}")

# Main execution for Baum-Welch - OPTIMIZED
print("\n" + "="*60)
print("BAUM-WELCH UNSUPERVISED HMM TRAINING")
print("="*60)

# Prepare data efficiently
print("Preparing data for unsupervised training...")

# Use generator expressions for memory efficiency
train_sequences = prepare_unsupervised_data(
    (example["tokens"] for example in train),
    word_to_id
)

val_sequences = prepare_unsupervised_data(
    (example["tokens"] for example in val),
    word_to_id
)

# Filter out empty sequences
train_sequences = [seq for seq in train_sequences if len(seq) > 0]
val_sequences = [seq for seq in val_sequences if len(seq) > 0]

print(f"Training with {len(train_sequences)} non-empty sequences")

# Initialize and train with optimized parameters
print("Training HMM with Baum-Welch algorithm...")

hmm_bw = HMMBaumWelch(
    num_states=num_tags,
    num_observations=num_words
)

# Train using optimized Baum-Welch
pi_bw, A_bw, B_bw = hmm_bw.baum_welch(
    train_sequences,
    max_iter=60,
    tol=1e-6,
    verbose=True
)

# Plot convergence
plot_likelihood_convergence(hmm_bw.log_likelihoods)

# Evaluate optimized model
print("\n" + "="*40)
print("EVALUATING BAUM-WELCH TRAINED HMM")
print("="*40)

# Test set evaluation
y_true_bw_test, y_pred_bw_test = evaluate_baum_welch(
    hmm_bw, test_corpus, id_to_tag, "Test"
)

# Validation set evaluation
y_true_bw_val, y_pred_bw_val = evaluate_baum_welch(
    hmm_bw, val_corpus, id_to_tag, "Validation"
)

# Compare with supervised HMM
print("\n" + "="*50)
print("COMPARISON: SUPERVISED vs UNSUPERVISED (BAUM-WELCH)")
print("="*50)

# Semi-supervised approach with optimized initialization
print("\n" + "="*50)
print("SEMI-SUPERVISED APPROACH")
print("="*50)

print("Initializing with supervised parameters, then refining with Baum-Welch...")

hmm_semi = initialize_from_supervised(train_corpus, num_tags, num_words)

# Refine with optimized Baum-Welch
pi_semi, A_semi, B_semi = hmm_semi.baum_welch(
    train_sequences,
    max_iter=60,
    tol=1e-6,
    verbose=True
)

# Evaluate semi-supervised model
y_true_semi_test, y_pred_semi_test = evaluate_baum_welch(
    hmm_semi, test_corpus, id_to_tag, "Semi-supervised Test"
)

print("\n" + "="*60)
print("BAUM-WELCH TRAINING COMPLETE")
print("="*60)

# Optimized parameter analysis
def analyze_learned_parameters(hmm_model, id_to_tag, id_to_word, top_n=10):
    """
    Optimized parameter analysis with vectorized operations
    """
    print("\n" + "="*40)
    print("ANALYSIS OF LEARNED PARAMETERS - OPTIMIZED")
    print("="*40)

    # Vectorized analysis of initial probabilities
    print("\nTop initial states:")
    top_pi_indices = np.argsort(hmm_model.pi)[::-1][:top_n]
    for idx in top_pi_indices:
        tag_name = id_to_tag.get(idx, f"TAG_{idx}")
        prob = hmm_model.pi[idx]
        print(f"  {tag_name}: {prob:.4f}")

    # Vectorized analysis of transition probabilities
    print(f"\nStrongest transitions (top {top_n}):")
    # Flatten and get top indices
    flat_A = hmm_model.A.flatten()
    top_transitions = np.argsort(flat_A)[::-1][:top_n]
    
    for flat_idx in top_transitions:
        i = flat_idx // hmm_model.num_states
        j = flat_idx % hmm_model.num_states
        from_tag = id_to_tag.get(i, f"TAG_{i}")
        to_tag = id_to_tag.get(j, f"TAG_{j}")
        prob = hmm_model.A[i, j]
        print(f"  {from_tag} -> {to_tag}: {prob:.4f}")

    # Optimized emission probability analysis
    important_tags = ['B-PERSON', 'B-LOCATION', 'B-ORGANIZATION', 'O']
    tag_indices = {tag: idx for idx, tag in id_to_tag.items() if tag in important_tags}

    for tag_name, tag_idx in tag_indices.items():
        print(f"\nTop words for {tag_name}:")
        top_word_indices = np.argsort(hmm_model.B[tag_idx])[::-1][:top_n]
        for word_idx in top_word_indices:
            if word_idx in id_to_word and word_idx != 0:
                word = id_to_word[word_idx]
                prob = hmm_model.B[tag_idx, word_idx]
                print(f"  {word}: {prob:.4f}")

# Analyze optimized parameters
analyze_learned_parameters(hmm_bw, id_to_tag, id_to_word)

# Optimized parameter comparison
print("\n" + "="*40)
print("COMPARISON WITH SUPERVISED PARAMETERS")
print("="*40)

def compare_parameters(hmm1, hmm2, id_to_tag, param_name):
    """Optimized parameter comparison"""
    if param_name == 'pi':
        param1, param2 = hmm1.pi, hmm2.pi
    elif param_name == 'A':
        param1, param2 = hmm1.A.flatten(), hmm2.A.flatten()
    elif param_name == 'B':
        param1, param2 = hmm1.B.flatten(), hmm2.B.flatten()

    correlation = np.corrcoef(param1, param2)[0, 1]
    mse = np.mean((param1 - param2) ** 2)
    cosine_sim = np.dot(param1, param2) / (np.linalg.norm(param1) * np.linalg.norm(param2))

    print(f"{param_name} - Correlation: {correlation:.4f}, MSE: {mse:.6f}, Cosine: {cosine_sim:.4f}")

# Create supervised HMM for comparison
hmm_supervised = HMMBaumWelch(num_tags, num_words)
hmm_supervised.pi, hmm_supervised.A, hmm_supervised.B = initial_prob, transition_prob, emission_prob

print("Parameter comparison (Supervised vs Baum-Welch):")
compare_parameters(hmm_supervised, hmm_bw, id_to_tag, 'pi')
compare_parameters(hmm_supervised, hmm_bw, id_to_tag, 'A')
compare_parameters(hmm_supervised, hmm_bw, id_to_tag, 'B')

print("\n" + "="*60)
print("HMM BAUM-WELCH COMPLETED SUCCESSFULLY")
print("="*60)
````

Since Baum-Welch is unsupervised, it often collapses to trivial solutions (e.g., predicting mostly “O”). Accuracy and F1 may remain poor even as log-likelihood improves.

**Can you think of any ideas to get better results?**

**ANSWER:**

Since the Baum-Welch algorithm is unsupervised, it often converges to trivial solutions such as predicting almost all tokens as “O”. This happens because “O” dominates the data, so maximizing likelihood favors that tag even though NER performance (F1) remains poor.

To get better results, several strategies can be used:

1. **Better Initialization:**
   Initialize the HMM parameters using a small amount of labeled data or heuristic rules (e.g., tokens starting with capital letters or matching gazetteer entries) instead of random initialization. This helps the EM process start from a more meaningful state and avoid collapsing to all “O”.

2. **Semi-supervised Learning:**
   Combine Baum-Welch with a small labeled subset to guide the model. The labeled data anchors the entity tags, while the unlabeled portion improves coverage.

3. **Gazetteers and Lexical Cues:**
   Use external name lists (for persons, locations, organizations) to bias emission probabilities for relevant words. This makes it more likely that entity tokens are tagged correctly.

4. **Posterior Constraints or Priors:**
   Add constraints that prevent all tokens from being assigned to “O” (e.g., enforce a minimum expected proportion of entity tags). This discourages the trivial solution.

5. **Early Stopping Based on Validation F1:**
   Since log-likelihood alone can be misleading, monitor F1 on a small labeled validation set and stop EM when performance stops improving.

In short, the idea is to **introduce weak supervision, smart initialization, or constraints** so that Baum-Welch doesn’t exploit the easy “all-O” path but learns meaningful patterns of entity transitions and emissions.

#### 1) Chaitanya starts with HMM (for NER)

Chaitanya wants to build a Named Entity Recognizer (NER).  
First, he tries **Hidden Markov Model (HMM).**

**Good parts of HMM:**
- Can learn **unsupervised** from raw text (using algorithms like Baum-Welch).  
- Handles variable-length sentences.  
- Can decode fast using the **Viterbi algorithm**.

**Problems he notices:**
- HMM is a **generative model**. It assumes tags generate words independently.  
  → This makes it hard to add **overlapping or arbitrary features**.  
- It mostly looks at **current word + previous tag**, not the full context.  
- It models the **joint probability** $P(X, Y)$ (words + tags together).  
  → But in NER, what we really want is $P(Y \mid X)$ (tags given the words).  
- Because of its independence assumptions, it cannot use rich features like:
  - Capitalization (*Delhi* vs *delhi*)  
  - Prefixes/suffixes (*-pur*, *-nagar*)  
  - Gazetteers (lists of names, places, etc.)

**Chaitanya’s thought:**  
*"NER needs context. HMM feels too blind and restricted."*

 Note: In **supervised settings**, HMM parameters can be learned by counting.  
But its real strength is in **unsupervised learning**, unlike MEMM/CRF.

#### 2) Chaitanya tries MEMM (for NER)

After struggling with HMM, Chaitanya thinks:  
*"What if I could use more information about the words, not just fixed probabilities?"*  
That’s when he discovers **Maximum Entropy Markov Model (MEMM).**

**What is MEMM?**
- Imagine HMM is a person who only checks the **previous tag + current word** before guessing the next tag.  
- MEMM is a smarter person who says:  
  *“Why not use more clues? Like capitalization, suffixes, and surrounding words.”*  
- MEMM is a **discriminative model**.  
- It replaces HMM’s fixed probabilities with a **maximum entropy classifier** (like logistic regression).  
- This classifier predicts the next tag using **features of the current word and its context**.

**Why MEMM felt better:**
- With MEMM, Chaitanya could finally use rich signals:
  - If a word starts with a capital → maybe a **PERSON**.  
  - If a word ends with “-pur” → maybe a **LOCATION**.  
  - If the word before is *Mr.* → next word likely a **PERSON**.  
- MEMM directly models $P(Y \mid X)$, not $P(X, Y)$.  
- Training is efficient — each local classifier is trained independently.  

**Chaitanya felt relieved:**  
*"Finally! My NER system can use context and features, not just blind probabilities."*

#### 3) But MEMM has a hidden problem — Label Bias

Chaitanya was happy with MEMM… until he saw a big issue: **label bias.**

**What is Label Bias?**
- Suppose MEMM is tagging: *"Robbin flies to the bank."*  
  - *flies* can be verb or noun.  
  - *bank* can be a riverbank or financial bank.  
- Imagine the tag **VERB** has many possible next tags (high entropy).  
- The tag **NOUN** has fewer next tags (low entropy).  
- In MEMM, probabilities are **normalized locally at each state**.  
  - Fewer transitions → higher chance by default.  
  - More transitions → probability spread out.  
- So MEMM gets biased toward **states with fewer options**, even if features say otherwise.  

**In short:**  
MEMM is greedy at each step. It can’t use **future context** to fix wrong choices.  

**Chaitanya’s frustration:**  
*"Oh no… MEMM uses features, but it still makes biased local decisions!"*

 This problem is well known — called **label bias** in the original CRF paper (Lafferty et al., 2001).

# Conditional Random Fields (CRFs)

## Setup & Notation

We work with an observed sequence and a state/label sequence:

$$
\mathbf{x} = (x_1, \dots, x_m)
$$

$$
\mathbf{s} = (s_1, \dots, s_m), \quad s_j \in \mathcal{S}
$$

$$
\mathcal{S}^m = \text{set of all length-}m\text{ state sequences}
$$

We model the conditional distribution:

$$
p(\mathbf{s}\mid \mathbf{x})
$$


## Global Feature Vector and the Log-Linear CRF

Define a global feature vector:

$$
\Phi(\mathbf{x},\mathbf{s}) \in \mathbb{R}^d
$$

The conditional distribution is:

$$
p(\mathbf{s}\mid \mathbf{x};\mathbf{w})
=
\frac{\exp\!\big(\mathbf{w}^\top \Phi(\mathbf{x},\mathbf{s})\big)}
{\sum_{\mathbf{s'}\in \mathcal{S}^m} \exp\!\big(\mathbf{w}^\top \Phi(\mathbf{x},\mathbf{s'})\big)}
$$


## Local-to-Global Feature Decomposition

Assume features decompose along adjacent transitions(**“Adjacent transitions” simply means the pairs of neighboring states $(s_{j-1}, s_j)$ along the sequence.**)
:

$$
\Phi(\mathbf{x},\mathbf{s})
=
\sum_{j=1}^{m} \phi\!\big(x_j, s_{j-1}, s_j\big)
$$

For each coordinate:

$$
\Phi_k(\mathbf{x},\mathbf{s})
=
\sum_{j=1}^{m} \phi_k\!\big(x_j, s_{j-1}, s_j\big)
$$



## Decoding (Most Likely Sequence)

Goal:

$$
\arg\max_{\mathbf{s}\in \mathcal{S}^m} p(\mathbf{s}\mid \mathbf{x};\mathbf{w})
=
\arg\max_{\mathbf{s}\in \mathcal{S}^m} \mathbf{w}^\top \Phi(\mathbf{x},\mathbf{s})
=
\arg\max_{\mathbf{s}\in \mathcal{S}^m}
\sum_{j=1}^{m} \mathbf{w}^\top \phi\!\big(x_j, s_{j-1}, s_j\big)
$$

Define the transition score:

$$
\text{score}\!\left(s_{j-1}\!\to s_j; x_j\right)
=
\mathbf{w}^\top \phi\!\big(x_j, s_{j-1}, s_j\big)
$$

### Viterbi-Style Dynamic Program $(O(m|\mathcal{S}|^2))$

Initialization (with special start state $ s_0 $ described in text):

$$
V[1,s] = \mathbf{w}^\top \phi(x_1, s_0, s)
$$

Recurrence (for $ j=2,\dots,m $):

$$
V[j,s] = \max_{a\in \mathcal{S}}\Big( V[j-1,a] + \mathbf{w}^\top \phi(x_j, a, s) \Big)
$$

Termination:

$$
\max_{\mathbf{s}\in\mathcal{S}^m}
\sum_{j=1}^{m} \mathbf{w}^\top \phi(x_j, s_{j-1}, s_j)
=
\max_{s\in\mathcal{S}} V[m,s]
$$

(Use backpointers to reconstruct the best sequence.)



## Parameter Estimation (Training)

Regularized conditional log-likelihood over labeled pairs $ (\mathbf{x}^{(i)}, \mathbf{s}^{(i)}) $, $ i=1,\dots,n $:

$$
\mathcal{L}(\mathbf{w})
=
\sum_{i=1}^{n}
\log p\!\big(\mathbf{s}^{(i)} \mid \mathbf{x}^{(i)}; \mathbf{w}\big)
-
\frac{\lambda}{2}\,\lVert \mathbf{w}\rVert_2^2
$$

Estimator:

$$
\hat{\mathbf{w}}
=
\arg\max_{\mathbf{w}\in\mathbb{R}^d} \mathcal{L}(\mathbf{w})
$$

### Gradient (Empirical − Expected Counts − L2)

$$
\frac{\partial \mathcal{L}}{\partial w_k}
=
\sum_{i=1}^{n}\Phi_k\!\big(\mathbf{x}^{(i)}, \mathbf{s}^{(i)}\big)
-
\sum_{i=1}^{n}\sum_{\mathbf{s}\in\mathcal{S}^m}
p(\mathbf{s}\mid \mathbf{x}^{(i)}; \mathbf{w})\,
\Phi_k\!\big(\mathbf{x}^{(i)}, \mathbf{s}\big)
-
\lambda w_k
$$

Using the local decomposition:

$$
\sum_{\mathbf{s}}
p(\mathbf{s}\mid \mathbf{x};\mathbf{w})\,
\Phi_k(\mathbf{x},\mathbf{s})
=
\sum_{j=1}^{m}\sum_{a\in\mathcal{S}}\sum_{b\in\mathcal{S}}
q_j(a,b)\,\phi_k(x_j,a,b)
$$

Edge marginal definition:

$$
q_j(a,b)
=
\Pr\!\big(s_{j-1}=a,\, s_{j}=b \mid \mathbf{x}; \mathbf{w}\big)
$$


## Forward–Backward for Edge Marginals

Local potentials:

$$
\psi_j(a,b)
=
\exp\!\big(\mathbf{w}^\top \phi(x_j,a,b)\big)
$$

Forward messages:

$$
\alpha_1(b) = \psi_1(s_0,b)
$$

$$
\alpha_j(b) = \sum_{a\in\mathcal{S}} \alpha_{j-1}(a)\,\psi_j(a,b)
$$

Backward messages:

$$
\beta_m(b) = 1
$$

$$
\beta_{j-1}(a) = \sum_{b\in\mathcal{S}} \psi_j(a,b)\,\beta_j(b)
$$

Partition function:

$$
Z(\mathbf{x}) = \sum_{s\in\mathcal{S}} \alpha_m(s)
$$

Edge marginals:

$$
q_j(a,b)
=
\frac{\alpha_{j-1}(a)\,\psi_j(a,b)\,\beta_j(b)}
{Z(\mathbf{x})}
$$



## Complexity

Decoding (Viterbi):

$$
O(m\,|\mathcal{S}|^2)
$$

Forward–Backward expectations:

$$
O(m\,|\mathcal{S}|^2)
$$

Linear-chain CRFs are efficient thanks to dynamic programming over the chain.

# Why CRF solved his problems:
- CRF is like MEMM but **globally normalized**.  
- It still models $P(Y \mid X)$ (discriminative).  
- But instead of local decisions, CRF looks at the **whole sequence**.  
- It scores features and transitions **globally** and normalizes across all possible tag sequences.  
- This removes the **label bias problem**.

 In **linear-chain CRF** (used for NER), training uses methods like gradient descent  
It’s heavier than MEMM but gives better results.

### Benefits for NER:
- Like MEMM, CRF allows rich features.  
- But CRF makes decisions using the **whole sentence**.  
- Example: *Delhi* will correctly be tagged as a **LOCATION**, even if local bias wanted to mark it as “O”.  
- In modern NER, CRF is often combined with deep models (e.g., BiLSTM-CRF) for even higher accuracy.

````python
# Cell 1: Import and setup
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
import logging

# Set up logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
logger.info(f'Using device: {device}')
````

````python
# Dataset class
class NERDataset(Dataset):
    def __init__(self, sentences, word_to_id, tag_to_id, max_length=128):
        self.sentences = sentences
        self.word_to_id = word_to_id
        self.tag_to_id = tag_to_id
        self.max_length = max_length
        
    def __len__(self):
        return len(self.sentences)
    
    def __getitem__(self, idx):
        sentence_data = self.sentences[idx]
        word_ids = sentence_data['words']
        tag_ids = sentence_data['tags']
        
        # Pad sequences
        padded_word_ids = word_ids[:self.max_length] + [self.word_to_id['<PAD>']] * max(0, self.max_length - len(word_ids))
        padded_tag_ids = tag_ids[:self.max_length] + [self.tag_to_id['O']] * max(0, self.max_length - len(tag_ids))
        
        # Create mask (1 for real tokens, 0 for padding)
        mask = [1] * min(len(word_ids), self.max_length) + [0] * max(0, self.max_length - len(word_ids))
        
        return {
            'word_ids': torch.tensor(padded_word_ids, dtype=torch.long),
            'tag_ids': torch.tensor(padded_tag_ids, dtype=torch.long),
            'mask': torch.tensor(mask, dtype=torch.bool),
            'length': len(word_ids)
        }
````

````python
# Cell 3: FeatureExtractor class
class FeatureExtractor:
    """
    Feature extraction for CRF model
    """
    def __init__(self, word_to_id, window_size=2):
        self.word_to_id = word_to_id
        self.window_size = window_size
        self.feature_dim = self._calculate_feature_dim()

    def _calculate_feature_dim(self):
        # Word identity features
        dim = len(self.word_to_id)

        # Context window features (left and right)
        dim += 2 * self.window_size * len(self.word_to_id)

        # Position features
        dim += 10  # For position bins

        # Word shape features
        dim += 5   # For different word patterns

        return dim

    def get_features(self, position, word_ids, word_to_id):
        """
        Extract features for a word at given position
        """
        features = []

        # 1. Current word identity (one-hot)
        current_word_features = self._one_hot(word_ids[position], len(word_to_id))
        features.extend(current_word_features)

        # 2. Context window features
        context_features = self._get_context_features(position, word_ids, word_to_id)
        features.extend(context_features)

        # 3. Position features
        position_features = self._get_position_features(position, len(word_ids))
        features.extend(position_features)

        # 4. Word shape features (simplified for Hindi)
        shape_features = self._get_word_shape_features(position, word_ids, word_to_id)
        features.extend(shape_features)

        return features

    def _one_hot(self, index, size):
        vec = [0] * size
        if index < size:
            vec[index] = 1
        return vec

    def _get_context_features(self, position, word_ids, word_to_id):
        features = []
        sentence_length = len(word_ids)
        vocab_size = len(word_to_id)

        # Left context
        for i in range(1, self.window_size + 1):
            if position - i >= 0:
                features.extend(self._one_hot(word_ids[position - i], vocab_size))
            else:
                features.extend([0] * vocab_size)  # Padding

        # Right context
        for i in range(1, self.window_size + 1):
            if position + i < sentence_length:
                features.extend(self._one_hot(word_ids[position + i], vocab_size))
            else:
                features.extend([0] * vocab_size)  # Padding

        return features

    def _get_position_features(self, position, sentence_length):
        """Simple position features"""
        features = [0] * 10
        if sentence_length > 0:
            pos_bin = min(9, int((position / sentence_length) * 10))
            features[pos_bin] = 1
        return features

    def _get_word_shape_features(self, position, word_ids, word_to_id):
        """
        Simplified word shape features for Hindi
        In practice, you'd want more sophisticated features for Hindi
        """
        # Placeholder for actual word shape features
        # These would typically include:
        # - Word length
        # - Presence of specific characters/patterns
        # - Word prefixes/suffixes
        return [0, 0, 0, 0, 0]  # Simplified
````

````python
# Utility functions
def load_conll_file(file_path: str):
    sentences = []
    current_sentence = []
    
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                if current_sentence:
                    sentences.append(current_sentence)
                    current_sentence = []
            else:
                parts = line.split()
                if len(parts) >= 2:
                    token = parts[0]
                    tag = parts[1]
                    current_sentence.append((token, tag))
        
        if current_sentence:
            sentences.append(current_sentence)
    
    return sentences

def create_mappings(sentences):
    word_freq = Counter()
    tag_set = set()
    
    for sentence in sentences:
        for token, tag in sentence:
            word_freq[token] += 1
            tag_set.add(tag)
    
    sorted_words = [word for word, count in word_freq.most_common()]
    
    word_to_id = {'<UNK>': 0, '<PAD>': 1}
    id_to_word = {0: '<UNK>', 1: '<PAD>'}
    
    for idx, word in enumerate(sorted_words, 2):
        word_to_id[word] = idx
        id_to_word[idx] = word
    
    tag_list = sorted(list(tag_set))
    tag_to_id = {}
    id_to_tag = {}
    
    for idx, tag in enumerate(tag_list):
        tag_to_id[tag] = idx
        id_to_tag[idx] = tag
    
    return word_to_id, id_to_word, tag_to_id, id_to_tag

def prepare_corpus(sentences, word_to_id, tag_to_id):
    corpus = []
    
    for sentence in sentences:
        word_ids = []
        tag_ids = []
        
        for token, tag in sentence:
            word_id = word_to_id.get(token, word_to_id['<UNK>'])
            tag_id = tag_to_id[tag]
            
            word_ids.append(word_id)
            tag_ids.append(tag_id)
        
        corpus.append({
            'words': word_ids,
            'tags': tag_ids
        })
    
    return corpus
````

#### Class 2: CRFModel

Purpose: Defines the Neural Network + CRF layer.

Responsibilities:
- Word embedding lookup (nn.Embedding).
- Linear layer that projects embeddings into tag space (emission scores).
- CRF mechanics:
  - Forward algorithm (partition function).
  - Viterbi decoding for prediction.
  - Negative log-likelihood for training loss.

````python
# Fixed CRF Model
class CRFModel(nn.Module):
    def __init__(self, num_tags, vocab_size, embedding_dim=128, hidden_dim=256):
        super().__init__()
        self.num_tags = num_tags
        self.vocab_size = vocab_size
        
        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=1)
        
        # BiLSTM layer
        self.lstm = nn.LSTM(embedding_dim, hidden_dim // 2, 
                           num_layers=1, bidirectional=True, batch_first=True)
        
        # Linear layer to convert LSTM outputs to emission scores
        self.hidden2tag = nn.Linear(hidden_dim, num_tags)
        
        # CRF transition parameters
        self.transitions = nn.Parameter(torch.randn(num_tags, num_tags))
        
        # Constraints
        self.START_TAG = num_tags - 2
        self.STOP_TAG = num_tags - 1
        
        # Initialize transitions with constraints
        self.transitions.data[self.START_TAG, :] = -10000
        self.transitions.data[:, self.STOP_TAG] = -10000
        
    def forward(self, sentence_ids):
        """Forward pass - compute emission scores"""
        embeddings = self.embedding(sentence_ids)
        lstm_out, _ = self.lstm(embeddings)
        emissions = self.hidden2tag(lstm_out)
        return emissions
    
    def _forward_algorithm(self, emissions, mask):
        """Compute log partition function with masking"""
        batch_size, seq_length, num_tags = emissions.shape
        
        # Initialize alpha in log space
        alpha = torch.full((batch_size, num_tags), -10000.0).to(device)
        alpha[:, self.START_TAG] = 0.0
        
        # Iterate through sequence
        for t in range(seq_length):
            current_mask = mask[:, t].unsqueeze(1)  # (batch_size, 1)
            emit_scores = emissions[:, t]  # (batch_size, num_tags)
            
            # Broadcast alpha for transitions
            alpha_expanded = alpha.unsqueeze(2)  # (batch_size, num_tags, 1)
            transitions_expanded = self.transitions.unsqueeze(0)  # (1, num_tags, num_tags)
            
            # Compute scores for all transitions
            scores = alpha_expanded + transitions_expanded + emit_scores.unsqueeze(1)
            
            # Log-sum-exp over previous states
            new_alpha = torch.logsumexp(scores, dim=1)
            
            # Update alpha with masking
            alpha = new_alpha * current_mask + alpha * (~current_mask)
        
        # Add final transitions to STOP_TAG
        final_scores = alpha + self.transitions[:, self.STOP_TAG].unsqueeze(0)
        log_partition = torch.logsumexp(final_scores, dim=1)
        
        return log_partition
    
    def _score_sequence(self, emissions, tags, mask):
        """Compute score for gold tag sequence"""
        batch_size, seq_length, num_tags = emissions.shape
        
        # Score from START_TAG to first tag
        score = self.transitions[self.START_TAG, tags[:, 0]]
        score += emissions[:, 0].gather(1, tags[:, 0].unsqueeze(1)).squeeze(1)
        
        # Score for each transition
        for t in range(1, seq_length):
            current_mask = mask[:, t]
            prev_tags = tags[:, t-1]
            current_tags = tags[:, t]
            
            # Transition scores
            transition_score = self.transitions[prev_tags, current_tags]
            
            # Emission scores
            emission_score = emissions[:, t].gather(1, current_tags.unsqueeze(1)).squeeze(1)
            
            # Update score
            score += (transition_score + emission_score) * current_mask
        
        # Score to STOP_TAG from last tag
        last_positions = mask.sum(dim=1).long() - 1
        last_tags = tags.gather(1, last_positions.unsqueeze(1)).squeeze(1)
        score += self.transitions[last_tags, self.STOP_TAG]
        
        return score
    
    def neg_log_likelihood(self, sentence_ids, tags, mask):
        """Compute negative log likelihood loss"""
        emissions = self.forward(sentence_ids)
        
        # Compute scores
        gold_score = self._score_sequence(emissions, tags, mask)
        forward_score = self._forward_algorithm(emissions, mask)
        
        # Negative log likelihood (average over batch)
        nll = forward_score - gold_score
        return nll.mean()  # Return scalar for backpropagation
    
    def viterbi_decode(self, emissions, mask):
        """Viterbi decoding for prediction"""
        batch_size, seq_length, num_tags = emissions.shape
        
        # Initialize backpointers and viterbi variables
        backpointers = torch.zeros(batch_size, seq_length, num_tags, dtype=torch.long).to(device)
        viterbi = torch.full((batch_size, num_tags), -10000.0).to(device)
        viterbi[:, self.START_TAG] = 0.0
        
        # Forward pass
        for t in range(seq_length):
            current_mask = mask[:, t].unsqueeze(1)
            emit_scores = emissions[:, t]
            
            # Expand viterbi for transitions
            viterbi_expanded = viterbi.unsqueeze(2)  # (batch_size, num_tags, 1)
            transitions_expanded = self.transitions.unsqueeze(0)  # (1, num_tags, num_tags)
            
            # Compute scores
            scores = viterbi_expanded + transitions_expanded + emit_scores.unsqueeze(1)
            
            # Find best scores and backpointers
            best_scores, best_tags = torch.max(scores, dim=1)
            backpointers[:, t] = best_tags
            
            # Update viterbi with masking
            viterbi = best_scores * current_mask + viterbi * (~current_mask)
        
        # Add final transitions to STOP_TAG
        final_scores = viterbi + self.transitions[:, self.STOP_TAG].unsqueeze(0)
        _, best_final_tags = torch.max(final_scores, dim=1)
        
        # Backtrack to find best path
        best_paths = []
        for i in range(batch_size):
            best_path = [best_final_tags[i].item()]
            valid_length = mask[i].sum().item()
            
            for t in range(seq_length - 1, 0, -1):
                if t < valid_length:
                    best_tag = backpointers[i, t, best_path[0]]
                    best_path.insert(0, best_tag.item())
            
            best_paths.append(best_path)
        
        return best_paths
    
    def decode(self, sentence_ids, mask):
        """Decode best tag sequence"""
        emissions = self.forward(sentence_ids)
        return self.viterbi_decode(emissions, mask)
````

#### Class 3: Trainer

Purpose: Handles training and evaluation loop.

Responsibilities:
- Initialize optimizer and manage training epochs.
- Call neg_log_likelihood from CRFModel to compute loss.
- Backpropagation and optimizer step.
- Evaluate on validation/test sets:
  - Decode predictions using CRFModel.viterbi_decode.
  - Convert tag IDs back to tag strings.
  - Print classification metrics (Precision, Recall, F1).

````python
from tqdm import tqdm

# Trainer class
class Trainer:
    def __init__(self, model, device, lr=0.01):
        self.model = model.to(device)
        self.device = device
        self.optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
        
    def train_model(self, train_loader, val_loader, epochs=10):
        """Fixed training loop with proper loss handling"""
        train_losses = []
        val_losses = []
        
        for epoch in range(epochs):
            # Training
            self.model.train()
            total_train_loss = 0
            num_batches = 0
            
            for batch in tqdm(train_loader, desc=f'Epoch {epoch+1}/{epochs} - Training'):
                word_ids = batch['word_ids'].to(self.device)
                tag_ids = batch['tag_ids'].to(self.device)
                mask = batch['mask'].to(self.device)
                
                # Forward pass - get scalar loss
                loss = self.model.neg_log_likelihood(word_ids, tag_ids, mask)
                
                # Backward pass
                self.optimizer.zero_grad()
                loss.backward()  # Now this works because loss is scalar
                torch.nn.utils.clip_grad_norm_(self.model.parameters(), 5.0)
                self.optimizer.step()
                
                total_train_loss += loss.item()
                num_batches += 1
            
            avg_train_loss = total_train_loss / num_batches
            train_losses.append(avg_train_loss)
            
            # Validation
            self.model.eval()
            total_val_loss = 0
            num_val_batches = 0
            
            with torch.no_grad():
                for batch in tqdm(val_loader, desc=f'Epoch {epoch+1}/{epochs} - Validation'):
                    word_ids = batch['word_ids'].to(self.device)
                    tag_ids = batch['tag_ids'].to(self.device)
                    mask = batch['mask'].to(self.device)
                    
                    loss = self.model.neg_log_likelihood(word_ids, tag_ids, mask)
                    total_val_loss += loss.item()
                    num_val_batches += 1
            
            avg_val_loss = total_val_loss / num_val_batches
            val_losses.append(avg_val_loss)
            
            print(f"Epoch {epoch+1}/{epochs}: Train Loss = {avg_train_loss:.4f}, Val Loss = {avg_val_loss:.4f}")
        
        # Plot training history
        self.plot_training_history(train_losses, val_losses)
        
        return train_losses, val_losses
    
    def plot_training_history(self, train_losses, val_losses):
        plt.figure(figsize=(10, 6))
        plt.plot(train_losses, label='Training Loss')
        plt.plot(val_losses, label='Validation Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.title('Training History')
        plt.legend()
        plt.grid(True)
        plt.show()
    
    @torch.no_grad()
    def evaluate_model(self, test_loader, id_to_tag):
        """Run evaluation loop and print metrics."""
        self.model.eval()
        
        all_true_tags = []
        all_pred_tags = []
        
        for batch in tqdm(test_loader, desc='Evaluating'):
            word_ids = batch['word_ids'].to(self.device)
            mask = batch['mask'].to(self.device)
            true_tags = batch['tag_ids']
            lengths = batch['length']
            
            # Decode predictions
            pred_tag_sequences = self.model.decode(word_ids, mask)
            
            # Convert to flat lists
            for i, length in enumerate(lengths):
                true_seq = true_tags[i][:length].tolist()
                
                # Ensure we don't exceed the predicted sequence length
                if i < len(pred_tag_sequences):
                    pred_seq = pred_tag_sequences[i]
                    # Take only up to the actual length
                    pred_seq = pred_seq[:length]
                    
                    # Convert tag IDs to tag names (excluding START/STOP tags)
                    for j in range(min(len(true_seq), len(pred_seq))):
                        true_tag_id = true_seq[j]
                        pred_tag_id = pred_seq[j]
                        
                        # Only include if it's a valid tag (not START/STOP)
                        if true_tag_id in id_to_tag and id_to_tag[true_tag_id] not in ['<START>', '<STOP>']:
                            all_true_tags.append(id_to_tag[true_tag_id])
                            all_pred_tags.append(id_to_tag.get(pred_tag_id, 'O'))  # Default to 'O' if unknown
        
        # Print classification report
        print("\n=== Classification Report ===")
        print(classification_report(all_true_tags, all_pred_tags, zero_division=0))
        
        # Plot confusion matrix
        self.plot_confusion_matrix(all_true_tags, all_pred_tags, [tag for tag in id_to_tag.values() if tag not in ['<START>', '<STOP>']])
        
        return all_true_tags, all_pred_tags
    
    def plot_confusion_matrix(self, y_true, y_pred, labels):
        cm = confusion_matrix(y_true, y_pred, labels=labels)
        
        plt.figure(figsize=(12, 10))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                   xticklabels=labels, yticklabels=labels)
        plt.title('Confusion Matrix')
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.xticks(rotation=45)
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()
````

### Q1. What is the main difference between a Hidden Markov Model (HMM) and a Conditional Random Field (CRF) in terms of how they model probabilities? (10 marks)

The main difference lies in how they model the joint versus conditional probabilities.
An **HMM** is a *generative model* that learns the **joint probability** of the observations and the tag sequence, i.e., ( P(X, Y) = P(Y)P(X|Y) ). It assumes strong independence between observations given the tags.

In contrast, a **CRF** is a *discriminative model* that directly models the **conditional probability** of the tag sequence given the observations, i.e., ( P(Y|X) ). This allows CRFs to consider *arbitrary and overlapping features* of the input without making independence assumptions.

As a result, CRFs generally achieve higher accuracy in sequence labeling tasks like NER because they focus on learning decision boundaries rather than modeling the data generation process.


### Q2. Why are CRFs less prone to the label bias problem compared to Maximum Entropy Markov Models (MEMMs)? (10 marks)

CRFs are less prone to the **label bias problem** because they **normalize over the entire sequence of tags**, not just locally at each state.

In **MEMMs**, each state makes independent decisions by normalizing transition probabilities only over its outgoing edges. This causes states with fewer outgoing transitions to become “biased” toward themselves, even if they are not the best global choice.

In contrast, **CRFs compute a single global normalization factor** over all possible tag sequences, which allows information to flow across the entire sequence. This global normalization ensures that decisions depend on the **overall context** rather than local transition probabilities, thereby reducing label bias.


### Q3. In a linear-chain CRF, what are the two main components of the score for a tag sequence? $(5 \times 3 = 15\ marks)$

* (a) Emission scores and transition scores
* (b) Transition scores and loss scores
* (c) Emission scores and gradient scores

✅ **Correct answer:** (a) **Emission scores and transition scores**

In a linear-chain CRF, the total score for a tag sequence is computed as the **sum of emission scores** (which measure how well each tag fits the observed word or features) and **transition scores** (which measure how likely one tag is to follow another). These two components together capture both the local evidence from the input and the dependencies between consecutive tags.


### Q4. Suppose you have a sentence with 5 words and you are tagging with 3 possible labels. How many possible tag sequences exist for this sentence in a CRF model? (20 marks)

Each word in the sentence can take **any of the 3 possible labels**, and there are **5 words** in total.
Therefore, the total number of possible tag sequences is:

[
3^5 = 3 \times 3 \times 3 \times 3 \times 3 = 243
]

✅ **Answer:** There are **243 possible tag sequences** for the 5-word sentence.


### Q5. During CRF training, we maximize the conditional log-likelihood. Which two terms are needed to compute this objective?  $(5 \times 3 = 15\ marks)$

* (a) Score of the gold sequence and partition function
* (b) Only the emission score
* (c) Only the transition score

✅ **Correct answer:** (a) **Score of the gold sequence and partition function**

In CRF training, the **conditional log-likelihood** measures how probable the correct (gold) tag sequence is compared to all possible tag sequences. It is computed as:

$
\log P(Y|X) = \text{Score}(X, Y) - \log Z(X)
$

where **Score(X, Y)** is the total score (sum of emission and transition scores) for the gold tag sequence, and **Z(X)** is the **partition function**, which normalizes over all possible tag sequences.

# Introduction and Motivation

## Introduction
Sequential data is at the heart of many real-world applications such as speech recognition, machine translation, sentiment analysis, and named entity recognition (NER). The challenge lies in effectively modeling temporal dependencies and contextual relationships across varying sequence lengths. Over time, different approaches have been proposed to handle sequential data, starting from probabilistic models to deep neural architectures.


## Motivation
Early approaches such as **Hidden Markov Models (HMMs)** provided a probabilistic framework for modeling sequences, where hidden states generate observable sequences. While successful in certain domains, HMMs rely on strong independence assumptions and struggle to capture long-term dependencies. Training with algorithms like **Baum-Welch** often led to challenges with data sparsity and scalability.

To overcome some of these limitations, **Conditional Random Fields (CRFs)** were introduced. Unlike HMMs, CRFs are discriminative models that directly model the conditional probability of label sequences given inputs. They allow the inclusion of richer features and are effective in tasks like part-of-speech tagging and NER. However, CRFs still depend heavily on hand-engineered features and cannot automatically learn hierarchical representations.

The emergence of **Recurrent Neural Networks (RNNs)** shifted the paradigm by enabling models to learn representations directly from raw data. RNNs can, in principle, capture sequential dependencies but suffer from vanishing and exploding gradient problems, which limit their ability to handle long sequences.

**Long Short-Term Memory (LSTM) networks** addressed these issues through gating mechanisms that control information flow, allowing them to retain relevant information over longer contexts. This made LSTMs powerful for a wide range of sequential tasks, from language modeling to speech recognition.

Building on this, **Bidirectional LSTMs (BiLSTMs)** enhanced performance by processing input sequences in both forward and backward directions, thus capturing context from both past and future tokens. Extending this further, **Multi-layer BiLSTMs** stack several layers to learn increasingly abstract and hierarchical representations of sequential data.

In summary, the journey from HMMs and CRFs to LSTMs and their variants demonstrates a clear progression: from **probabilistic models with restrictive assumptions and manual features** to **neural architectures that automatically learn rich, long-range, and hierarchical representations**. This evolution motivates the use of LSTMs, BiLSTMs, and Multi-layer BiLSTMs for modern sequence modeling tasks.


## Task 1: Vanilla LSTM Implementation (25 Marks)

### Conceptual Background
- RNNs suffer from vanishing/exploding gradients, making them ineffective for long-term dependencies.  
- LSTM introduces a cell state ($C_t$) and gates (forget, input, output) to regulate memory flow.  

### Mathematical Formulation
Let input be $x_t$, previous hidden be $h_{t-1}$, previous cell be $C_{t-1}$.

**1. Forget Gate**
$$
f_t = \sigma\big(W_f x_t + U_f h_{t-1} + b_f\big)
$$

**2. Input Gate**
$$
i_t = \sigma\big(W_i x_t + U_i h_{t-1} + b_i\big)
$$

**3. Candidate Cell State**
$$
\tilde{C}_t = \tanh\big(W_c x_t + U_c h_{t-1} + b_c\big)
$$

**4. Cell State Update**
$$
C_t = f_t \odot C_{t-1} + i_t \odot \tilde{C}_t
$$

**5. Output Gate**
$$
o_t = \sigma\big(W_o x_t + U_o h_{t-1} + b_o\big)
$$

**6. Hidden State**
$$
h_t = o_t \odot \tanh(C_t)
$$

> Here $\sigma(\cdot)$ is the logistic sigmoid, $\tanh(\cdot)$ is the hyperbolic tangent, and $\odot$ denotes element-wise multiplication.

## Task 2: Bidirectional LSTM (BiLSTM) (25 Marks)

### Conceptual Background
- A standard LSTM captures *past* context only.  
- BiLSTM uses two LSTMs:
  - **Forward LSTM** processes $x_1 \to x_T$.
  - **Backward LSTM** processes $x_T \to x_1$.

### Mathematical Formulation

**Forward LSTM**
$$
\overrightarrow{h}_t = \mathrm{LSTM}_{\text{fwd}}(x_t, \overrightarrow{h}_{t-1})
$$

**Backward LSTM**
$$
\overleftarrow{h}_t = \mathrm{LSTM}_{\text{bwd}}(x_t, \overleftarrow{h}_{t+1})
$$

**Combined Hidden State** (concatenation)
$$
h_t = \big[\,\overrightarrow{h}_t \,;\, \overleftarrow{h}_t\,\big]
$$

**Output Sequence**
$$
H = \big(h_1, h_2, \dots, h_T\big)
$$

## Task 3: Multi-layer BiLSTM (Stacked BiLSTM) (25 Marks)

### Conceptual Background
- A single BiLSTM layer may not capture hierarchical or deep temporal patterns.  
- Stacking multiple BiLSTM layers allows successive layers to build higher-level representations: lower layers capture local patterns; higher layers capture more abstract, global features.

### Notation
- Let $L$ be the number of stacked BiLSTM layers.
- Superscript $(l)$ denotes the layer index (e.g., first layer $l=1$, final layer $l=L$).
- Subscript $t$ denotes the time step.

### First BiLSTM Layer ($l=1$)
**Forward direction**
$$
\overrightarrow{h}_t^{(1)} = \mathrm{LSTM}_{\text{fwd}}^{(1)}\big(x_t,\; \overrightarrow{h}_{t-1}^{(1)}\big)
$$

**Backward direction**
$$
\overleftarrow{h}_t^{(1)} = \mathrm{LSTM}_{\text{bwd}}^{(1)}\big(x_t,\; \overleftarrow{h}_{t+1}^{(1)}\big)
$$

**Concatenate to form layer-1 representation**
$$
h_t^{(1)} = \big[\,\overrightarrow{h}_t^{(1)} \,;\, \overleftarrow{h}_t^{(1)}\,\big]
$$

### Higher Layers ($l = 2, \dots, L$)

For **layer $l$**, the input at each time step is the concatenated output from layer $l-1$, i.e., $h_t^{(l-1)}$.

**Forward direction**
$$
\overrightarrow{h}_t^{(l)} = \mathrm{LSTM}_{\text{fwd}}^{(l)}\big(h_t^{(l-1)},\; \overrightarrow{h}_{t-1}^{(l)}\big)
$$

**Backward direction**
$$
\overleftarrow{h}_t^{(l)} = \mathrm{LSTM}_{\text{bwd}}^{(l)}\big(h_t^{(l-1)},\; \overleftarrow{h}_{t+1}^{(l)}\big)
$$

**Concatenate to form layer-$l$ representation**
$$
h_t^{(l)} = \big[\,\overrightarrow{h}_t^{(l)} \,;\, \overleftarrow{h}_t^{(l)}\,\big]
$$

### Final Output after $L$ layers
$$
H^{(L)} = \big(h_1^{(L)},\; h_2^{(L)},\; \dots,\; h_T^{(L)}\big)
$$

Implement LSTM cell for single step

````python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
import logging
from tqdm import tqdm

# Set up logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
logger.info(f'Using device: {device}')

## Task 1: Vanilla LSTM Implementation

class LSTMCell(nn.Module):
    """
    A single LSTM cell that implements the logic for one time step.
    """
    def __init__(self, input_size, hidden_size):
        super(LSTMCell, self).__init__()
        self.input_size = input_size
        self.hidden_size = hidden_size

        # Input gate weights
        self.W_xi = nn.Linear(input_size, hidden_size, bias=False)
        self.W_hi = nn.Linear(hidden_size, hidden_size, bias=True)

        # Forget gate weights
        self.W_xf = nn.Linear(input_size, hidden_size, bias=False)
        self.W_hf = nn.Linear(hidden_size, hidden_size, bias=True)

        # Output gate weights
        self.W_xo = nn.Linear(input_size, hidden_size, bias=False)
        self.W_ho = nn.Linear(hidden_size, hidden_size, bias=True)

        # Cell candidate weights
        self.W_xc = nn.Linear(input_size, hidden_size, bias=False)
        self.W_hc = nn.Linear(hidden_size, hidden_size, bias=True)

        # Initialize weights
        self._init_weights()

    def _init_weights(self):
        """Initialize weights using Xavier initialization"""
        for name, param in self.named_parameters():
            if 'weight' in name:
                nn.init.xavier_uniform_(param)
            elif 'bias' in name:
                nn.init.constant_(param, 0)
                # Set forget gate bias to 1 to help with gradient flow
                if 'W_hf' in name:
                    nn.init.constant_(param, 1)

    def forward(self, x, states):
        """
        Forward pass of the LSTM cell.

        Args:
            x: Input tensor at current time step (batch_size, input_size)
            states: Tuple of (h_prev, c_prev) - previous hidden and cell states

        Returns:
            h_next: Next hidden state (batch_size, hidden_size)
            c_next: Next cell state (batch_size, hidden_size)
        """
        h_prev, c_prev = states

        # Input gate
        i_t = torch.sigmoid(self.W_xi(x) + self.W_hi(h_prev))

        # Forget gate
        f_t = torch.sigmoid(self.W_xf(x) + self.W_hf(h_prev))

        # Output gate
        o_t = torch.sigmoid(self.W_xo(x) + self.W_ho(h_prev))

        # Cell candidate
        c_tilde = torch.tanh(self.W_xc(x) + self.W_hc(h_prev))

        # Cell state update
        c_next = f_t * c_prev + i_t * c_tilde

        # Hidden state update
        h_next = o_t * torch.tanh(c_next)

        return h_next, c_next

````

Implement full general LSTM layer

````python
class LSTM(nn.Module):
    """
    A full LSTM layer that uses LSTMCell to process an input sequence.
    """
    def __init__(self, input_size, hidden_size, bidirectional=False, num_layers=1, dropout=0.3):
        super(LSTM, self).__init__()
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.bidirectional = bidirectional
        self.num_layers = num_layers
        self.num_directions = 2 if bidirectional else 1

        # Create LSTM cells for each layer and direction
        self.cells = nn.ModuleDict()

        for layer in range(num_layers):
            layer_input_size = input_size if layer == 0 else hidden_size * self.num_directions

            if bidirectional:
                # Forward and backward cells for each layer
                self.cells[f'forward_{layer}'] = LSTMCell(layer_input_size, hidden_size)
                self.cells[f'backward_{layer}'] = LSTMCell(layer_input_size, hidden_size)
            else:
                # Single forward cell for each layer
                self.cells[f'forward_{layer}'] = LSTMCell(layer_input_size, hidden_size)

        self.dropout = nn.Dropout(dropout) if dropout > 0 else nn.Identity()

    def forward(self, input_seq, states=None):
        """
        Forward pass of the full LSTM layer.

        Args:
            input_seq: Input sequence (batch_size, seq_len, input_size)
            states: Tuple of (h_0, c_0) - initial hidden and cell states

        Returns:
            output: Output sequence (batch_size, seq_len, num_directions * hidden_size)
            (h_n, c_n): Final hidden and cell states
        """
        batch_size, seq_len, _ = input_seq.shape

        # Initialize states if not provided
        if states is None:
            h_0 = torch.zeros(self.num_layers * self.num_directions, batch_size, self.hidden_size, device=input_seq.device)
            c_0 = torch.zeros(self.num_layers * self.num_directions, batch_size, self.hidden_size, device=input_seq.device)
        else:
            h_0, c_0 = states

        # Process each layer
        layer_outputs = []
        layer_hidden_states = []
        layer_cell_states = []

        current_input = input_seq  # (batch_size, seq_len, input_size)

        for layer in range(self.num_layers):
            if self.bidirectional:
                # Forward pass
                h_forward, c_forward, output_forward = self._process_direction(
                    current_input, 'forward', layer,
                    h_0[layer * 2], c_0[layer * 2]
                )

                # Backward pass
                h_backward, c_backward, output_backward = self._process_direction(
                    current_input, 'backward', layer,
                    h_0[layer * 2 + 1], c_0[layer * 2 + 1]
                )

                # Concatenate forward and backward outputs
                output = torch.cat([output_forward, output_backward], dim=-1)
                h_n = torch.stack([h_forward, h_backward])
                c_n = torch.stack([c_forward, c_backward])

            else:
                # Unidirectional forward pass
                h_forward, c_forward, output = self._process_direction(
                    current_input, 'forward', layer,
                    h_0[layer], c_0[layer]
                )
                h_n = h_forward.unsqueeze(0)
                c_n = c_forward.unsqueeze(0)

            # Apply dropout between layers (except last layer)
            if layer < self.num_layers - 1:
                output = self.dropout(output)

            layer_outputs.append(output)
            layer_hidden_states.append(h_n)
            layer_cell_states.append(c_n)

            # Input to next layer is output from current layer
            current_input = output

        # Stack outputs and states
        final_output = layer_outputs[-1]
        h_n = torch.cat(layer_hidden_states, dim=0)
        c_n = torch.cat(layer_cell_states, dim=0)

        return final_output, (h_n, c_n)

    def _process_direction(self, input_seq, direction, layer, h_0, c_0):
        """
        Process sequence in one direction (forward or backward)
        """
        batch_size, seq_len, input_size = input_seq.shape
        cell = self.cells[f'{direction}_{layer}']

        # Initialize hidden and cell states
        h_t = h_0  # (batch_size, hidden_size)
        c_t = c_0  # (batch_size, hidden_size)

        outputs = []
        hidden_states = []
        cell_states = []

        # Determine processing order
        if direction == 'forward':
            indices = range(seq_len)
        else:  # backward
            indices = range(seq_len - 1, -1, -1)

        # Process sequence
        for t in indices:
            x_t = input_seq[:, t, :]  # (batch_size, input_size)
            h_t, c_t = cell(x_t, (h_t, c_t))
            outputs.append(h_t)

        # Stack outputs in correct order
        if direction == 'forward':
            output = torch.stack(outputs, dim=1)  # (batch_size, seq_len, hidden_size)
        else:
            output = torch.stack(outputs[::-1], dim=1)  # Reverse for backward

        return h_t, c_t, output
````

Implement Vanilla LSTM using custom LSTM created above




````python
class VanillaLSTM(nn.Module):
    """
    A Vanilla LSTM model for Named Entity Recognition (NER).
    """
    def __init__(self, vocab_size, tagset_size, embedding_dim, hidden_size, num_layers=1, dropout=0.3):
        super().__init__()
        self.vocab_size = vocab_size
        self.tagset_size = tagset_size
        self.embedding_dim = embedding_dim
        self.hidden_size = hidden_size
        self.num_layers = num_layers

        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)

        # LSTM layer
        self.lstm = LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            bidirectional=False,
            num_layers=num_layers,
            dropout=dropout
        )

        # Output layer
        self.hidden2tag = nn.Linear(hidden_size, tagset_size)

        # Dropout
        self.dropout = nn.Dropout(dropout)

        # Initialize weights
        self._init_weights()

    def _init_weights(self):
        """Initialize weights"""
        nn.init.xavier_uniform_(self.embedding.weight)
        nn.init.xavier_uniform_(self.hidden2tag.weight)
        nn.init.constant_(self.hidden2tag.bias, 0)

    def forward(self, sentence):
        """
        Forward pass of the Vanilla LSTM NER model.

        Args:
            sentence: Input sequence of token indices (batch_size, seq_len)

        Returns:
            tag_scores: Output scores for each tag (batch_size, seq_len, tagset_size)
        """
        # Embedding layer
        embeds = self.embedding(sentence)  # (batch_size, seq_len, embedding_dim)
        embeds = self.dropout(embeds)

        # LSTM layer
        lstm_out, _ = self.lstm(embeds)  # (batch_size, seq_len, hidden_size)
        lstm_out = self.dropout(lstm_out)

        # Output layer
        tag_scores = self.hidden2tag(lstm_out)  # (batch_size, seq_len, tagset_size)

        return tag_scores
````

Implement  BiLSTM using custom LSTM created above


````python
class BiLSTM(nn.Module):
    """
    A Bi-directional LSTM model for Named Entity Recognition (NER).
    """
    def __init__(self, vocab_size, tagset_size, embedding_dim, hidden_size, num_layers=1, dropout=0.3):
        super().__init__()
        self.vocab_size = vocab_size
        self.tagset_size = tagset_size
        self.embedding_dim = embedding_dim
        self.hidden_size = hidden_size
        self.num_layers = num_layers

        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)

        # BiLSTM layer
        self.bilstm = LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            bidirectional=True,
            num_layers=num_layers,
            dropout=dropout
        )

        # Output layer (2 * hidden_size because bidirectional)
        self.hidden2tag = nn.Linear(2 * hidden_size, tagset_size)

        # Dropout
        self.dropout = nn.Dropout(dropout)

        # Initialize weights
        self._init_weights()

    def _init_weights(self):
        """Initialize weights"""
        nn.init.xavier_uniform_(self.embedding.weight)
        nn.init.xavier_uniform_(self.hidden2tag.weight)
        nn.init.constant_(self.hidden2tag.bias, 0)

    def forward(self, sentence):
        """
        Forward pass of the BiLSTM NER model.

        Args:
            sentence: Input sequence of token indices (batch_size, seq_len)

        Returns:
            tag_scores: Output scores for each tag (batch_size, seq_len, tagset_size)
        """
        # Embedding layer
        embeds = self.embedding(sentence)  # (batch_size, seq_len, embedding_dim)
        embeds = self.dropout(embeds)

        # BiLSTM layer
        bilstm_out, _ = self.bilstm(embeds)  # (batch_size, seq_len, 2 * hidden_size)
        bilstm_out = self.dropout(bilstm_out)

        # Output layer
        tag_scores = self.hidden2tag(bilstm_out)  # (batch_size, seq_len, tagset_size)

        return tag_scores
````

````python
class MultiLayerBiLSTM(nn.Module):
    """
    A multi-layer Bidirectional LSTM built by stacking LSTM layers.
    """
    def __init__(self, vocab_size, tagset_size, embedding_dim, hidden_size, num_layers, dropout=0.3):
        super().__init__()
        self.vocab_size = vocab_size
        self.tagset_size = tagset_size
        self.embedding_dim = embedding_dim
        self.hidden_size = hidden_size
        self.num_layers = num_layers

        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)

        # Multi-layer BiLSTM
        self.multilayer_bilstm = LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            bidirectional=True,
            num_layers=num_layers,
            dropout=dropout
        )

        # Output layer
        self.hidden2tag = nn.Linear(2 * hidden_size, tagset_size)

        # Dropout
        self.dropout = nn.Dropout(dropout)

        # Initialize weights
        self._init_weights()

    def _init_weights(self):
        """Initialize weights"""
        nn.init.xavier_uniform_(self.embedding.weight)
        nn.init.xavier_uniform_(self.hidden2tag.weight)
        nn.init.constant_(self.hidden2tag.bias, 0)

    def forward(self, sequence):
        """
        Forward pass of the Multi-layer BiLSTM.

        Args:
            sequence: Input tensor of shape (batch_size, seq_len)

        Returns:
            tag_scores: Output tensor of shape (batch_size, seq_len, tagset_size)
        """
        # Embedding layer
        embeds = self.embedding(sequence)  # (batch_size, seq_len, embedding_dim)
        embeds = self.dropout(embeds)

        # Multi-layer BiLSTM
        lstm_out, states = self.multilayer_bilstm(embeds)  # (batch_size, seq_len, 2 * hidden_size)
        lstm_out = self.dropout(lstm_out)

        # Output layer
        tag_scores = self.hidden2tag(lstm_out)  # (batch_size, seq_len, tagset_size)

        return tag_scores
````

````python
## Training and Evaluation Framework

class NERLSTMDataset(Dataset):
    """Dataset for LSTM-based NER models"""
    def __init__(self, sentences, word_to_id, tag_to_id, max_len=100):
        self.sentences = sentences
        self.word_to_id = word_to_id
        self.tag_to_id = tag_to_id
        self.max_len = max_len
        self.id_to_tag = {v: k for k, v in tag_to_id.items()}

    def __len__(self):
        return len(self.sentences)

    def __getitem__(self, idx):
        sentence_data = self.sentences[idx]
        words = sentence_data['words'][:self.max_len]
        tags = sentence_data['tags'][:self.max_len]

        # Pad sequences
        if len(words) < self.max_len:
            pad_len = self.max_len - len(words)
            words = words + [0] * pad_len  # 0 is padding index
            tags = tags + [self.tag_to_id['O']] * pad_len  # Pad with 'O' tag

        return {
            'words': torch.tensor(words, dtype=torch.long),
            'tags': torch.tensor(tags, dtype=torch.long),
            'length': min(len(sentence_data['words']), self.max_len)
        }

def collate_fn_lstm(batch):
    """Custom collate function for LSTM datasets"""
    words = torch.stack([item['words'] for item in batch])
    tags = torch.stack([item['tags'] for item in batch])
    lengths = torch.tensor([item['length'] for item in batch])

    return {
        'words': words,
        'tags': tags,
        'lengths': lengths
    }

class LSTMTrainer:
    """Trainer for LSTM models"""
    def __init__(self, model, device, learning_rate=0.001):
        self.model = model.to(device)
        self.device = device
        self.optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=1e-4)
        self.criterion = nn.CrossEntropyLoss(ignore_index=tag_to_id['O'])  # Ignore 'O' tag in loss

        # Learning rate scheduler
        self.scheduler = optim.lr_scheduler.ReduceLROnPlateau(self.optimizer, patience=3, factor=0.5)

    def train_epoch(self, dataloader):
        """Train for one epoch"""
        self.model.train()
        total_loss = 0
        num_batches = 0

        for batch in tqdm(dataloader, desc="Training"):
            # Move data to device
            words = batch['words'].to(self.device)
            tags = batch['tags'].to(self.device)
            lengths = batch['lengths']

            # Zero gradients
            self.optimizer.zero_grad()

            # Forward pass
            tag_scores = self.model(words)

            # Reshape for loss calculation
            tag_scores_flat = tag_scores.view(-1, tag_scores.shape[-1])
            tags_flat = tags.view(-1)

            # Compute loss
            loss = self.criterion(tag_scores_flat, tags_flat)

            # Backward pass
            loss.backward()

            # Gradient clipping
            torch.nn.utils.clip_grad_norm_(self.model.parameters(), 5.0)

            # Update weights
            self.optimizer.step()

            total_loss += loss.item()
            num_batches += 1

        return total_loss / num_batches

    def evaluate(self, dataloader, id_to_tag):
        """Evaluate model on dataset"""
        self.model.eval()
        all_true_tags = []
        all_pred_tags = []

        with torch.no_grad():
            for batch in tqdm(dataloader, desc="Evaluating"):
                words = batch['words'].to(self.device)
                tags = batch['tags']
                lengths = batch['lengths']

                # Forward pass
                tag_scores = self.model(words)

                # Get predictions
                _, predicted = torch.max(tag_scores, dim=-1)

                # Convert to lists
                for i, length in enumerate(lengths):
                    true_seq = tags[i][:length].tolist()
                    pred_seq = predicted[i][:length].tolist()

                    all_true_tags.extend(true_seq)
                    all_pred_tags.extend(pred_seq)

        # Convert to tag names
        true_tag_names = [id_to_tag[tag_id] for tag_id in all_true_tags]
        pred_tag_names = [id_to_tag[tag_id] for tag_id in all_pred_tags]

        return true_tag_names, pred_tag_names

    def train(self, train_loader, val_loader, id_to_tag, epochs=50):
        """Full training loop"""
        train_losses = []
        val_f1_scores = []

        best_f1 = 0
        patience = 7
        patience_counter = 0

        for epoch in range(epochs):
            # Train
            train_loss = self.train_epoch(train_loader)
            train_losses.append(train_loss)

            # Validate
            true_tags, pred_tags = self.evaluate(val_loader, id_to_tag)

            # Compute metrics
            report = classification_report(true_tags, pred_tags, output_dict=True, zero_division=0)
            val_f1 = report['macro avg']['f1-score']
            val_f1_scores.append(val_f1)

            # Update learning rate
            self.scheduler.step(val_f1)

            current_lr = self.optimizer.param_groups[0]['lr']
            logger.info(f'Epoch {epoch+1}/{epochs}: Train Loss = {train_loss:.4f}, Val F1 = {val_f1:.4f}, LR = {current_lr:.6f}')

            # Early stopping
            if val_f1 > best_f1:
                best_f1 = val_f1
                patience_counter = 0
                # Save best model
                torch.save(self.model.state_dict(), f'best_{self.model.__class__.__name__.lower()}.pth')
            else:
                patience_counter += 1
                if patience_counter >= patience:
                    logger.info(f'Early stopping at epoch {epoch+1}')
                    break

        # Load best model
        self.model.load_state_dict(torch.load(f'best_{self.model.__class__.__name__.lower()}.pth'))

        return train_losses, val_f1_scores

def plot_comparison(results_dict):
    """Plot comparison of different models"""
    plt.figure(figsize=(12, 5))

    # Plot training loss
    plt.subplot(1, 2, 1)
    for model_name, (train_losses, val_f1_scores) in results_dict.items():
        plt.plot(train_losses, label=f'{model_name} Train Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training Loss Comparison')
    plt.legend()
    plt.grid(True, alpha=0.3)

    # Plot validation F1
    plt.subplot(1, 2, 2)
    for model_name, (train_losses, val_f1_scores) in results_dict.items():
        plt.plot(val_f1_scores, label=f'{model_name} Val F1')
    plt.xlabel('Epoch')
    plt.ylabel('F1 Score')
    plt.title('Validation F1 Score Comparison')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
````

### Q1: Which of the above models assumes the Markov assumption in text processing? (20 marks)

- Identify and explain which models rely on the Markov assumption (i.e., that the current state depends only on the previous state).
- What are the limitations of this assumption in NER tasks?

The **Hidden Markov Model (HMM)** and the **Baum-Welch HMM** explicitly rely on the **Markov assumption**, which states that the current tag (state) depends **only on the previous tag**, not on any earlier tags or future context. Formally, this is written as:
$
P(y_t | y_{t-1}, y_{t-2}, \dots, y_1) = P(y_t | y_{t-1})
$
This simplifies sequence modeling and makes probability computation tractable. The **Baum-Welch algorithm** further uses this assumption during its unsupervised parameter estimation.

In contrast, models like **CRF**, **LSTM**, and **BiLSTM** do **not strictly follow the Markov assumption**.

* **CRF** models the *conditional probability* of the entire tag sequence given all observations, allowing it to consider the whole sentence at once.
* **LSTM** and **BiLSTM** go even further by capturing **long-range dependencies** through memory cells, with BiLSTM using both past and future context.

**Limitations of the Markov assumption in NER:**

* It restricts the model to very **short-term dependencies**, ignoring useful context beyond one token.
* It struggles with **multi-word entities** or situations where the correct tag depends on words farther away.
* It often leads to **boundary errors** and misclassification in longer or complex sentences.

Hence, while HMMs are simple and efficient, their reliance on the Markov assumption limits their effectiveness for tasks like NER that require **global and contextual understanding** of the sentence.

### Q2: Which parts of the NLP pipeline assume the Markov assumption for processing?  (20 + 5 + 5 + 5 = 35 marks)

* Explain how this assumption affects:
   * Sequence labeling
   * Transition probabilities (in HMM and CRF)
   * Inference algorithms (Viterbi)
   

The **Markov assumption** is mainly used in the **sequence labeling** part of the NLP pipeline, where models predict a sequence of tags (e.g., NER, POS tagging). It simplifies the tagging process by assuming that **each tag depends only on the previous tag**, not on the entire history of tags or words.

#### **1. Sequence Labeling:**

In models like **HMM** and **Baum-Welch HMM**, sequence labeling relies directly on the Markov assumption. This means the decision for the current tag ( y_t ) depends only on the previous tag ( y_{t-1} ).

* This makes computation efficient but limits the model’s ability to capture **long-range context**.
* As a result, HMMs may mislabel entities that depend on information from distant words in the sentence.

#### **2. Transition Probabilities (in HMM and CRF):**

* In **HMMs**, transition probabilities ( P(y_t | y_{t-1}) ) are defined explicitly under the Markov assumption. These control how likely one tag is to follow another (e.g., `B-ORG → I-ORG`).
* In **CRFs**, although transitions between labels are still modeled, the overall framework is **not limited** by the Markov assumption on observations. CRFs can incorporate **features from the entire input sequence**, but the tag dependencies (transition features) are still **first-order Markov**, i.e., depend on adjacent tags.

#### **3. Inference Algorithms (Viterbi):**

The **Viterbi algorithm**, used in HMMs and CRFs for decoding, is based entirely on the **first-order Markov assumption**. It computes the most probable sequence of tags efficiently by reusing partial results — this is only possible because each state depends **only on the previous one**.

* Without this assumption, inference would become exponentially expensive, as the algorithm would need to consider all possible tag histories.

**In summary:**
The Markov assumption simplifies **sequence labeling**, defines **transition probabilities** in HMM and CRF, and enables efficient **decoding via Viterbi**. However, it limits the ability to model **long-distance dependencies**, which is why deep learning models like **LSTM** and **BiLSTM** often outperform purely Markov-based approaches in NLP tasks such as NER.

### Q8: If the task had to run on noisy text (e.g. social media, OCR, speech transcripts) with spelling mistakes or non-standard spellings, then, (10 + 10 = 20 marks)

* Which models would be more robust?
* What preprocessing or model additions could help?

#### **Which models would be more robust:**

Among all the models — **LSTM** and especially **BiLSTM** would be **most robust** to noisy or misspelled text.

* **Reason:** Neural models learn **distributed representations (embeddings)** of words rather than relying on exact surface forms. Even if a word is slightly misspelled or unseen, its embedding can still be close to similar words, allowing the model to generalize better.
* **CRF** performs moderately well if strong **character-level or morphological features** are used but struggles without them since it relies on exact token matches.
* **HMM (Viterbi / Baum–Welch)** perform **poorly** on noisy data because their emission probabilities are based on exact word-tag co-occurrence counts. Any unseen or misspelled token gets an extremely low or zero probability.

✅ **Robustness ranking (most → least):**

> **BiLSTM > LSTM > CRF > HMM (Baum–Welch) > HMM (Viterbi)**

---

#### **What preprocessing or model additions could help:**

1. **Preprocessing Techniques:**

   * **Spell correction / normalization:** Automatically correct common misspellings and expand abbreviations.
   * **Lowercasing and Unicode normalization:** Reduce surface variability in text.
   * **Noise filtering:** Remove or standardize symbols, emojis, and transliterated tokens (especially in social media or OCR text).

2. **Model-Level Improvements:**

   * **Character-level embeddings:** Capture subword patterns and handle unseen words (e.g., “govt” → “government”).
   * **Subword tokenization (Byte-Pair Encoding or WordPiece):** Helps LSTM and CRF models recognize fragments of rare or misspelled words.
   * **Pretrained embeddings (FastText, BERT, IndicBERT):** These embeddings are robust to noise and can generalize across variations in spelling.
   * **Noise-robust training:** Add synthetic noise during training (e.g., random misspellings or dropped characters) to improve generalization.

✅ **Summary:**
For noisy text, **BiLSTM with character/subword embeddings** performs best because it can learn from **both directions** and generalize beyond exact word forms. HMMs fail on such data due to their dependence on **exact token matches**. Proper preprocessing and the use of **robust embeddings** significantly enhance model performance under noisy conditions.

### Q9: If you only had limited annotated data (say only 10-20% of the training set annotated), how would you modify your modeling approach (semi-supervised learning, transfer learning, etc.)? (10 marks)

**(Give some brief description about the approach you have used.)**

When only a small portion of the data is annotated, the best strategy is to use **semi-supervised and transfer learning approaches** to leverage both labeled and unlabeled data effectively.

#### **1. Semi-Supervised Learning (Using Baum-Welch or Self-Training):**

* Start by training a simple supervised model (e.g., HMM or CRF) on the available 10–20% labeled data.
* Then, use that model to **predict labels on the remaining unlabeled data**.
* Iteratively retrain the model using both true and pseudo-labeled data — this is known as **self-training** or **bootstrapping**.
* Alternatively, the **Baum-Welch algorithm** (unsupervised EM) can initialize HMM parameters from unlabeled data and then be fine-tuned on the small labeled set to improve emission probabilities.

#### **2. Transfer Learning (Using Pretrained Embeddings or Models):**

* Use **pretrained word embeddings** (like FastText, GloVe, or IndicBERT for Hindi) in LSTM/CRF models to transfer linguistic knowledge from large corpora.
* Fine-tune a **pretrained BiLSTM or transformer model (e.g., BERT)** on the small labeled dataset. This dramatically improves performance because the model already encodes general language understanding.

#### **3. Data Augmentation (Optional Enhancement):**

* Generate synthetic training data by **replacing entities** with similar ones (e.g., “दिल्ली” → “मुंबई”), or by **adding noise** to improve robustness.
* Use back-translation or paraphrasing to expand sentence diversity.

✅ **Summary:**

With limited annotated data, a **semi-supervised CRF or BiLSTM model with pretrained embeddings** is most effective. It combines the strengths of unlabeled data (via self-training) and general linguistic knowledge (via transfer learning), leading to much better generalization than purely supervised HMMs or CRFs trained on small datasets.

### Q10: In a setting where inference time is limited (mobile, embedded systems), which model would you choose? (10 marks)

* What and how trade-offs (accuracy vs speed vs memory) will guide your decision?


In a **resource-constrained environment** such as mobile or embedded systems, the best choice would be a **lightweight probabilistic model like HMM (Viterbi)** or a **compressed CRF** model, depending on the accuracy requirements.

#### **Model Choice:**

* **HMM (Viterbi)** is preferred when **speed and memory efficiency** are critical. It has very low computational complexity (**O(T × N²)**) and minimal storage needs for transition and emission probabilities.
* **CRF (linear-chain)** could be chosen if slightly more computation and memory are acceptable, since it offers **better accuracy** by considering full-sequence dependencies while still maintaining reasonable decoding speed.

#### **Trade-offs (Accuracy vs Speed vs Memory):**

| Factor                                   | HMM (Viterbi)      | CRF           | LSTM / BiLSTM |
| ---------------------------------------- | ------------------ | ------------- | ------------- |
| **Speed**                                | ✅ Very fast (best) | ⚙️ Moderate   | ❌ Slow        |
| **Memory Usage**                         | ✅ Very low         | ⚙️ Moderate   | ❌ Very high   |
| **Accuracy (especially for non-O tags)** | ⚙️ Moderate        | ✅ High        | ✅✅ Highest    |
| **Suitability for low-power devices**    | ✅ Excellent        | ⚙️ Acceptable | ❌ Poor        |

#### **Decision Rationale:**

* On mobile/edge devices, **real-time response** and **low memory footprint** are more important than achieving the absolute highest F1 score.
* Hence, a **compact HMM model** or a **feature-pruned CRF** offers the **best trade-off** — fast inference, small model size, and decent accuracy.
* Deep models like **LSTM/BiLSTM**, while accurate, require large memory and computation (matrix multiplications, embeddings), making them unsuitable for embedded deployment without heavy quantization or pruning.

✅ **Conclusion:**
For limited inference time and hardware constraints, choose **HMM (Viterbi)** for maximum speed and minimal resource use, or a **lightweight CRF** if slightly higher accuracy is needed. This balances the trade-off between **speed, memory, and accuracy** effectively for mobile or embedded NER applications.
