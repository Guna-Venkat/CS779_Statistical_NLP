# **Historical & Related Work**

The history of NER is tied to the evolution of NLP itself:

## Rule-based systems (1980s–1990s)
- Early systems relied on **hand-crafted linguistic rules** and **gazetteers** (dictionaries of names).  
- Example: If a word follows *“Mr.”*, it is likely a **Person**.  
- **Limitation:** Not scalable across domains or languages.  

## Statistical Models (1990s–2000s)
- With more annotated data, researchers shifted to **probabilistic models** like:  
  - Hidden Markov Models (HMMs)  
  - Maximum Entropy Models  
  - Conditional Random Fields (CRFs)  
- CRFs became a **gold standard** for NER, as they model label dependencies.  

## Neural Models (2010s–present)
- With the rise of deep learning, **Recurrent Neural Networks (RNNs)**, especially **Long Short-Term Memory (LSTM)** models, improved performance by capturing **contextual dependencies** in text.  
- Later, **BiLSTM-CRF** became the state-of-the-art (combining sequence modeling with structured prediction).  
- Today, **transformers** (BERT, RoBERTa, GPT-based models) dominate NER, but LSTM remains important for **foundational understanding** and **resource-constrained setups**.

# **Why Data Analysis Matters in Classical NLP**

Before deep learning, classical NLP models heavily depended on:

- **Feature engineering** (POS tags, capitalization, prefixes/suffixes).  
- **Statistical assumptions** (Markov property in HMMs, linear dependencies in CRFs).  
- **Data preprocessing** (removing noise, balancing entity types).  

Thus, **data analysis** was crucial:  

- Understanding class distribution helped avoid bias.  
- Identifying rare entity types guided feature design.  
- Detecting errors in annotation improved robustness.  

Even in neural methods, exploratory analysis remains vital for **token distributions, sequence lengths, and vocabulary coverage**.

For creation of a Hindi NER system, there are various linguistic challenges like:

### **No Capitalization:**

Unlike English, Hindi does not have capitalization as a feature, which could have been helpful for performing the NER task.  
For example:  
In English, if we see **Aman**, we can detect it is a name.  
But in Hindi, if it is written as **अमन**, you cannot get any idea from capitalization.

### **Ambiguity:**

Proper nouns in Hindi can be ambiguous as the same word can belong to a different PoS category.  
For example, a common Indian name like **Kiran** can be both a proper noun (used as a person’s name) and a common noun meaning **a ray of light**.

### **Spellings:**

The spelling of some words in Hindi can differ depending on the local region in India.  
For example, the word “Sabzi” is written as **“सब्ज़ी”** or **“सब्जी”**, though both forms denote vegetable.

### **Word Order:**

Languages like Hindi have relatively more flexible word order compared to English, so the same meaning can appear in different forms.  
**Example:**  
**गौतम घर गया। (Gautam went home.)**  
**घर गया गौतम। (Home went Gautam.)**

Both mean the same, but entities **गौतम (Gautam)** (PERSON) and **घर (LOCATION)** shift places, unlike English where word order is fixed.  
In English, the sentence is fixed as “Gautam went home.”  
If we change the order (like “Went home Gautam”), it sounds odd.  
This makes the NER task more challenging as computational approaches cannot be complemented with a pattern of PoS tags or strict word order.

````python
def sentence_word_counts(df, col="tokens"):
    """
    Return a list with the number of words in each sentence.
    """
    return [len(tokens) for tokens in df[col]]

def plot_sentence_length_distribution(counts, title="Sentence Length Distribution"):
    """
    Plot histogram of sentence lengths.
    """
    plt.figure(figsize=(10, 6))
    plt.hist(counts, bins=50, alpha=0.7, color='skyblue', edgecolor='black')
    plt.xlabel('Sentence Length (number of tokens)')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.grid(axis='y', alpha=0.75)
    plt.show()
    
    # Print statistics
    print(f"Mean sentence length: {np.mean(counts):.2f}")
    print(f"Median sentence length: {np.median(counts):.2f}")
    print(f"Std sentence length: {np.std(counts):.2f}")
    print(f"Max sentence length: {np.max(counts)}")
    print(f"Min sentence length: {np.min(counts)}")
````

````python
def word_lengths(df, col="tokens"):
    """
    Return a list with the lengths of all words in the dataset.
    """
    all_lengths = []
    for tokens in df[col]:
        all_lengths.extend([len(token) for token in tokens])
    return all_lengths
````

````python
def plot_word_length_distribution(lengths, title="Word Length Distribution"):
    """
    Plot histogram of word lengths.
    """
    plt.figure(figsize=(10, 6))
    plt.hist(lengths, bins=30, alpha=0.7, color='lightcoral', edgecolor='black')
    plt.xlabel('Word Length (number of characters)')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.grid(axis='y', alpha=0.75)
    plt.show()
    
    # Print statistics
    print(f"Mean word length: {np.mean(lengths):.2f}")
    print(f"Median word length: {np.median(lengths):.2f}")
    print(f"Std word length: {np.std(lengths):.2f}")
    print(f"Max word length: {np.max(lengths)}")
    print(f"Min word length: {np.min(lengths)}")
````

### 1.3 NER Tag Distribution


````python
def extract_ner_tags(df, col="ner_tags"):
    """
    Return a flat list of all NER tags in the dataset.
    """
    all_tags = []
    for tags in df[col]:
        all_tags.extend(tags)
    return all_tags

def plot_ner_tag_distribution(tags, label_list, title="NER Tag Distribution", top_n=None):
    """
    Plot distribution of NER tags (flattened).
    """
    # Convert tag indices to names
    tag_names = [label_list[tag] for tag in tags]
    tag_counts = Counter(tag_names)
    
    if top_n:
        # Get top N most common tags
        most_common = tag_counts.most_common(top_n)
        labels, counts = zip(*most_common)
    else:
        labels, counts = zip(*tag_counts.most_common())
    
    plt.figure(figsize=(12, 6))
    bars = plt.bar(labels, counts, color='lightgreen', alpha=0.7, edgecolor='black')
    plt.xlabel('NER Tags')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', alpha=0.75)
    
    # Add value labels on bars
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                f'{int(height)}', ha='center', va='bottom')
    
    plt.tight_layout()
    plt.show()
````

````python
def tag_popularity(tags, label_list, top_k=5):
    """
    Return most and least frequent tags.
    """
    tag_names = [label_list[tag] for tag in tags]
    tag_counts = Counter(tag_names)
    
    most_common = tag_counts.most_common(top_k)
    least_common = tag_counts.most_common()[-top_k:]
    
    print("Most frequent tags:")
    for tag, count in most_common:
        print(f"  {tag}: {count} ({count/len(tags)*100:.2f}%)")
    
    print("\nLeast frequent tags:")
    for tag, count in least_common:
        print(f"  {tag}: {count} ({count/len(tags)*100:.2f}%)")
    
    return most_common, least_common
````

### 1.4 Top Words per Entity Type


````python
def top_words_for_tag(df, token_col="tokens", tag_col="ner_tags", target_tag_id=None, target_tag_name=None, top_n=10):
    """
    Return the most frequent words assigned a given tag (by index or name).
    """
    if target_tag_name and not target_tag_id:
        target_tag_id = LABEL_LIST.index(target_tag_name)
    
    word_counter = Counter()
    for tokens, tags in zip(df[token_col], df[tag_col]):
        for token, tag in zip(tokens, tags):
            if tag == target_tag_id:
                word_counter[token] += 1
    
    return word_counter.most_common(top_n)
````

````python
def plot_top_words_for_tag(tag_name, top_words, title_prefix="Top Words for"):
    """
    Plot the top words associated with a given tag.
    """
    if not top_words:
        print(f"No words found for tag: {tag_name}")
        return
    
    words, counts = zip(*top_words)
    
    plt.figure(figsize=(10, 6))
    bars = plt.barh(words, counts, color='lightblue', alpha=0.7, edgecolor='black')
    plt.xlabel('Frequency')
    plt.ylabel('Words')
    plt.title(f'{title_prefix} {tag_name}')
    plt.gca().invert_yaxis()  # Display highest frequency at top
    plt.grid(axis='x', alpha=0.75)
    
    # Add value labels on bars
    for bar in bars:
        width = bar.get_width()
        plt.text(width, bar.get_y() + bar.get_height()/2.,
                f'{int(width)}', ha='left', va='center')
    
    plt.tight_layout()
    plt.show()
````

### **1.5 N-Gram Tag Distributions**

N-gram tag frequency distributions at unigram, bigram, and trigram levels.

**(i). Unigram Tag Frequencies:**

Unigram distributions show the overall frequency of individual tags.
  * This distribution typically follows Zipf's law.
  (Do you remember this concept form you past assignments and does the plots you get are following the law?) .
  
  * The unigram frequencies serve as the baseline emission probabilities in HMM models, indicating how likely each word is to be tagged with a specific tag.

    * For example, in English corpora, words like articles and prepositions dominate the frequency distribution, while content words like specific nouns appear less frequently. This pattern reflects the fundamental structure of natural language where a small number of elements are used repeatedly.

**(ii). Bigram Tag Frequencies:**

Bigram tag sequences capture the transition patterns between consecutive tags, which form the core of HMM transition probabilities.


**(iii). Trigram Tag Frequencies:**

Trigram patterns provide deeper contextual understanding by capturing longer-range dependencies between three consecutive tags.

````python
def generate_ngrams(seq, n):
    """
    Yield n-grams from a sequence.
    """
    for i in range(len(seq) - n + 1):
        yield tuple(seq[i:i + n])

def tag_ngram_distribution(df_tags, n, label_list):
    """
    Return n-gram tag frequency Counter.
    """
    ngram_counter = Counter()
    for tags in df_tags:
        tag_names = [label_list[tag] for tag in tags]
        ngrams = list(generate_ngrams(tag_names, n))
        ngram_counter.update(ngrams)
    return ngram_counter
````

````python
def plot_ngram_distribution(counter, title, top_n=10):
    """
    Plot top-N n-grams as bar chart.
    """
    if not counter:
        print(f"No n-grams found for {title}")
        return
    
    most_common = counter.most_common(top_n)
    ngrams, counts = zip(*most_common)
    
    # Convert n-grams to strings for plotting
    ngram_labels = [' '.join(ngram) for ngram in ngrams]
    
    plt.figure(figsize=(12, 6))
    bars = plt.bar(ngram_labels, counts, color='orange', alpha=0.7, edgecolor='black')
    plt.xlabel(f'{len(ngrams[0])}-grams')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', alpha=0.75)
    
    # Add value labels on bars
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                f'{int(height)}', ha='center', va='bottom')
    
    plt.tight_layout()
    plt.show()
````

````python
### 1.6 Execution Block

print("=== Dataset Summary ===")
print(f"Training set size: {len(df_train)} sentences")
print(f"Validation set size: {len(df_val)} sentences")
print(f"Test set size: {len(df_test)} sentences")

# Calculate total tokens
total_tokens_train = sum(len(tokens) for tokens in df_train['tokens'])
total_tokens_val = sum(len(tokens) for tokens in df_val['tokens'])
total_tokens_test = sum(len(tokens) for tokens in df_test['tokens'])

print(f"Total tokens in training set: {total_tokens_train}")
print(f"Total tokens in validation set: {total_tokens_val}")
print(f"Total tokens in test set: {total_tokens_test}")

# 1. Sentence Length Distribution
print("\n=== Sentence Length Analysis (Training Set) ===")
train_sent_lengths = sentence_word_counts(df_train)
plot_sentence_length_distribution(train_sent_lengths, "Training Set - Sentence Length Distribution")

# 2. Word Length Distribution
print("\n=== Word Length Analysis (Training Set) ===")
train_word_lengths = word_lengths(df_train)
plot_word_length_distribution(train_word_lengths, "Training Set - Word Length Distribution")

# 3. Tag Distribution
print("\n=== NER Tag Distribution (Training Set) ===")
train_tags = extract_ner_tags(df_train)
plot_ner_tag_distribution(train_tags, LABEL_LIST, "Training Set - NER Tag Distribution", top_n=15)

# Most/Least Frequent Tags
print("\n=== Tag Popularity (Training Set) ===")
most_common, least_common = tag_popularity(train_tags, LABEL_LIST, top_k=5)
print(f"Most Common Tag: {most_common}")
print(f"Least Common Tag: {least_common}")

# 4. Top Words per Tag
print("\n=== Top Words per Entity Type (Training Set) ===")
# Analyze top words for some important entity types
important_tags = ['B-PERSON', 'B-LOCATION', 'B-ORGANIZATION', 'B-TIMEX', 'O']

for tag in important_tags:
    print(f"\n--- {tag} ---")
    top_words = top_words_for_tag(df_train, target_tag_name=tag, top_n=10)
    if top_words:
        print("Top words:")
        for word, count in top_words:
            print(f"  {word}: {count}")
        plot_top_words_for_tag(tag, top_words)

# 5. Tag N-Gram Distributions
print("\n=== N-Gram Tag Distributions (Training Set) ===")

# Unigram
print("\n--- Unigram Tag Frequencies ---")
unigram_counter = tag_ngram_distribution(df_train['ner_tags'], 1, LABEL_LIST)
plot_ngram_distribution(unigram_counter, "Top Unigram Tag Patterns", top_n=15)

# Check Zipf's Law for unigrams
print("\n--- Zipf's Law Analysis for Unigrams ---")
unigram_freqs = [freq for _, freq in unigram_counter.most_common()]
ranks = range(1, len(unigram_freqs) + 1)

plt.figure(figsize=(10, 6))
plt.loglog(ranks, unigram_freqs, 'bo-', alpha=0.7)
plt.xlabel('Rank (log)')
plt.ylabel('Frequency (log)')
plt.title("Zipf's Law: Rank vs Frequency for NER Tags")
plt.grid(True, alpha=0.3)
plt.show()

# Bigram
print("\n--- Bigram Tag Frequencies ---")
bigram_counter = tag_ngram_distribution(df_train['ner_tags'], 2, LABEL_LIST)
plot_ngram_distribution(bigram_counter, "Top Bigram Tag Patterns", top_n=15)

# Trigram
print("\n--- Trigram Tag Frequencies ---")
trigram_counter = tag_ngram_distribution(df_train['ner_tags'], 3, LABEL_LIST)
plot_ngram_distribution(trigram_counter, "Top Trigram Tag Patterns", top_n=15)

# Additional analysis: Entity length distribution
print("\n=== Entity Length Analysis ===")
def analyze_entity_lengths(df, label_list):
    entity_lengths = defaultdict(list)
    
    for tags in df['ner_tags']:
        tag_names = [label_list[tag] for tag in tags]
        
        current_entity = None
        current_length = 0
        
        for tag in tag_names:
            if tag.startswith('B-'):
                # Save previous entity
                if current_entity and current_length > 0:
                    entity_lengths[current_entity].append(current_length)
                # Start new entity
                current_entity = tag[2:]  # Remove 'B-' prefix
                current_length = 1
            elif tag.startswith('I-') and current_entity == tag[2:]:
                # Continue current entity
                current_length += 1
            else:
                # Save previous entity and reset
                if current_entity and current_length > 0:
                    entity_lengths[current_entity].append(current_length)
                current_entity = None
                current_length = 0
        
        # Don't forget the last entity in the sequence
        if current_entity and current_length > 0:
            entity_lengths[current_entity].append(current_length)
    
    return entity_lengths

entity_lengths = analyze_entity_lengths(df_train, LABEL_LIST)

print("Average entity lengths by type:")
for entity_type, lengths in entity_lengths.items():
    if lengths:  # Only show if we have entities of this type
        avg_length = np.mean(lengths)
        print(f"  {entity_type}: {avg_length:.2f} (based on {len(lengths)} entities)")

print("\n=== Analysis Complete ===")
````

````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter, defaultdict
import os

def load_file(file_path: str):
    """
    Parses a CoNLL file.
    Returns a list of sentences, where each sentence is a list of (token, tag) tuples.
    """
    sentences = []
    current_sentence = []
    
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:  # Empty line indicates sentence boundary
                if current_sentence:
                    sentences.append(current_sentence)
                    current_sentence = []
            else:
                parts = line.split()
                if len(parts) >= 2:
                    token = parts[0]
                    tag = parts[1]
                    current_sentence.append((token, tag))
        
        # Add the last sentence if file doesn't end with newline
        if current_sentence:
            sentences.append(current_sentence)
    
    return sentences

def create_mappings(sentences):
    """
    Creates word-to-id and tag-to-id and vice-versa mappings from the data.
    """
    # Count word frequencies
    word_freq = Counter()
    tag_set = set()
    
    for sentence in sentences:
        for token, tag in sentence:
            word_freq[token] += 1
            tag_set.add(tag)
    
    # Sort words by frequency (descending)
    sorted_words = [word for word, count in word_freq.most_common()]
    
    # Create word mappings - UNK at index 0
    word_to_id = {'<UNK>': 0}
    id_to_word = {0: '<UNK>'}
    
    for idx, word in enumerate(sorted_words, 1):  # Start from 1 since 0 is UNK
        word_to_id[word] = idx
        id_to_word[idx] = word
    
    # Create tag mappings
    tag_list = sorted(list(tag_set))  # Sort for consistent ordering
    tag_to_id = {}
    id_to_tag = {}
    
    for idx, tag in enumerate(tag_list):
        tag_to_id[tag] = idx
        id_to_tag[idx] = tag
    
    return word_to_id, id_to_word, tag_to_id, id_to_tag

def prepare_corpus(sentences, word_to_id, tag_to_id):
    """
    Converts sentences into a list of dictionaries with wordids and tagids
    """
    corpus = []
    
    for sentence in sentences:
        word_ids = []
        tag_ids = []
        
        for token, tag in sentence:
            # Map token to ID, use UNK if not found
            word_id = word_to_id.get(token, word_to_id['<UNK>'])
            tag_id = tag_to_id[tag]
            
            word_ids.append(word_id)
            tag_ids.append(tag_id)
        
        corpus.append({
            'words': word_ids,
            'tags': tag_ids
        })
    
    return corpus
````

### **Task 2: HMM Training (20 Marks)**
---

With the data prepared, you'll now train the Hidden Markov Model. This involves calculating its three core probability distributions from the training corpus using **Maximum Likelihood Estimation**

A **Hidden Markov Model (HMM)** is a statistical model used to analyze sequences of data. It assumes that the system you are observing has a sequence of **hidden states** that you cannot see, which in turn generate a sequence of **observations** that you *can* see.

Think of it like trying to guess the weather (**hidden state**: Sunny or Rainy) just by observing whether someone carries an umbrella (**observation**: Yes or No). The HMM learns two key things:
1.  The probability of the weather changing (e.g., from Sunny to Rainy).
2.  The probability of seeing someone with an umbrella given the weather (e.g., high for Rainy, low for Sunny).


In NER, the goal is to label each word in a sentence with a tag like `PERSON`, `LOCATION`, or `O`. HMM is a natural fit for this task.

* **Observations**: The words in the sentence (e.g., "Ram", "went", "to", "Delhi").
* **Hidden States**: The sequence of NER tags you want to predict (e.g., `B-PER`, `O`, `O`, `B-LOC`).


The HMM is trained on a labeled corpus to learn pi, A, and O which will be defined shortly.

Once trained, the HMM uses the **Viterbi algorithm** to find the most probable sequence of hidden states (NER tags) for a new, unseen sentence (the observations).

#### **Conceptual & Mathematical Details**

An HMM is defined by the following three probability distributions (pi, A, O):

**1. Initial Probability Vector ($\pi$)**:
This vector stores the probability of a sentence starting with a particular NER tag. The $i$-th element of $\pi$, denoted $\pi_i$, is the probability that the first word of a sentence is assigned `tag_i`.

$$\pi_i = P(\text{tag}_i \text{ at start}) = \frac{\text{Count}(\text{sentences starting with tag}_i)}{\text{Total number of sentences}}$$

**2. Transition Probability Matrix ($A$)**:
This matrix captures the relationship between consecutive NER tags. An element $A_{ij}$ stores the probability of transitioning from `tag_i` to `tag_j`. This is crucial for learning patterns like `B-PER` is often followed by `I-PER`.

$$A_{ij} = P(\text{tag}_j | \text{tag}_i) = \frac{\text{Count}(\text{tag}_i \text{ is followed by tag}_j)}{\text{Count}(\text{tag}_i)}$$

**3. Emission Probability Matrix ($O$)**:
This matrix stores the probability of a given NER tag "emitting" a specific word. An element $O_{jk}$ stores the probability that `tag_j` is associated with `word_k`. For example, the word "दिल्ली" will have a high probability of being emitted by the `B-LOC` tag.

$$O_{jk} = P(\text{word}_k | \text{tag}_j) = \frac{\text{Count}(\text{word}_k \text{ appears with tag}_j)}{\text{Count}(\text{tag}_j)}$$

2.  Iterate through each sentence in the `train_corpus` to populate these count arrays:
    * **Initial ($\pi$):** For each sentence, increment the count for its starting tag. (5 marks)
    * **Transition ($A$):** For each adjacent pair of tags `(tag_i, tag_j)`, increment the count for that transition. (5 marks)
    * **Emission ($O$):** For each `(word_id, tag_id)` pair, increment the count for that specific emission. (5 marks)
3.  After counting is complete, **normalize** the arrays to convert them into probability distributions. For example, to get the final `transition_prob` matrix, you'll divide each row in `transition_counts` by its sum. (2 marks)

````python
def train_hmm(train_corpus, num_tags, num_words):
    """
    Trains a Hidden Markov Model by computing the initial state vector pi,
    transition probability matrix A, and emission probability matrix O
    from a training corpus.
    """
    # Initialize count arrays
    initial_counts = np.zeros(num_tags)
    transition_counts = np.zeros((num_tags, num_tags))
    emission_counts = np.zeros((num_tags, num_words))
    
    # Count occurrences
    for sentence in train_corpus:
        tags = sentence['tags']
        words = sentence['words']
        
        # Initial counts - first tag of sentence
        initial_counts[tags[0]] += 1
        
        # Transition counts - pairs of consecutive tags
        for i in range(len(tags) - 1):
            current_tag = tags[i]
            next_tag = tags[i + 1]
            transition_counts[current_tag, next_tag] += 1
        
        # Emission counts - word-tag pairs
        for word, tag in zip(words, tags):
            emission_counts[tag, word] += 1
    
    # Normalize to get probabilities
    # Initial probabilities
    total_sentences = len(train_corpus)
    initial_prob = initial_counts / total_sentences
    
    # Transition probabilities
    transition_prob = np.zeros_like(transition_counts)
    for i in range(num_tags):
        row_sum = np.sum(transition_counts[i])
        if row_sum > 0:
            transition_prob[i] = transition_counts[i] / row_sum
        else:
            transition_prob[i] = 1.0 / num_tags  # Uniform distribution if no transitions
    
    # Emission probabilities
    emission_prob = np.zeros_like(emission_counts)
    for i in range(num_tags):
        row_sum = np.sum(emission_counts[i])
        if row_sum > 0:
            emission_prob[i] = emission_counts[i] / row_sum
        else:
            emission_prob[i] = 1.0 / num_words  # Uniform distribution if no emissions
    
    return initial_prob, transition_prob, emission_prob
````

### **Task 3: HMM Training (with Laplace Smoothing) (10 Marks)**
---

A model trained on simple counts suffers from the **zero-probability problem**: it assumes events not seen in training are impossible. For example, if it only saw "Sachin" tagged as a Person (`B-PER`), it would assign a zero probability to "Rahul" also being a Person, causing it to fail on new data.

**Laplace smoothing** (or add-1 smoothing) fixes this.

* **The Idea**: We pretend to have seen every possible outcome one extra time by **adding one** to all our counts before calculating probabilities.

* **Why it Helps**: This ensures no probability is ever exactly zero. The probability of an unseen event like `"Rahul" | B-PER` becomes very small instead of impossible, making the model more **robust** and allowing it to **generalize** to new data.


#### **The Algorithmic Difference**
This idea is implemented by adding 1 to our raw counts in the numerator and adding the number of possible outcomes to the denominator. Here's how it formally changes our calculations:

* **Initial Probability ($\pi$) with Smoothing:**
    $$
    \pi_i = \frac{\text{Count(sentences starting with tag\_i)} + 1}{\text{Total number of sentences} + N_{\text{tags}}}
    $$
    (Where $N_{\text{tags}}$ is the total number of unique tags).

* **Transition Probability ($A$) with Smoothing:**
    $$
    A_{ij} = P(\text{tag\_j} \mid \text{tag\_i}) = \frac{\text{Count(tag\_i} \rightarrow \text{tag\_j}) + 1}{\text{Count(tag\_i)} + N_{\text{tags}}}
    $$
    (Where $N_{\text{tags}}$ is the total number of unique tags).

* **Emission Probability ($O$) with Smoothing:**
    $$
    O_{jk} = P(\text{word\_k} \mid \text{tag\_j}) = \frac{\text{Count}(\text{word\_k}, \text{tag\_j}) + 1}{\text{Count}(\text{tag\_j}) + V}
    $$
    (Where $V$ is the total number of unique words in the vocabulary).

````python
# Task 3: HMM Training (with Laplace smoothing)

def train_hmm_laplace(train_corpus, num_tags, num_words):
    """
    Trains a Hidden Markov Model with Laplace smoothing.
    """
    # Initialize count arrays (same as before)
    initial_counts = np.zeros(num_tags)
    transition_counts = np.zeros((num_tags, num_tags))
    emission_counts = np.zeros((num_tags, num_words))
    
    # Count occurrences (same as before)
    for sentence in train_corpus:
        tags = sentence['tags']
        words = sentence['words']
        
        initial_counts[tags[0]] += 1
        
        for i in range(len(tags) - 1):
            current_tag = tags[i]
            next_tag = tags[i + 1]
            transition_counts[current_tag, next_tag] += 1
        
        for word, tag in zip(words, tags):
            emission_counts[tag, word] += 1
    
    # Apply Laplace smoothing by adding 1 to all counts
    initial_counts_smoothed = initial_counts + 1
    transition_counts_smoothed = transition_counts + 1
    emission_counts_smoothed = emission_counts + 1
    
    # Normalize smoothed counts
    # Initial probabilities
    total_sentences = len(train_corpus)
    initial_prob = initial_counts_smoothed / (total_sentences + num_tags)
    
    # Transition probabilities
    transition_prob = np.zeros_like(transition_counts_smoothed)
    for i in range(num_tags):
        row_sum = np.sum(transition_counts_smoothed[i])
        transition_prob[i] = transition_counts_smoothed[i] / row_sum
    
    # Emission probabilities
    emission_prob = np.zeros_like(emission_counts_smoothed)
    for i in range(num_tags):
        row_sum = np.sum(emission_counts_smoothed[i])
        emission_prob[i] = emission_counts_smoothed[i] / row_sum
    
    return initial_prob, transition_prob, emission_prob
````

### **Task 4: Viterbi Decoding (15 Marks)**
---

Now that you have your trained HMM parameters (pi, A, O), the final step is to use them to predict the most likely tag sequence for a new sentence. This process is known as **decoding**.


#### **Decoding Approaches**

**1. Brute-Force Decoding**
The most obvious approach is to generate every possible tag sequence for the sentence, calculate the probability of each one, and pick the sequence with the highest score.

* **Problem**: For a sentence with **T** words and **N** possible tags, there are **N<sup>T</sup>** possible sequences.
* **Time Complexity**: This method has an **exponential** time complexity of **O(N<sup>T</sup>)**, which is computationally infeasible.

**2. Viterbi Decoding**
To decode efficiently, we use the **Viterbi algorithm**. It is a **dynamic programming** approach that cleverly finds the single best path without having to check every possibility.

* **Time Complexity**: Viterbi has a **polynomial** time complexity of **O(T \* N<sup>2</sup>)**, making it dramatically faster and the standard method for HMM decoding.


#### **The Viterbi Algorithm in Detail**

The algorithm works by moving through the sentence one word at a time, building two matrices to keep track of the best path found so far.

![viterbi](https://i.postimg.cc/Dyq4KdqD/Screenshot-2025-09-15-at-12-15-52-AM.png)
(source: https://princeton-nlp.github.io/cos484/lectures/L6_sp25.pdf)
(for the derivation and further intuition, please check the above references.)

**The `M` and `B` Matrices**

* **Probability Matrix (`M`)**: `M[i, j]` stores the probability of the **most probable sequence of tags ending with tag `j` at the `i`-th token**. It essentially keeps track of the "score" of the best path to each tag at each position.
* **Backpointer Matrix (`B`)**: `B[i, j]` is the **tag of the `(i-1)`-th token in the most probable sequence ending with tag `j` at the `i`-th token**. It acts like a trail of breadcrumbs, remembering the path taken to reach each state, which is crucial for reconstruction.

**The Process**

The algorithm proceeds in three main steps:

![algo](https://i.postimg.cc/VkCLm5mH/Screenshot-2025-09-16-at-4-54-46-PM.png)
(source: https://web.stanford.edu/~jurafsky/slp3/A.pdf)

**1. Initialization**
For the first word (at index `i=0`), we calculate the initial probability for each possible tag `j`. This is the product of the start probability and the emission probability of the first word given that tag.
<!-- $$
M[0, j] = \pi_j \times \text{emission_prob}[j, \text{word}_0]
$$ -->
$$
M[0, j] = \pi_j \times \text{O}[j, \text{word}_0]
$$

**2. Recursion**
For each subsequent word `i` and each possible tag `j`, we find the path from the previous step that has the highest probability of leading to it. We consider all possible previous tags `k`. We store the maximum probability in `M` and the "winning" previous tag's index in `B`.
<!-- $$
M[i, j] = \max_{k} (M[i-1, k] \times \text{transition_prob}[k, j]) \times \text{emission_prob}[j, \text{word}_i]
$$ -->
$$
M[i, j] = \max_{k} (M[i-1, k] \times \text{A}[k, j]) \times \text{O}[j, \text{word}_i]
$$

<!-- $$
B[i, j] = \arg\max_{k} (M[i-1, k] \times \text{transition_prob}[k, j]) \times \text{emission_prob}[j, \text{word}_i]
$$ -->

$$
B[i, j] = \arg\max_{k} (M[i-1, k] \times \text{A}[k, j]) \times \text{O}[j, \text{word}_i]
$$

**3. Backtracking**
After filling the matrices, the algorithm finishes by:
* Finding the highest probability in the **last column** of the `M` matrix to identify the final tag of the best sequence.
* Using the **`B` matrix** to trace the path backward from this final tag to the beginning, revealing the complete optimal sequence.

        transition_prob (np.array): The transition probability matrix (A).

        list: The sequence of predicted tag IDs.
    """

* State matrix M (size sentence_length x num_tags) and the backpointer matrix B (same size as M).
* **Process**: Implement the three phases—Initialization (5 marks), Recursion (5 marks), and Backtracking(5 marks)—as described by the formulas above.
* **Return**: A list of the predicted tag IDs for the input sentence.

````python
# Task 4: Viterbi Decoding

def viterbi_decode(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags):
    """
    Decodes the most likely sequence of tags for a sentence using the Viterbi algorithm.
    """
    T = len(sentence_word_ids)  # Length of sentence
    if T == 0:
        return []
    
    # Initialize matrices
    M = np.zeros((T, num_tags))  # Probability matrix
    B = np.zeros((T, num_tags), dtype=int)  # Backpointer matrix
    
    # 1. Initialization
    first_word = sentence_word_ids[0]
    for j in range(num_tags):
        M[0, j] = initial_prob[j] * emission_prob[j, first_word]
        B[0, j] = -1  # No previous state for first word
    
    # 2. Recursion
    for t in range(1, T):
        word_id = sentence_word_ids[t]
        for j in range(num_tags):
            # Find the maximum probability and the best previous state
            max_prob = -1
            best_prev = -1
            for k in range(num_tags):
                prob = M[t-1, k] * transition_prob[k, j] * emission_prob[j, word_id]
                if prob > max_prob:
                    max_prob = prob
                    best_prev = k
            
            M[t, j] = max_prob
            B[t, j] = best_prev
    
    # 3. Backtracking
    # Find the best final state
    best_path = []
    best_final_state = np.argmax(M[T-1, :])
    best_path.append(best_final_state)
    
    # Trace back through the backpointers
    current_state = best_final_state
    for t in range(T-1, 0, -1):
        current_state = B[t, current_state]
        best_path.insert(0, current_state)
    
    return best_path
````

````python
# Task 5: Model Evaluation

def get_confusion_matrix(y_true, y_pred, labels):
    """
    Computes confusion matrix from true and predicted labels.
    """
    num_labels = len(labels)
    cm = np.zeros((num_labels, num_labels), dtype=int)
    
    for true, pred in zip(y_true, y_pred):
        cm[true, pred] += 1
    
    return cm

def get_classification_report(y_true, y_pred, id_to_tag):
    """
    Computes detailed classification report including precision, recall, F1-score.
    """
    num_labels = len(id_to_tag)
    labels = list(range(num_labels))
    
    # Get confusion matrix
    cm = get_confusion_matrix(y_true, y_pred, labels)
    
    # Calculate metrics per class
    precision_per_class = []
    recall_per_class = []
    f1_per_class = []
    support_per_class = []
    
    for i in range(num_labels):
        # True positives: diagonal element
        tp = cm[i, i]
        
        # False positives: sum of column i minus tp
        fp = np.sum(cm[:, i]) - tp
        
        # False negatives: sum of row i minus tp
        fn = np.sum(cm[i, :]) - tp
        
        # True negatives: all other elements
        tn = np.sum(cm) - tp - fp - fn
        
        # Support (number of true instances)
        support = np.sum(cm[i, :])
        support_per_class.append(support)
        
        # Precision
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
        precision_per_class.append(precision)
        
        # Recall
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
        recall_per_class.append(recall)
        
        # F1-score
        if precision + recall > 0:
            f1 = 2 * precision * recall / (precision + recall)
        else:
            f1 = 0.0
        f1_per_class.append(f1)
    
    # Overall accuracy
    accuracy = np.sum(np.diag(cm)) / np.sum(cm)
    
    # Calculate averages
    # Macro average (simple mean of per-class metrics)
    macro_precision = np.mean([p for p, s in zip(precision_per_class, support_per_class) if s > 0])
    macro_recall = np.mean([r for r, s in zip(recall_per_class, support_per_class) if s > 0])
    macro_f1 = np.mean([f for f, s in zip(f1_per_class, support_per_class) if s > 0])
    
    # Micro average (calculate from overall TP, FP, FN)
    total_tp = np.sum(np.diag(cm))
    total_fp = np.sum(cm) - total_tp  # Simplified for micro average
    total_fn = total_fp  # In multi-class single-label, FP = FN for micro
    
    micro_precision = total_tp / (total_tp + total_fp) if (total_tp + total_fp) > 0 else 0.0
    micro_recall = total_tp / (total_tp + total_fn) if (total_tp + total_fn) > 0 else 0.0
    micro_f1 = 2 * micro_precision * micro_recall / (micro_precision + micro_recall) if (micro_precision + micro_recall) > 0 else 0.0
    
    # Weighted average (weight by support)
    total_support = np.sum(support_per_class)
    weighted_precision = np.sum([p * s for p, s in zip(precision_per_class, support_per_class)]) / total_support
    weighted_recall = np.sum([r * s for r, s in zip(recall_per_class, support_per_class)]) / total_support
    weighted_f1 = np.sum([f * s for f, s in zip(f1_per_class, support_per_class)]) / total_support
    
    # Format report
    report = []
    report.append("{:15s} {:>10s} {:>10s} {:>10s} {:>10s}".format(
        "Tag", "Precision", "Recall", "F1-Score", "Support"))
    report.append("-" * 65)
    
    for i in range(num_labels):
        if support_per_class[i] > 0:  # Only show tags with support
            report.append("{:15s} {:10.4f} {:10.4f} {:10.4f} {:10d}".format(
                id_to_tag[i], precision_per_class[i], recall_per_class[i], 
                f1_per_class[i], support_per_class[i]))
    
    report.append("-" * 65)
    report.append("{:15s} {:10.4f} {:10.4f} {:10.4f} {:10d}".format(
        "accuracy", accuracy, accuracy, accuracy, total_support))
    report.append("{:15s} {:10.4f} {:10.4f} {:10.4f} {:10d}".format(
        "macro avg", macro_precision, macro_recall, macro_f1, total_support))
    report.append("{:15s} {:10.4f} {:10.4f} {:10.4f} {:10d}".format(
        "weighted avg", weighted_precision, weighted_recall, weighted_f1, total_support))
    
    return "\n".join(report), cm

def evaluate_model(corpus, initial_prob, transition_prob, emission_prob, num_tags, id_to_tag, dataset_name=""):
    """
    Evaluates HMM model on given corpus and prints classification report.
    """
    y_true = []
    y_pred = []
    
    for sentence_data in corpus:
        words = sentence_data['words']
        true_tags = sentence_data['tags']
        
        # Predict tags using Viterbi
        predicted_tags = viterbi_decode(words, initial_prob, transition_prob, emission_prob, num_tags)
        
        y_true.extend(true_tags)
        y_pred.extend(predicted_tags)
    
    # Get classification report
    report, cm = get_classification_report(y_true, y_pred, id_to_tag)
    
    print(f"\n=== Evaluation on {dataset_name} ===")
    print(report)
    
    # Plot confusion matrix
    plt.figure(figsize=(12, 10))
    tag_names = [id_to_tag[i] for i in range(num_tags)]
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=tag_names, yticklabels=tag_names)
    plt.title(f'Confusion Matrix - {dataset_name}')
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.xticks(rotation=45)
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    
    return y_true, y_pred

````

### **Task 6: ROC AUC Curve Evaluation (20 Marks)**

While metrics like F1-score evaluate the quality of the single best tag sequence predicted by the Viterbi algorithm, they don't tell the whole story. They are based on "hard" decisions (this word *is* `B-PER`, this one *is not*). A **Receiver Operating Characteristic (ROC)** curve provides a more nuanced evaluation by showing how well the model can distinguish between classes across all possible decision thresholds.

---

#### **Conceptual Background**

* **What is an ROC Curve?** An ROC curve plots the **True Positive Rate (TPR)** against the **False Positive Rate (FPR)** at various threshold settings.
    * **TPR (Recall):** $TPR = \frac{TP}{TP + FN}$. It answers: "Of all the actual positive samples, how many did we correctly identify?"
    * **FPR:** $FPR = \frac{FP}{FP + TN}$. It answers: "Of all the actual negative samples, how many did we incorrectly identify as positive?"
    * A model that is good at discriminating will have a curve that bows towards the top-left corner, indicating a high TPR for a low FPR. A random model follows the diagonal line (y=x).

* **Area Under the Curve (AUC):** The AUC is the area under the ROC curve. It provides a single number to summarize the model's performance.
    * **AUC = 1.0:** Perfect classifier.
    * **AUC = 0.5:** Random classifier (no discriminative ability).
    * **AUC < 0.5:** Worse than random.



#### **Why is the Forward-Backward Algorithm Needed?**

To plot an ROC curve, we need a continuous **score** or probability for each class, not just a single, final prediction.
* The **Viterbi algorithm** (Task 4) is a "hard-decision" algorithm. It finds the single most likely sequence of tags. It doesn't tell you the probability of a specific tag at a specific position, considering the *entire* sentence.


* The **Forward–Backward algorithm**, in contrast, can compute the **posterior probability**, which is exactly what we need. It calculates the probability of a tag at a specific time step \(t\), given the *entire* observation sequence. This probability, denoted as \(\gamma_t(j)\), is calculated as:

$$
\gamma_t(j)
= P\bigl(\text{tag}_t = j \mid \text{all words in sentence}\bigr)
= \frac{\alpha_t(j)\,\beta_t(j)}{\sum_{i=1}^{N} \alpha_T(i)}
$$

Where \(\alpha_t(j)\) is the forward probability, \(\beta_t(j)\) is the backward probability,  
and the denominator is the total probability of the observed sequence.  

This posterior probability serves as the perfect score to generate the ROC curve. By varying the threshold against this score (e.g., “predict `B-PER` if its posterior probability is > 0.6”), we can compute the TPR and FPR to trace the curve.

````python
# Task 6: ROC AUC Curve Evaluation

def _forward_pass(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags):
    """
    Computes forward probabilities (alpha) for a sentence.
    """
    T = len(sentence_word_ids)
    alpha = np.zeros((T, num_tags))
    
    # Initialization
    first_word = sentence_word_ids[0]
    for j in range(num_tags):
        alpha[0, j] = initial_prob[j] * emission_prob[j, first_word]
    
    # Recursion
    for t in range(1, T):
        word_id = sentence_word_ids[t]
        for j in range(num_tags):
            total = 0.0
            for i in range(num_tags):
                total += alpha[t-1, i] * transition_prob[i, j]
            alpha[t, j] = total * emission_prob[j, word_id]
    
    return alpha

def _backward_pass(sentence_word_ids, transition_prob, emission_prob, num_tags):
    """
    Computes backward probabilities (beta) for a sentence.
    """
    T = len(sentence_word_ids)
    beta = np.zeros((T, num_tags))
    
    # Initialization (end of sequence)
    beta[T-1, :] = 1.0  # All states can be final states
    
    # Recursion (backwards)
    for t in range(T-2, -1, -1):
        next_word = sentence_word_ids[t+1]
        for i in range(num_tags):
            total = 0.0
            for j in range(num_tags):
                total += (transition_prob[i, j] * 
                         emission_prob[j, next_word] * 
                         beta[t+1, j])
            beta[t, i] = total
    
    return beta

def get_posterior_scores(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags):
    """
    Computes posterior probabilities (gamma) for each word in a sentence.
    """
    alpha = _forward_pass(sentence_word_ids, initial_prob, transition_prob, emission_prob, num_tags)
    beta = _backward_pass(sentence_word_ids, transition_prob, emission_prob, num_tags)
    
    T = len(sentence_word_ids)
    gamma = np.zeros((T, num_tags))
    
    # Compute normalization factor (probability of entire sequence)
    # This is sum of alpha at final time step for all states
    P_O = np.sum(alpha[T-1, :])
    
    # Compute posterior probabilities
    for t in range(T):
        for j in range(num_tags):
            gamma[t, j] = (alpha[t, j] * beta[t, j]) / P_O
    
    return gamma

def roc_curve_from_scratch(y_true_binary, y_scores):
    """
    Computes ROC curve from scratch.
    """
    # Sort by scores descending
    sorted_indices = np.argsort(y_scores)[::-1]
    y_true_sorted = y_true_binary[sorted_indices]
    y_scores_sorted = y_scores[sorted_indices]
    
    # Initialize
    tpr = [0.0]
    fpr = [0.0]
    
    # Count positive and negative examples
    n_pos = np.sum(y_true_binary)
    n_neg = len(y_true_binary) - n_pos
    
    # Current counts
    tp = 0
    fp = 0
    
    # Previous score for threshold grouping
    prev_score = None
    
    for i in range(len(y_scores_sorted)):
        # If score changes, add new point to ROC
        if y_scores_sorted[i] != prev_score:
            tpr.append(tp / n_pos if n_pos > 0 else 0.0)
            fpr.append(fp / n_neg if n_neg > 0 else 0.0)
            prev_score = y_scores_sorted[i]
        
        # Update counts
        if y_true_sorted[i] == 1:
            tp += 1
        else:
            fp += 1
    
    # Add final point
    tpr.append(tp / n_pos if n_pos > 0 else 0.0)
    fpr.append(fp / n_neg if n_neg > 0 else 0.0)
    
    return np.array(fpr), np.array(tpr), None

def auc_from_scratch(fpr, tpr):
    """
    Computes AUC using trapezoidal rule.
    """
    # Sort by FPR
    sorted_indices = np.argsort(fpr)
    fpr_sorted = fpr[sorted_indices]
    tpr_sorted = tpr[sorted_indices]
    
    # Calculate area using trapezoidal rule
    auc = 0.0
    for i in range(1, len(fpr_sorted)):
        width = fpr_sorted[i] - fpr_sorted[i-1]
        avg_height = (tpr_sorted[i] + tpr_sorted[i-1]) / 2.0
        auc += width * avg_height
    
    return auc

def plot_one_vs_rest_roc(y_true, y_scores, id_to_tag_map):
    """
    Computes and plots One-vs-Rest ROC curves for each class.
    """
    from sklearn.preprocessing import label_binarize
    
    num_tags = len(id_to_tag_map)
    
    # Binarize labels for one-vs-rest
    y_true_binarized = label_binarize(y_true, classes=range(num_tags))
    
    # Initialize arrays for ROC curves
    fpr = [None] * num_tags
    tpr = [None] * num_tags
    roc_auc = [0.0] * num_tags
    
    # Compute ROC and AUC for each class
    for i in range(num_tags):
        fpr[i], tpr[i], _ = roc_curve_from_scratch(y_true_binarized[:, i], y_scores[:, i])
        roc_auc[i] = auc_from_scratch(fpr[i], tpr[i])
    
    # Plot all ROC curves
    plt.figure(figsize=(10, 8))
    colors = plt.cm.rainbow(np.linspace(0, 1, num_tags))
    
    for i in range(num_tags):
        plt.plot(fpr[i], tpr[i], color=colors[i], lw=2,
                label=f'{id_to_tag_map[i]} (AUC = {roc_auc[i]:.3f})')
    
    # Plot random classifier line
    plt.plot([0, 1], [0, 1], 'k--', lw=2, label='Random Classifier')
    
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('One-vs-Rest ROC Curves')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Print per-tag AUC scores
    print('Per tag AUC scores:')
    per_tag_aucs = {id_to_tag_map[i]: roc_auc[i] for i in range(num_tags)}
    for tag, auc_val in sorted(per_tag_aucs.items(), key=lambda x: x[1], reverse=True):
        print(f'  {tag}: {auc_val:.4f}')
    
    return per_tag_aucs

def evaluate_and_plot_roc(corpus, initial_prob, transition_prob, emission_prob, num_tags, id_to_tag):
    """
    Main function to evaluate and plot ROC curves.
    """
    y_true = []
    y_scores_list = []
    
    for sentence_data in corpus:
        words = sentence_data['words']
        true_tags = sentence_data['tags']
        
        # Get posterior scores
        posterior_scores = get_posterior_scores(words, initial_prob, transition_prob, emission_prob, num_tags)
        
        y_true.extend(true_tags)
        y_scores_list.append(posterior_scores)
    
    # Concatenate all scores
    y_scores = np.vstack(y_scores_list)
    
    # Plot ROC curves
    per_tag_aucs = plot_one_vs_rest_roc(y_true, y_scores, id_to_tag)
    
    return y_true, y_scores, per_tag_aucs
````

## Introduction to the Expectation-Maximization (EM) Algorithm

The **Expectation-Maximization (EM)** algorithm is a general framework for finding maximum likelihood estimates of parameters in probabilistic models when some variables are unobserved or "hidden". It is especially useful when direct computation of the likelihood function is difficult because the data includes latent variables.

**EM works in two main iterative steps:**

- **E-step (Expectation):** Compute an expectation of the log-likelihood with respect to the current estimate of the distribution for the latent variables.
- **M-step (Maximization):** Maximize this expected log-likelihood to update the parameters.

This process is repeated until convergence. EM guarantees that the likelihood will never decrease with each iteration.

---

## Baum-Welch Algorithm: EM for HMMs

The **Baum-Welch algorithm** is a specific case of the EM algorithm for training **Hidden Markov Models (HMMs)** on observation sequences when the "true" hidden states are unknown (in unsupervised/semi-supervised settings).

**Goal:**  
Given observed data sequences (words/events), estimate the optimal HMM parameters [Initial Distribution $(\pi)$, Transition Matrix $(A)$, and Emission Matrix $(B)$] to maximize the likelihood of the observed data.

**Why Baum-Welch?**  
In many practical cases (like speech, text, sequence tagging), we only observe the sequence of emitted symbols (e.g., words), not the underlying sequence of states (e.g., POS tags, NER tags). We want the HMM parameters that best explain the observed data.

---

## The Baum-Welch Algorithm: Step-by-Step

### 1. Initialization

Start with an initial guess for all HMM parameters:

- Initial state probabilities $(\pi)$
- Transition probabilities $(A)$
- Emission probabilities $(B)$

These can be random, uniform, or based on some prior knowledge.

### 2. E-step: Expected Sufficient Statistics

Given the current model parameters, run **Forward-Backward** algorithm over all training sequences:

- Forward: $\alpha_t(i)$ — Compute the probability of observing the partial sequence up to time $t$, ending in state $i$.  
- Backward: $\beta_t(i)$ — Compute the probability of remaining sequence started from state $i$ at time $t$.

From these, compute:

- $\gamma_t(i)$ — expected probability of being in state $i$ at time $t$ given the observed sequence and current parameters.

- $\xi_t(i,j)$ — expected probability of transitioning from state $i$ at $t$ to state $j$ at $t+1$ given the observed sequence and current parameters.

### 3. M-step: Update Model Parameters

Update each HMM parameter to maximize the expected log-likelihood:

* **Initial probabilities $(\pi)$:**

$$ \pi_i = \gamma_1(i) $$

(average probability of starting in state $i$)

* **Transition matrix $(A)$:**

$$
A_{i,j} = \frac{\sum_{t=1}^{T-1} \xi_t(i,j)}{\sum_{t=1}^{T-1} \gamma_t(i)}
$$

(expected number of transitions from $i$ to $j$ divided by expected number in $i$)

* **Emission Matrix $(B)$**:

$$
B_{j,k} = \frac{\sum_{t=1}^{T} \gamma_t(j) \cdot \mathbf{1}[o_t = k]}{\sum_{t=1}^{T} \gamma_t(j)}
$$

(expected number of times state $j$ emitted symbol $k$, normalized)

### 4. Iterate Until Convergence

Repeat the E-step and M-step until the model parameters converge (i.e., the change in likelihood is below a small threshold, or a max number of iterations is reached), until log-likelihood stabilizes.

---

## Formal Statement of the Problem

Given:

- $N$: Number of hidden states (tags)
- $M$: Number of observable symbols (words)
- $O = (o_1, o_2, \dots, o_T)$: Observed sequence  

Estimate HMM parameters $\lambda = (\pi, A, B)$ to maximize:

$$
P(O \mid \lambda)
$$

Where:

- $\pi_i$: Prob. of starting in state $i$  
- $A_{i,j}$: Prob. of transitioning $i \rightarrow j$  
- $B_{j,k}$: Prob. of emitting symbol $k$ in state $j$

````python
import numpy as np
from collections import defaultdict, Counter
import logging

# Set up logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

def initialize_hmm_parameters(train_sents, train_tags=None, num_states=None):
    """
    Initialize HMM parameters (π, A, B) from training data.
    
    Args:
        train_sents: List of word sequences
        train_tags: List of corresponding tag sequences (optional for supervised initialization)
        num_states: Number of hidden states (if None, will be inferred from train_tags)
    
    Returns:
        pi: Initial state probabilities (N,)
        A: Transition matrix (N, N)
        B: Emission matrix (N, V)
        tag2idx: Tag to index mapping
        word2idx: Word to index mapping
    """
    
    # Create vocabulary
    word_freq = Counter()
    for sent in train_sents:
        word_freq.update(sent)
    
    # Create word to index mapping (include UNK token)
    words = ['<UNK>'] + sorted([word for word, count in word_freq.most_common()])
    word2idx = {word: idx for idx, word in enumerate(words)}
    V = len(word2idx)  # Vocabulary size
    
    # Determine number of states and create tag mapping
    if train_tags is not None:
        # Supervised initialization - use provided tags
        tag_set = set()
        for tags in train_tags:
            tag_set.update(tags)
        
        tags = sorted(list(tag_set))
        tag2idx = {tag: idx for idx, tag in enumerate(tags)}
        N = len(tags)
    else:
        # Unsupervised initialization
        if num_states is None:
            # Use a reasonable default based on common NER tags
            N = 12  # Common NER tags: PER, LOC, ORG, MISC, etc. with B- and I- prefixes
        else:
            N = num_states
        tag2idx = {f'STATE_{i}': i for i in range(N)}
    
    logger.info(f"State count N={N}, Vocab size V={V}")
    
    # Initialize parameters
    if train_tags is not None:
        # Supervised initialization using counts
        pi = np.zeros(N)
        A = np.zeros((N, N))
        B = np.zeros((N, V))
        
        # Count occurrences
        for sent_tags, sent_words in zip(train_tags, train_sents):
            if len(sent_tags) == 0:
                continue
                
            # Initial probabilities
            first_tag = sent_tags[0]
            pi[tag2idx[first_tag]] += 1
            
            # Transition probabilities
            for i in range(len(sent_tags) - 1):
                current_tag = sent_tags[i]
                next_tag = sent_tags[i + 1]
                A[tag2idx[current_tag], tag2idx[next_tag]] += 1
            
            # Emission probabilities
            for tag, word in zip(sent_tags, sent_words):
                word_idx = word2idx.get(word, word2idx['<UNK>'])
                B[tag2idx[tag], word_idx] += 1
        
        # Normalize with Laplace smoothing to avoid zeros
        pi = (pi + 1) / (len(train_tags) + N)
        for i in range(N):
            A[i] = (A[i] + 1) / (np.sum(A[i]) + N)
            B[i] = (B[i] + 1) / (np.sum(B[i]) + V)
            
    else:
        # Unsupervised initialization - use uniform or random initialization
        # Initial probabilities - uniform
        pi = np.ones(N) / N
        
        # Transition probabilities - encourage self-transitions slightly
        A = np.random.dirichlet(np.ones(N) * 1.5, size=N)
        
        # Emission probabilities - random but normalized
        B = np.random.dirichlet(np.ones(V) * 0.1, size=N)
    
    logger.info("Initialized π, A, B shapes: %s, %s, %s", pi.shape, A.shape, B.shape)
    logger.info("π sum: %.4f, A row sums: %s", np.sum(pi), [f"{s:.4f}" for s in np.sum(A, axis=1)])
    
    return pi, A, B, tag2idx, word2idx
````

````python
def forward_scaled(obs_seq, pi, A, B):
    """
    Forward algorithm with scaling to prevent numerical underflow.
    
    Args:
        obs_seq: Sequence of observation indices
        pi: Initial state probabilities (N,)
        A: Transition matrix (N, N)
        B: Emission matrix (N, V)
    
    Returns:
        alpha: Scaled forward probabilities (T, N)
        scales: Scaling factors (T,)
        log_prob: Log probability of the sequence
    """
    T = len(obs_seq)
    N = len(pi)
    
    # Initialize alpha matrix and scales array
    alpha = np.zeros((T, N))
    scales = np.zeros(T)
    
    # Initialization step (t=0)
    for i in range(N):
        alpha[0, i] = pi[i] * B[i, obs_seq[0]]
    
    # Scale initial alpha
    scales[0] = 1.0 / np.sum(alpha[0])
    alpha[0] *= scales[0]
    
    # Recursion step
    for t in range(1, T):
        for j in range(N):
            # Sum over previous states
            total = 0.0
            for i in range(N):
                total += alpha[t-1, i] * A[i, j]
            alpha[t, j] = total * B[j, obs_seq[t]]
        
        # Scale current alpha
        scales[t] = 1.0 / np.sum(alpha[t])
        alpha[t] *= scales[t]
    
    # Compute log probability of sequence
    log_prob = -np.sum(np.log(scales))
    
    return alpha, scales, log_prob

def backward_scaled(obs_seq, A, B, scales):
    """
    Backward algorithm using scaling factors from forward pass.
    
    Args:
        obs_seq: Sequence of observation indices
        A: Transition matrix (N, N)
        B: Emission matrix (N, V)
        scales: Scaling factors from forward algorithm (T,)
    
    Returns:
        beta: Scaled backward probabilities (T, N)
    """
    T = len(obs_seq)
    N = A.shape[0]
    
    # Initialize beta matrix
    beta = np.zeros((T, N))
    
    # Initialization (last time step)
    beta[T-1] = scales[T-1]  # Since alpha[T-1] was scaled by scales[T-1]
    
    # Recursion (backwards)
    for t in range(T-2, -1, -1):
        for i in range(N):
            total = 0.0
            for j in range(N):
                total += A[i, j] * B[j, obs_seq[t+1]] * beta[t+1, j]
            beta[t, i] = total * scales[t]
    
    return beta

def compute_gamma(alpha, beta):
    """
    Compute state occupation probabilities gamma.
    
    Args:
        alpha: Forward probabilities (T, N)
        beta: Backward probabilities (T, N)
    
    Returns:
        gamma: State occupation probabilities (T, N)
    """
    # gamma_t(i) = alpha_t(i) * beta_t(i) / P(O|λ)
    # Since we're using scaled versions, alpha_t(i) * beta_t(i) already gives the correct probability
    gamma = alpha * beta
    
    # Normalize (though it should already be normalized due to scaling)
    row_sums = np.sum(gamma, axis=1, keepdims=True)
    gamma = gamma / row_sums
    
    return gamma

def compute_xi(obs_seq, alpha, beta, A, B):
    """
    Compute transition probabilities xi.
    
    Args:
        obs_seq: Sequence of observation indices
        alpha: Forward probabilities (T, N)
        beta: Backward probabilities (T, N)
        A: Transition matrix (N, N)
        B: Emission matrix (N, V)
    
    Returns:
        xi: Transition probabilities (T-1, N, N)
    """
    T = len(obs_seq)
    N = A.shape[0]
    
    xi = np.zeros((T-1, N, N))
    
    for t in range(T-1):
        denominator = 0.0
        
        # Compute denominator for normalization
        for i in range(N):
            for j in range(N):
                denominator += alpha[t, i] * A[i, j] * B[j, obs_seq[t+1]] * beta[t+1, j]
        
        # Avoid division by zero
        if denominator == 0:
            denominator = 1e-10
        
        # Compute xi values
        for i in range(N):
            for j in range(N):
                numerator = alpha[t, i] * A[i, j] * B[j, obs_seq[t+1]] * beta[t+1, j]
                xi[t, i, j] = numerator / denominator
    
    return xi
````

````python
def baum_welch(train_sents, pi, A, B, word2idx, max_iter=50, tol=1e-6):
    """
    Implement the Baum-Welch algorithm for HMM parameter estimation.
    
    Args:
        train_sents: List of training sentences (word sequences)
        pi: Initial state probabilities (N,)
        A: Transition probabilities (N, N)
        B: Emission probabilities (N, V)
        word2idx: Word to index mapping
        max_iter: Maximum iterations
        tol: Convergence tolerance
    
    Returns:
        Updated pi, A, B parameters and log likelihood history
    """
    N = len(pi)
    V = B.shape[1]
    
    # Convert sentences to observation sequences
    obs_sequences = []
    for sent in train_sents:
        obs_seq = [word2idx.get(word, word2idx['<UNK>']) for word in sent]
        obs_sequences.append(obs_seq)
    
    log_likelihood_history = []
    prev_log_likelihood = -np.inf
    
    logger.info("Starting Baum-Welch training with %d sequences", len(obs_sequences))
    
    for iteration in range(max_iter):
        # Initialize accumulators
        pi_num = np.zeros(N)
        A_num = np.zeros((N, N))
        A_den = np.zeros(N)
        B_num = np.zeros((N, V))
        B_den = np.zeros(N)
        
        total_log_likelihood = 0.0
        num_sequences = len(obs_sequences)
        
        # E-step: Compute expected counts
        for obs_seq in obs_sequences:
            T = len(obs_seq)
            if T == 0:
                continue
                
            # Run forward-backward algorithm
            alpha, scales, log_prob = forward_scaled(obs_seq, pi, A, B)
            beta = backward_scaled(obs_seq, A, B, scales)
            
            total_log_likelihood += log_prob
            
            # Compute gamma and xi
            gamma = compute_gamma(alpha, beta)
            xi = compute_xi(obs_seq, alpha, beta, A, B)
            
            # Accumulate statistics
            # For initial probabilities
            pi_num += gamma[0]
            
            # For transition probabilities
            for t in range(T-1):
                for i in range(N):
                    A_den[i] += gamma[t, i]
                    for j in range(N):
                        A_num[i, j] += xi[t, i, j]
            
            # For emission probabilities
            for t in range(T):
                for i in range(N):
                    B_den[i] += gamma[t, i]
                    B_num[i, obs_seq[t]] += gamma[t, i]
        
        # M-step: Update parameters
        # Update initial probabilities
        if np.sum(pi_num) > 0:
            pi_new = pi_num / np.sum(pi_num)
        else:
            pi_new = pi.copy()
        
        # Update transition probabilities
        A_new = np.zeros((N, N))
        for i in range(N):
            if A_den[i] > 0:
                A_new[i] = A_num[i] / A_den[i]
            else:
                A_new[i] = A[i].copy()
        
        # Update emission probabilities
        B_new = np.zeros((N, V))
        for i in range(N):
            if B_den[i] > 0:
                B_new[i] = B_num[i] / B_den[i]
            else:
                B_new[i] = B[i].copy()
        
        # Check for convergence
        avg_log_likelihood = total_log_likelihood / num_sequences
        log_likelihood_history.append(avg_log_likelihood)
        
        likelihood_change = avg_log_likelihood - prev_log_likelihood
        
        logger.info("Iteration %d: Avg Log Likelihood = %.6f, Change = %.6f", 
                   iteration + 1, avg_log_likelihood, likelihood_change)
        
        # Check convergence
        if iteration > 0 and likelihood_change < tol:
            logger.info("Converged after %d iterations", iteration + 1)
            break
        
        # Update parameters for next iteration
        pi = pi_new
        A = A_new
        B = B_new
        prev_log_likelihood = avg_log_likelihood
    
    else:
        logger.info("Reached maximum iterations (%d)", max_iter)
    
    logger.info("Baum-Welch training completed")
    return pi, A, B, log_likelihood_history

def plot_training_history(log_likelihood_history):
    """
    Plot the training log likelihood over iterations.
    """
    import matplotlib.pyplot as plt
    
    plt.figure(figsize=(10, 6))
    plt.plot(log_likelihood_history, 'b-', linewidth=2)
    plt.xlabel('Iteration')
    plt.ylabel('Average Log Likelihood')
    plt.title('Baum-Welch Training Progress')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
````

# NER USING CRF(CONDITIONAL RANDOM FIELD)

## ***Why CRF for NER? — Let understand it with a small story  with the help of main character Chaitanya, TA of this course!!***

### 1) Chaitanya starts with HMM (for NER)

Chaitanya wants to build a Named Entity Recognizer (NER).  
First, he tries **Hidden Markov Model (HMM).**

**Good parts of HMM:**
- Can learn **unsupervised** from raw text (using algorithms like Baum-Welch).  
- Handles variable-length sentences.  
- Can decode fast using the **Viterbi algorithm**.

**Problems he notices:**
- HMM is a **generative model**. It assumes tags generate words independently.  
  → This makes it hard to add **overlapping or arbitrary features**.  
- It mostly looks at **current word + previous tag**, not the full context.  
- It models the **joint probability** $P(X, Y)$ (words + tags together).  
  → But in NER, what we really want is $P(Y \mid X)$ (tags given the words).  
- Because of its independence assumptions, it cannot use rich features like:
  - Capitalization (*Delhi* vs *delhi*)  
  - Prefixes/suffixes (*-pur*, *-nagar*)  
  - Gazetteers (lists of names, places, etc.)

**Chaitanya’s thought:**  
*"NER needs context. HMM feels too blind and restricted."*

 Note: In **supervised settings**, HMM parameters can be learned by counting.  
But its real strength is in **unsupervised learning**, unlike MEMM/CRF.

### 2) Chaitanya tries MEMM (for NER)

After struggling with HMM, Chaitanya thinks:  
*"What if I could use more information about the words, not just fixed probabilities?"*  
That’s when he discovers **Maximum Entropy Markov Model (MEMM).**

**What is MEMM?**
- Imagine HMM is a person who only checks the **previous tag + current word** before guessing the next tag.  
- MEMM is a smarter person who says:  
  *“Why not use more clues? Like capitalization, suffixes, and surrounding words.”*  
- MEMM is a **discriminative model**.  
- It replaces HMM’s fixed probabilities with a **maximum entropy classifier** (like logistic regression).  
- This classifier predicts the next tag using **features of the current word and its context**.

**Why MEMM felt better:**
- With MEMM, Chaitanya could finally use rich signals:
  - If a word starts with a capital → maybe a **PERSON**.  
  - If a word ends with “-pur” → maybe a **LOCATION**.  
  - If the word before is *Mr.* → next word likely a **PERSON**.  
- MEMM directly models $P(Y \mid X)$, not $P(X, Y)$.  
- Training is efficient — each local classifier is trained independently.  

**Chaitanya felt relieved:**  
*"Finally! My NER system can use context and features, not just blind probabilities."*

### 3) But MEMM has a hidden problem — Label Bias

Chaitanya was happy with MEMM… until he saw a big issue: **label bias.**

**What is Label Bias?**
- Suppose MEMM is tagging: *"Robbin flies to the bank."*  
  - *flies* can be verb or noun.  
  - *bank* can be a riverbank or financial bank.  
- Imagine the tag **VERB** has many possible next tags (high entropy).  
- The tag **NOUN** has fewer next tags (low entropy).  
- In MEMM, probabilities are **normalized locally at each state**.  
  - Fewer transitions → higher chance by default.  
  - More transitions → probability spread out.  
- So MEMM gets biased toward **states with fewer options**, even if features say otherwise.  

**In short:**  
MEMM is greedy at each step. It can’t use **future context** to fix wrong choices.  

**Chaitanya’s frustration:**  
*"Oh no… MEMM uses features, but it still makes biased local decisions!"*

 This problem is well known — called **label bias** in the original CRF paper (Lafferty et al., 2001).

# Conditional Random Fields (CRFs)

## Setup & Notation

We work with an observed sequence and a state/label sequence:

$$
\mathbf{x} = (x_1, \dots, x_m)
$$

$$
\mathbf{s} = (s_1, \dots, s_m), \quad s_j \in \mathcal{S}
$$

$$
\mathcal{S}^m = \text{set of all length-}m\text{ state sequences}
$$

We model the conditional distribution:

$$
p(\mathbf{s}\mid \mathbf{x})
$$


## Global Feature Vector and the Log-Linear CRF

Define a global feature vector:

$$
\Phi(\mathbf{x},\mathbf{s}) \in \mathbb{R}^d
$$

The conditional distribution is:

$$
p(\mathbf{s}\mid \mathbf{x};\mathbf{w})
=
\frac{\exp\!\big(\mathbf{w}^\top \Phi(\mathbf{x},\mathbf{s})\big)}
{\sum_{\mathbf{s'}\in \mathcal{S}^m} \exp\!\big(\mathbf{w}^\top \Phi(\mathbf{x},\mathbf{s'})\big)}
$$


## Local-to-Global Feature Decomposition

Assume features decompose along adjacent transitions(**“Adjacent transitions” simply means the pairs of neighboring states $(s_{j-1}, s_j)$ along the sequence.**)
:

$$
\Phi(\mathbf{x},\mathbf{s})
=
\sum_{j=1}^{m} \phi\!\big(x_j, s_{j-1}, s_j\big)
$$

For each coordinate:

$$
\Phi_k(\mathbf{x},\mathbf{s})
=
\sum_{j=1}^{m} \phi_k\!\big(x_j, s_{j-1}, s_j\big)
$$



## Decoding (Most Likely Sequence)

Goal:

$$
\arg\max_{\mathbf{s}\in \mathcal{S}^m} p(\mathbf{s}\mid \mathbf{x};\mathbf{w})
=
\arg\max_{\mathbf{s}\in \mathcal{S}^m} \mathbf{w}^\top \Phi(\mathbf{x},\mathbf{s})
=
\arg\max_{\mathbf{s}\in \mathcal{S}^m}
\sum_{j=1}^{m} \mathbf{w}^\top \phi\!\big(x_j, s_{j-1}, s_j\big)
$$

Define the transition score:

$$
\text{score}\!\left(s_{j-1}\!\to s_j; x_j\right)
=
\mathbf{w}^\top \phi\!\big(x_j, s_{j-1}, s_j\big)
$$

### Viterbi-Style Dynamic Program $(O(m|\mathcal{S}|^2))$

Initialization (with special start state $ s_0 $ described in text):

$$
V[1,s] = \mathbf{w}^\top \phi(x_1, s_0, s)
$$

Recurrence (for $ j=2,\dots,m $):

$$
V[j,s] = \max_{a\in \mathcal{S}}\Big( V[j-1,a] + \mathbf{w}^\top \phi(x_j, a, s) \Big)
$$

Termination:

$$
\max_{\mathbf{s}\in\mathcal{S}^m}
\sum_{j=1}^{m} \mathbf{w}^\top \phi(x_j, s_{j-1}, s_j)
=
\max_{s\in\mathcal{S}} V[m,s]
$$

(Use backpointers to reconstruct the best sequence.)



## Parameter Estimation (Training)

Regularized conditional log-likelihood over labeled pairs $ (\mathbf{x}^{(i)}, \mathbf{s}^{(i)}) $, $ i=1,\dots,n $:

$$
\mathcal{L}(\mathbf{w})
=
\sum_{i=1}^{n}
\log p\!\big(\mathbf{s}^{(i)} \mid \mathbf{x}^{(i)}; \mathbf{w}\big)
-
\frac{\lambda}{2}\,\lVert \mathbf{w}\rVert_2^2
$$

Estimator:

$$
\hat{\mathbf{w}}
=
\arg\max_{\mathbf{w}\in\mathbb{R}^d} \mathcal{L}(\mathbf{w})
$$

### Gradient (Empirical − Expected Counts − L2)

$$
\frac{\partial \mathcal{L}}{\partial w_k}
=
\sum_{i=1}^{n}\Phi_k\!\big(\mathbf{x}^{(i)}, \mathbf{s}^{(i)}\big)
-
\sum_{i=1}^{n}\sum_{\mathbf{s}\in\mathcal{S}^m}
p(\mathbf{s}\mid \mathbf{x}^{(i)}; \mathbf{w})\,
\Phi_k\!\big(\mathbf{x}^{(i)}, \mathbf{s}\big)
-
\lambda w_k
$$

Using the local decomposition:

$$
\sum_{\mathbf{s}}
p(\mathbf{s}\mid \mathbf{x};\mathbf{w})\,
\Phi_k(\mathbf{x},\mathbf{s})
=
\sum_{j=1}^{m}\sum_{a\in\mathcal{S}}\sum_{b\in\mathcal{S}}
q_j(a,b)\,\phi_k(x_j,a,b)
$$

Edge marginal definition:

$$
q_j(a,b)
=
\Pr\!\big(s_{j-1}=a,\, s_{j}=b \mid \mathbf{x}; \mathbf{w}\big)
$$


## Forward–Backward for Edge Marginals

Local potentials:

$$
\psi_j(a,b)
=
\exp\!\big(\mathbf{w}^\top \phi(x_j,a,b)\big)
$$

Forward messages:

$$
\alpha_1(b) = \psi_1(s_0,b)
$$

$$
\alpha_j(b) = \sum_{a\in\mathcal{S}} \alpha_{j-1}(a)\,\psi_j(a,b)
$$

Backward messages:

$$
\beta_m(b) = 1
$$

$$
\beta_{j-1}(a) = \sum_{b\in\mathcal{S}} \psi_j(a,b)\,\beta_j(b)
$$

Partition function:

$$
Z(\mathbf{x}) = \sum_{s\in\mathcal{S}} \alpha_m(s)
$$

Edge marginals:

$$
q_j(a,b)
=
\frac{\alpha_{j-1}(a)\,\psi_j(a,b)\,\beta_j(b)}
{Z(\mathbf{x})}
$$



## Complexity

Decoding (Viterbi):

$$
O(m\,|\mathcal{S}|^2)
$$

Forward–Backward expectations:

$$
O(m\,|\mathcal{S}|^2)
$$

Linear-chain CRFs are efficient thanks to dynamic programming over the chain.

## So now lets continue Chaitanya story...
### Why CRF solved his problems:
- CRF is like MEMM but **globally normalized**.  
- It still models $P(Y \mid X)$ (discriminative).  
- But instead of local decisions, CRF looks at the **whole sequence**.  
- It scores features and transitions **globally** and normalizes across all possible tag sequences.  
- This removes the **label bias problem**.

 In **linear-chain CRF** (used for NER), training uses methods like gradient descent  
It’s heavier than MEMM but gives better results.

### Benefits for NER:
- Like MEMM, CRF allows rich features.  
- But CRF makes decisions using the **whole sentence**.  
- Example: *Delhi* will correctly be tagged as a **LOCATION**, even if local bias wanted to mark it as “O”.  
- In modern NER, CRF is often combined with deep models (e.g., BiLSTM-CRF) for even higher accuracy.

**Chaitanya’s final happiness:**  
*"Yes! CRF gives me MEMM’s power but without bias.  
Now my NER system finally sees the full picture."*

#### Class 1: NERDataset

Purpose: Handles all data preparation and batching.

**Responsibilities**:

* Store and manage the dataset splits (train/val/test).
* Convert words/tags into their corresponding IDs using vocab (word_to_id, tag_to_id).
* Provide PyTorch DataLoader support with batching and padding for variable-length sequences.

````python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Dataset class
class NERDataset(Dataset):
    def __init__(self, sentences, word_to_id, tag_to_id, max_length=128):
        self.sentences = sentences
        self.word_to_id = word_to_id
        self.tag_to_id = tag_to_id
        self.max_length = max_length
        
    def __len__(self):
        return len(self.sentences)
    
    def __getitem__(self, idx):
        sentence_data = self.sentences[idx]
        word_ids = sentence_data['words']
        tag_ids = sentence_data['tags']
        
        # Pad sequences
        padded_word_ids = word_ids[:self.max_length] + [self.word_to_id['<PAD>']] * max(0, self.max_length - len(word_ids))
        padded_tag_ids = tag_ids[:self.max_length] + [self.tag_to_id['O']] * max(0, self.max_length - len(tag_ids))
        
        # Create mask (1 for real tokens, 0 for padding)
        mask = [1] * min(len(word_ids), self.max_length) + [0] * max(0, self.max_length - len(word_ids))
        
        return {
            'word_ids': torch.tensor(padded_word_ids, dtype=torch.long),
            'tag_ids': torch.tensor(padded_tag_ids, dtype=torch.long),
            'mask': torch.tensor(mask, dtype=torch.bool),
            'length': len(word_ids)
        }
````

````python
# Utility functions
def load_conll_file(file_path: str):
    sentences = []
    current_sentence = []
    
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                if current_sentence:
                    sentences.append(current_sentence)
                    current_sentence = []
            else:
                parts = line.split()
                if len(parts) >= 2:
                    token = parts[0]
                    tag = parts[1]
                    current_sentence.append((token, tag))
        
        if current_sentence:
            sentences.append(current_sentence)
    
    return sentences

def create_mappings(sentences):
    word_freq = Counter()
    tag_set = set()
    
    for sentence in sentences:
        for token, tag in sentence:
            word_freq[token] += 1
            tag_set.add(tag)
    
    sorted_words = [word for word, count in word_freq.most_common()]
    
    word_to_id = {'<UNK>': 0, '<PAD>': 1}
    id_to_word = {0: '<UNK>', 1: '<PAD>'}
    
    for idx, word in enumerate(sorted_words, 2):
        word_to_id[word] = idx
        id_to_word[idx] = word
    
    tag_list = sorted(list(tag_set))
    tag_to_id = {}
    id_to_tag = {}
    
    for idx, tag in enumerate(tag_list):
        tag_to_id[tag] = idx
        id_to_tag[idx] = tag
    
    return word_to_id, id_to_word, tag_to_id, id_to_tag

def prepare_corpus(sentences, word_to_id, tag_to_id):
    corpus = []
    
    for sentence in sentences:
        word_ids = []
        tag_ids = []
        
        for token, tag in sentence:
            word_id = word_to_id.get(token, word_to_id['<UNK>'])
            tag_id = tag_to_id[tag]
            
            word_ids.append(word_id)
            tag_ids.append(tag_id)
        
        corpus.append({
            'words': word_ids,
            'tags': tag_ids
        })
    
    return corpus
````

#### Class 2: CRFModel

Purpose: Defines the Neural Network + CRF layer.

Responsibilities:
- Word embedding lookup (nn.Embedding).
- Linear layer that projects embeddings into tag space (emission scores).
- CRF mechanics:
  - Forward algorithm (partition function).
  - Viterbi decoding for prediction.
  - Negative log-likelihood for training loss.

````python
# Fixed CRF Model
class CRFModel(nn.Module):
    def __init__(self, num_tags, vocab_size, embedding_dim=128, hidden_dim=256):
        super().__init__()
        self.num_tags = num_tags
        self.vocab_size = vocab_size
        
        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=1)
        
        # BiLSTM layer
        self.lstm = nn.LSTM(embedding_dim, hidden_dim // 2, 
                           num_layers=1, bidirectional=True, batch_first=True)
        
        # Linear layer to convert LSTM outputs to emission scores
        self.hidden2tag = nn.Linear(hidden_dim, num_tags)
        
        # CRF transition parameters
        self.transitions = nn.Parameter(torch.randn(num_tags, num_tags))
        
        # Constraints
        self.START_TAG = num_tags - 2
        self.STOP_TAG = num_tags - 1
        
        # Initialize transitions with constraints
        self.transitions.data[self.START_TAG, :] = -10000
        self.transitions.data[:, self.STOP_TAG] = -10000
        
    def forward(self, sentence_ids):
        """Forward pass - compute emission scores"""
        embeddings = self.embedding(sentence_ids)
        lstm_out, _ = self.lstm(embeddings)
        emissions = self.hidden2tag(lstm_out)
        return emissions
    
    def _forward_algorithm(self, emissions, mask):
        """Compute log partition function with masking"""
        batch_size, seq_length, num_tags = emissions.shape
        
        # Initialize alpha in log space
        alpha = torch.full((batch_size, num_tags), -10000.0).to(device)
        alpha[:, self.START_TAG] = 0.0
        
        # Iterate through sequence
        for t in range(seq_length):
            current_mask = mask[:, t].unsqueeze(1)  # (batch_size, 1)
            emit_scores = emissions[:, t]  # (batch_size, num_tags)
            
            # Broadcast alpha for transitions
            alpha_expanded = alpha.unsqueeze(2)  # (batch_size, num_tags, 1)
            transitions_expanded = self.transitions.unsqueeze(0)  # (1, num_tags, num_tags)
            
            # Compute scores for all transitions
            scores = alpha_expanded + transitions_expanded + emit_scores.unsqueeze(1)
            
            # Log-sum-exp over previous states
            new_alpha = torch.logsumexp(scores, dim=1)
            
            # Update alpha with masking
            alpha = new_alpha * current_mask + alpha * (~current_mask)
        
        # Add final transitions to STOP_TAG
        final_scores = alpha + self.transitions[:, self.STOP_TAG].unsqueeze(0)
        log_partition = torch.logsumexp(final_scores, dim=1)
        
        return log_partition
    
    def _score_sequence(self, emissions, tags, mask):
        """Compute score for gold tag sequence"""
        batch_size, seq_length, num_tags = emissions.shape
        
        # Score from START_TAG to first tag
        score = self.transitions[self.START_TAG, tags[:, 0]]
        score += emissions[:, 0].gather(1, tags[:, 0].unsqueeze(1)).squeeze(1)
        
        # Score for each transition
        for t in range(1, seq_length):
            current_mask = mask[:, t]
            prev_tags = tags[:, t-1]
            current_tags = tags[:, t]
            
            # Transition scores
            transition_score = self.transitions[prev_tags, current_tags]
            
            # Emission scores
            emission_score = emissions[:, t].gather(1, current_tags.unsqueeze(1)).squeeze(1)
            
            # Update score
            score += (transition_score + emission_score) * current_mask
        
        # Score to STOP_TAG from last tag
        last_positions = mask.sum(dim=1).long() - 1
        last_tags = tags.gather(1, last_positions.unsqueeze(1)).squeeze(1)
        score += self.transitions[last_tags, self.STOP_TAG]
        
        return score
    
    def neg_log_likelihood(self, sentence_ids, tags, mask):
        """Compute negative log likelihood loss"""
        emissions = self.forward(sentence_ids)
        
        # Compute scores
        gold_score = self._score_sequence(emissions, tags, mask)
        forward_score = self._forward_algorithm(emissions, mask)
        
        # Negative log likelihood (average over batch)
        nll = forward_score - gold_score
        return nll.mean()  # Return scalar for backpropagation
    
    def viterbi_decode(self, emissions, mask):
        """Viterbi decoding for prediction"""
        batch_size, seq_length, num_tags = emissions.shape
        
        # Initialize backpointers and viterbi variables
        backpointers = torch.zeros(batch_size, seq_length, num_tags, dtype=torch.long).to(device)
        viterbi = torch.full((batch_size, num_tags), -10000.0).to(device)
        viterbi[:, self.START_TAG] = 0.0
        
        # Forward pass
        for t in range(seq_length):
            current_mask = mask[:, t].unsqueeze(1)
            emit_scores = emissions[:, t]
            
            # Expand viterbi for transitions
            viterbi_expanded = viterbi.unsqueeze(2)  # (batch_size, num_tags, 1)
            transitions_expanded = self.transitions.unsqueeze(0)  # (1, num_tags, num_tags)
            
            # Compute scores
            scores = viterbi_expanded + transitions_expanded + emit_scores.unsqueeze(1)
            
            # Find best scores and backpointers
            best_scores, best_tags = torch.max(scores, dim=1)
            backpointers[:, t] = best_tags
            
            # Update viterbi with masking
            viterbi = best_scores * current_mask + viterbi * (~current_mask)
        
        # Add final transitions to STOP_TAG
        final_scores = viterbi + self.transitions[:, self.STOP_TAG].unsqueeze(0)
        _, best_final_tags = torch.max(final_scores, dim=1)
        
        # Backtrack to find best path
        best_paths = []
        for i in range(batch_size):
            best_path = [best_final_tags[i].item()]
            valid_length = mask[i].sum().item()
            
            for t in range(seq_length - 1, 0, -1):
                if t < valid_length:
                    best_tag = backpointers[i, t, best_path[0]]
                    best_path.insert(0, best_tag.item())
            
            best_paths.append(best_path)
        
        return best_paths
    
    def decode(self, sentence_ids, mask):
        """Decode best tag sequence"""
        emissions = self.forward(sentence_ids)
        return self.viterbi_decode(emissions, mask)
````

#### Class 3: Trainer

Purpose: Handles training and evaluation loop.

Responsibilities:
- Initialize optimizer and manage training epochs.
- Call neg_log_likelihood from CRFModel to compute loss.
- Backpropagation and optimizer step.
- Evaluate on validation/test sets:
  - Decode predictions using CRFModel.viterbi_decode.
  - Convert tag IDs back to tag strings.
  - Print classification metrics (Precision, Recall, F1).

````python

# Fixed Trainer class
class Trainer:
    def __init__(self, model, device, lr=0.01):
        self.model = model.to(device)
        self.device = device
        self.optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
        
    def train_model(self, train_loader, val_loader, epochs=10):
        """Fixed training loop with proper loss handling"""
        train_losses = []
        val_losses = []
        
        for epoch in range(epochs):
            # Training
            self.model.train()
            total_train_loss = 0
            num_batches = 0
            
            for batch in train_loader:
                word_ids = batch['word_ids'].to(self.device)
                tag_ids = batch['tag_ids'].to(self.device)
                mask = batch['mask'].to(self.device)
                
                # Forward pass - get scalar loss
                loss = self.model.neg_log_likelihood(word_ids, tag_ids, mask)
                
                # Backward pass
                self.optimizer.zero_grad()
                loss.backward()  # Now this works because loss is scalar
                torch.nn.utils.clip_grad_norm_(self.model.parameters(), 5.0)
                self.optimizer.step()
                
                total_train_loss += loss.item()
                num_batches += 1
            
            avg_train_loss = total_train_loss / num_batches
            train_losses.append(avg_train_loss)
            
            # Validation
            self.model.eval()
            total_val_loss = 0
            num_val_batches = 0
            
            with torch.no_grad():
                for batch in val_loader:
                    word_ids = batch['word_ids'].to(self.device)
                    tag_ids = batch['tag_ids'].to(self.device)
                    mask = batch['mask'].to(self.device)
                    
                    loss = self.model.neg_log_likelihood(word_ids, tag_ids, mask)
                    total_val_loss += loss.item()
                    num_val_batches += 1
            
            avg_val_loss = total_val_loss / num_val_batches
            val_losses.append(avg_val_loss)
            
            print(f"Epoch {epoch+1}/{epochs}: Train Loss = {avg_train_loss:.4f}, Val Loss = {avg_val_loss:.4f}")
        
        # Plot training history
        self.plot_training_history(train_losses, val_losses)
        
        return train_losses, val_losses
    
    def plot_training_history(self, train_losses, val_losses):
        plt.figure(figsize=(10, 6))
        plt.plot(train_losses, label='Training Loss')
        plt.plot(val_losses, label='Validation Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.title('Training History')
        plt.legend()
        plt.grid(True)
        plt.show()
    
    @torch.no_grad()
    def evaluate_model(self, test_loader, id_to_tag):
        """Run evaluation loop and print metrics."""
        self.model.eval()
        
        all_true_tags = []
        all_pred_tags = []
        
        for batch in test_loader:
            word_ids = batch['word_ids'].to(self.device)
            mask = batch['mask'].to(self.device)
            true_tags = batch['tag_ids']
            lengths = batch['length']
            
            # Decode predictions
            pred_tag_sequences = self.model.decode(word_ids, mask)
            
            # Convert to flat lists
            for i, length in enumerate(lengths):
                true_seq = true_tags[i][:length].tolist()
                
                # Ensure we don't exceed the predicted sequence length
                if i < len(pred_tag_sequences):
                    pred_seq = pred_tag_sequences[i]
                    # Take only up to the actual length
                    pred_seq = pred_seq[:length]
                    
                    # Convert tag IDs to tag names (excluding START/STOP tags)
                    for j in range(min(len(true_seq), len(pred_seq))):
                        true_tag_id = true_seq[j]
                        pred_tag_id = pred_seq[j]
                        
                        # Only include if it's a valid tag (not START/STOP)
                        if true_tag_id in id_to_tag and id_to_tag[true_tag_id] not in ['<START>', '<STOP>']:
                            all_true_tags.append(id_to_tag[true_tag_id])
                            all_pred_tags.append(id_to_tag.get(pred_tag_id, 'O'))  # Default to 'O' if unknown
        
        # Print classification report
        print("\n=== Classification Report ===")
        print(classification_report(all_true_tags, all_pred_tags, zero_division=0))
        
        # Plot confusion matrix
        self.plot_confusion_matrix(all_true_tags, all_pred_tags, [tag for tag in id_to_tag.values() if tag not in ['<START>', '<STOP>']])
        
        return all_true_tags, all_pred_tags
    
    def plot_confusion_matrix(self, y_true, y_pred, labels):
        cm = confusion_matrix(y_true, y_pred, labels=labels)
        
        plt.figure(figsize=(12, 10))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                   xticklabels=labels, yticklabels=labels)
        plt.title('Confusion Matrix')
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.xticks(rotation=45)
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()
````

# Introduction and Motivation

## Introduction
Sequential data is at the heart of many real-world applications such as speech recognition, machine translation, sentiment analysis, and named entity recognition (NER). The challenge lies in effectively modeling temporal dependencies and contextual relationships across varying sequence lengths. Over time, different approaches have been proposed to handle sequential data, starting from probabilistic models to deep neural architectures.


## Motivation
Early approaches such as **Hidden Markov Models (HMMs)** provided a probabilistic framework for modeling sequences, where hidden states generate observable sequences. While successful in certain domains, HMMs rely on strong independence assumptions and struggle to capture long-term dependencies. Training with algorithms like **Baum-Welch** often led to challenges with data sparsity and scalability.

To overcome some of these limitations, **Conditional Random Fields (CRFs)** were introduced. Unlike HMMs, CRFs are discriminative models that directly model the conditional probability of label sequences given inputs. They allow the inclusion of richer features and are effective in tasks like part-of-speech tagging and NER. However, CRFs still depend heavily on hand-engineered features and cannot automatically learn hierarchical representations.

The emergence of **Recurrent Neural Networks (RNNs)** shifted the paradigm by enabling models to learn representations directly from raw data. RNNs can, in principle, capture sequential dependencies but suffer from vanishing and exploding gradient problems, which limit their ability to handle long sequences.

**Long Short-Term Memory (LSTM) networks** addressed these issues through gating mechanisms that control information flow, allowing them to retain relevant information over longer contexts. This made LSTMs powerful for a wide range of sequential tasks, from language modeling to speech recognition.

Building on this, **Bidirectional LSTMs (BiLSTMs)** enhanced performance by processing input sequences in both forward and backward directions, thus capturing context from both past and future tokens. Extending this further, **Multi-layer BiLSTMs** stack several layers to learn increasingly abstract and hierarchical representations of sequential data.

In summary, the journey from HMMs and CRFs to LSTMs and their variants demonstrates a clear progression: from **probabilistic models with restrictive assumptions and manual features** to **neural architectures that automatically learn rich, long-range, and hierarchical representations**. This evolution motivates the use of LSTMs, BiLSTMs, and Multi-layer BiLSTMs for modern sequence modeling tasks.


## Task 1: Vanilla LSTM Implementation (25 Marks)

### Conceptual Background
- RNNs suffer from vanishing/exploding gradients, making them ineffective for long-term dependencies.  
- LSTM introduces a cell state ($C_t$) and gates (forget, input, output) to regulate memory flow.  

### Mathematical Formulation
Let input be $x_t$, previous hidden be $h_{t-1}$, previous cell be $C_{t-1}$.

**1. Forget Gate**
$$
f_t = \sigma\big(W_f x_t + U_f h_{t-1} + b_f\big)
$$

**2. Input Gate**
$$
i_t = \sigma\big(W_i x_t + U_i h_{t-1} + b_i\big)
$$

**3. Candidate Cell State**
$$
\tilde{C}_t = \tanh\big(W_c x_t + U_c h_{t-1} + b_c\big)
$$

**4. Cell State Update**
$$
C_t = f_t \odot C_{t-1} + i_t \odot \tilde{C}_t
$$

**5. Output Gate**
$$
o_t = \sigma\big(W_o x_t + U_o h_{t-1} + b_o\big)
$$

**6. Hidden State**
$$
h_t = o_t \odot \tanh(C_t)
$$

> Here $\sigma(\cdot)$ is the logistic sigmoid, $\tanh(\cdot)$ is the hyperbolic tangent, and $\odot$ denotes element-wise multiplication.

## Task 2: Bidirectional LSTM (BiLSTM) (25 Marks)

### Conceptual Background
- A standard LSTM captures *past* context only.  
- BiLSTM uses two LSTMs:
  - **Forward LSTM** processes $x_1 \to x_T$.
  - **Backward LSTM** processes $x_T \to x_1$.

### Mathematical Formulation

**Forward LSTM**
$$
\overrightarrow{h}_t = \mathrm{LSTM}_{\text{fwd}}(x_t, \overrightarrow{h}_{t-1})
$$

**Backward LSTM**
$$
\overleftarrow{h}_t = \mathrm{LSTM}_{\text{bwd}}(x_t, \overleftarrow{h}_{t+1})
$$

**Combined Hidden State** (concatenation)
$$
h_t = \big[\,\overrightarrow{h}_t \,;\, \overleftarrow{h}_t\,\big]
$$

**Output Sequence**
$$
H = \big(h_1, h_2, \dots, h_T\big)
$$

## Task 3: Multi-layer BiLSTM (Stacked BiLSTM) (25 Marks)

### Conceptual Background
- A single BiLSTM layer may not capture hierarchical or deep temporal patterns.  
- Stacking multiple BiLSTM layers allows successive layers to build higher-level representations: lower layers capture local patterns; higher layers capture more abstract, global features.

### Notation
- Let $L$ be the number of stacked BiLSTM layers.
- Superscript $(l)$ denotes the layer index (e.g., first layer $l=1$, final layer $l=L$).
- Subscript $t$ denotes the time step.

### First BiLSTM Layer ($l=1$)
**Forward direction**
$$
\overrightarrow{h}_t^{(1)} = \mathrm{LSTM}_{\text{fwd}}^{(1)}\big(x_t,\; \overrightarrow{h}_{t-1}^{(1)}\big)
$$

**Backward direction**
$$
\overleftarrow{h}_t^{(1)} = \mathrm{LSTM}_{\text{bwd}}^{(1)}\big(x_t,\; \overleftarrow{h}_{t+1}^{(1)}\big)
$$

**Concatenate to form layer-1 representation**
$$
h_t^{(1)} = \big[\,\overrightarrow{h}_t^{(1)} \,;\, \overleftarrow{h}_t^{(1)}\,\big]
$$

### Higher Layers ($l = 2, \dots, L$)

For **layer $l$**, the input at each time step is the concatenated output from layer $l-1$, i.e., $h_t^{(l-1)}$.

**Forward direction**
$$
\overrightarrow{h}_t^{(l)} = \mathrm{LSTM}_{\text{fwd}}^{(l)}\big(h_t^{(l-1)},\; \overrightarrow{h}_{t-1}^{(l)}\big)
$$

**Backward direction**
$$
\overleftarrow{h}_t^{(l)} = \mathrm{LSTM}_{\text{bwd}}^{(l)}\big(h_t^{(l-1)},\; \overleftarrow{h}_{t+1}^{(l)}\big)
$$

**Concatenate to form layer-$l$ representation**
$$
h_t^{(l)} = \big[\,\overrightarrow{h}_t^{(l)} \,;\, \overleftarrow{h}_t^{(l)}\,\big]
$$

### Final Output after $L$ layers
$$
H^{(L)} = \big(h_1^{(L)},\; h_2^{(L)},\; \dots,\; h_T^{(L)}\big)
$$

#### Implement LSTM cell for single step

````python
# Task 1: Vanilla LSTM Implementation

class LSTMCell(nn.Module):
    """
    A single LSTM cell that implements the logic for one time step.
    """
    def __init__(self, input_size, hidden_size):
        super(LSTMCell, self).__init__()
        self.input_size = input_size
        self.hidden_size = hidden_size
        
        # Learnable parameters for gates
        # We combine all gates in one matrix multiplication for efficiency
        self.weight_ih = nn.Parameter(torch.randn(4 * hidden_size, input_size))
        self.weight_hh = nn.Parameter(torch.randn(4 * hidden_size, hidden_size))
        self.bias = nn.Parameter(torch.zeros(4 * hidden_size))
        
        self.reset_parameters()
    
    def reset_parameters(self):
        """Initialize parameters using Xavier initialization"""
        nn.init.xavier_uniform_(self.weight_ih)
        nn.init.orthogonal_(self.weight_hh)
        # Set forget gate bias to 1 to help with gradient flow
        self.bias.data[self.hidden_size:2*self.hidden_size] = 1.0
    
    def forward(self, x, states):
        """
        Forward pass of the LSTM cell.
        
        Args:
            x: Input tensor of shape (batch_size, input_size)
            states: Tuple of (h_prev, c_prev) both of shape (batch_size, hidden_size)
        
        Returns:
            h_next: Next hidden state (batch_size, hidden_size)
            c_next: Next cell state (batch_size, hidden_size)
        """
        h_prev, c_prev = states
        
        # Combine input and hidden transformations
        gates = (x @ self.weight_ih.t() + h_prev @ self.weight_hh.t() + self.bias)
        
        # Split into input, forget, cell, output gates
        i_gate, f_gate, g_gate, o_gate = gates.chunk(4, 1)
        
        # Apply activation functions
        i_gate = torch.sigmoid(i_gate)  # Input gate
        f_gate = torch.sigmoid(f_gate)  # Forget gate
        g_gate = torch.tanh(g_gate)     # Candidate cell state
        o_gate = torch.sigmoid(o_gate)  # Output gate
        
        # Update cell state
        c_next = f_gate * c_prev + i_gate * g_gate
        
        # Update hidden state
        h_next = o_gate * torch.tanh(c_next)
        
        return h_next, c_next
````

#### Implement full general LSTM layer

````python
class LSTM(nn.Module):
    """
    A full LSTM layer that uses LSTMCell to process an input sequence.
    """
    def __init__(self, input_size, hidden_size, bidirectional=False, num_layers=1):
        super(LSTM, self).__init__()
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.bidirectional = bidirectional
        self.num_layers = num_layers
        self.num_directions = 2 if bidirectional else 1
        
        # Create LSTM cells for each layer and direction
        self.cells = nn.ModuleList()
        for layer in range(num_layers):
            layer_input_size = input_size if layer == 0 else hidden_size * self.num_directions
            if bidirectional:
                self.cells.append(LSTMCell(layer_input_size, hidden_size))  # Forward
                self.cells.append(LSTMCell(layer_input_size, hidden_size))  # Backward
            else:
                self.cells.append(LSTMCell(layer_input_size, hidden_size))
    
    def forward(self, input_seq, states=None):
        """
        Forward pass of the full LSTM layer.
        
        Args:
            input_seq: Input sequence of shape (batch_size, seq_len, input_size)
            states: Tuple of (h_0, c_0) or None
            
        Returns:
            output: Output features (batch_size, seq_len, num_directions * hidden_size)
            (h_n, c_n): Final hidden and cell states
        """
        batch_size, seq_len, _ = input_seq.shape
        
        # Initialize states if not provided
        if states is None:
            h_0 = torch.zeros(self.num_layers * self.num_directions, batch_size, self.hidden_size, device=input_seq.device)
            c_0 = torch.zeros(self.num_layers * self.num_directions, batch_size, self.hidden_size, device=input_seq.device)
        else:
            h_0, c_0 = states
        
        # Transpose for easier iteration (seq_len, batch_size, input_size)
        input_seq = input_seq.transpose(0, 1)
        
        # Initialize outputs
        outputs = []
        layer_outputs = [input_seq]  # Store outputs from each layer
        
        # Process each layer
        for layer in range(self.num_layers):
            layer_input = layer_outputs[-1]  # (seq_len, batch_size, layer_input_size)
            layer_output = []
            
            if self.bidirectional:
                # Forward direction
                h_forward = h_0[layer * 2]
                c_forward = c_0[layer * 2]
                forward_outputs = []
                
                for t in range(seq_len):
                    h_forward, c_forward = self.cells[layer * 2](layer_input[t], (h_forward, c_forward))
                    forward_outputs.append(h_forward)
                
                # Backward direction
                h_backward = h_0[layer * 2 + 1]
                c_backward = c_0[layer * 2 + 1]
                backward_outputs = []
                
                for t in range(seq_len-1, -1, -1):
                    h_backward, c_backward = self.cells[layer * 2 + 1](layer_input[t], (h_backward, c_backward))
                    backward_outputs.insert(0, h_backward)  # Insert at beginning to maintain order
                
                # Concatenate forward and backward outputs
                for t in range(seq_len):
                    combined = torch.cat([forward_outputs[t], backward_outputs[t]], dim=1)
                    layer_output.append(combined)
                
                # Store final states
                h_n_forward = h_forward.unsqueeze(0)
                c_n_forward = c_forward.unsqueeze(0)
                h_n_backward = h_backward.unsqueeze(0)
                c_n_backward = c_backward.unsqueeze(0)
                
                if layer == 0:
                    h_n = torch.cat([h_n_forward, h_n_backward], dim=0)
                    c_n = torch.cat([c_n_forward, c_n_backward], dim=0)
                else:
                    h_n = torch.cat([h_n, h_n_forward, h_n_backward], dim=0)
                    c_n = torch.cat([c_n, c_n_forward, c_n_backward], dim=0)
                    
            else:
                # Unidirectional LSTM
                h_layer = h_0[layer]
                c_layer = c_0[layer]
                
                for t in range(seq_len):
                    h_layer, c_layer = self.cells[layer](layer_input[t], (h_layer, c_layer))
                    layer_output.append(h_layer)
                
                # Store final state
                h_n_layer = h_layer.unsqueeze(0)
                c_n_layer = c_layer.unsqueeze(0)
                
                if layer == 0:
                    h_n = h_n_layer
                    c_n = c_n_layer
                else:
                    h_n = torch.cat([h_n, h_n_layer], dim=0)
                    c_n = torch.cat([c_n, c_n_layer], dim=0)
            
            layer_output = torch.stack(layer_output)  # (seq_len, batch_size, output_size)
            layer_outputs.append(layer_output)
        
        # Final output from last layer
        output = layer_outputs[-1].transpose(0, 1)  # (batch_size, seq_len, output_size)
        
        return output, (h_n, c_n)
````

````python
# Task 1: Vanilla LSTM Model
class VanillaLSTM(nn.Module):
    """
    A Vanilla LSTM model for Named Entity Recognition (NER).
    """
    def __init__(self, vocab_size, tagset_size, embedding_dim, hidden_size):
        super().__init__()
        self.embedding_dim = embedding_dim
        self.hidden_size = hidden_size
        self.vocab_size = vocab_size
        self.tagset_size = tagset_size
        
        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=1)
        
        # LSTM layer
        self.lstm = LSTM(embedding_dim, hidden_size, bidirectional=False)
        
        # Output layer
        self.output_layer = nn.Linear(hidden_size, tagset_size)
        
        # Dropout for regularization
        self.dropout = nn.Dropout(0.3)
    
    def forward(self, sentence):
        """
        Forward pass of the Vanilla LSTM NER model.
        
        Args:
            sentence: Input sequence of token indices, shape: (batch_size, seq_len)
            
        Returns:
            tag_scores: Output scores for each tag, shape: (batch_size, seq_len, tagset_size)
        """
        # Get embeddings
        embeddings = self.embedding(sentence)  # (batch_size, seq_len, embedding_dim)
        
        # Apply dropout to embeddings
        embeddings = self.dropout(embeddings)
        
        # Pass through LSTM
        lstm_out, _ = self.lstm(embeddings)  # (batch_size, seq_len, hidden_size)
        
        # Apply dropout to LSTM output
        lstm_out = self.dropout(lstm_out)
        
        # Get tag scores
        tag_scores = self.output_layer(lstm_out)  # (batch_size, seq_len, tagset_size)
        
        return tag_scores
````

````python
# Task 2: Bidirectional LSTM Model
class BiLSTM(nn.Module):
    """
    A Bi-directional LSTM model for Named Entity Recognition (NER).
    """
    def __init__(self, vocab_size, tagset_size, embedding_dim, hidden_size):
        super().__init__()
        self.embedding_dim = embedding_dim
        self.hidden_size = hidden_size
        self.vocab_size = vocab_size
        self.tagset_size = tagset_size
        
        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=1)
        
        # Bidirectional LSTM layer
        self.lstm = LSTM(embedding_dim, hidden_size, bidirectional=True)
        
        # Output layer (hidden_size * 2 because bidirectional)
        self.output_layer = nn.Linear(hidden_size * 2, tagset_size)
        
        # Dropout for regularization
        self.dropout = nn.Dropout(0.3)
    
    def forward(self, sentence):
        """
        Forward pass of the BiLSTM NER model.
        """
        # Get embeddings
        embeddings = self.embedding(sentence)
        embeddings = self.dropout(embeddings)
        
        # Pass through bidirectional LSTM
        lstm_out, _ = self.lstm(embeddings)  # (batch_size, seq_len, hidden_size * 2)
        
        # Apply dropout
        lstm_out = self.dropout(lstm_out)
        
        # Get tag scores
        tag_scores = self.output_layer(lstm_out)
        
        return tag_scores
````

````python
# Task 3: Multi-layer BiLSTM Model
class MultiLayerBiLSTM(nn.Module):
    """
    A multi-layer Bidirectional LSTM built by stacking LSTM layers.
    """
    def __init__(self, vocab_size, tagset_size, embedding_dim, hidden_size, num_layers):
        super().__init__()
        self.embedding_dim = embedding_dim
        self.hidden_size = hidden_size
        self.vocab_size = vocab_size
        self.tagset_size = tagset_size
        self.num_layers = num_layers
        
        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=1)
        
        # Multi-layer bidirectional LSTM
        self.lstm = LSTM(embedding_dim, hidden_size, bidirectional=True, num_layers=num_layers)
        
        # Output layer
        self.output_layer = nn.Linear(hidden_size * 2, tagset_size)
        
        # Dropout for regularization
        self.dropout = nn.Dropout(0.3)
    
    def forward(self, sequence):
        """
        Forward pass of the Multi-layer BiLSTM.
        """
        # Get embeddings
        embeddings = self.embedding(sequence)
        embeddings = self.dropout(embeddings)
        
        # Pass through multi-layer bidirectional LSTM
        lstm_out, states = self.lstm(embeddings)  # (batch_size, seq_len, hidden_size * 2)
        
        # Apply dropout
        lstm_out = self.dropout(lstm_out)
        
        # Get tag scores
        tag_scores = self.output_layer(lstm_out)
        
        return tag_scores
````

````python
# Dataset class for NER
class NERDataset(Dataset):
    def __init__(self, sentences, word_to_id, tag_to_id, max_length=128):
        self.sentences = sentences
        self.word_to_id = word_to_id
        self.tag_to_id = tag_to_id
        self.max_length = max_length
        
    def __len__(self):
        return len(self.sentences)
    
    def __getitem__(self, idx):
        sentence_data = self.sentences[idx]
        word_ids = sentence_data['words']
        tag_ids = sentence_data['tags']
        
        # Pad sequences
        padded_word_ids = word_ids[:self.max_length] + [self.word_to_id['<PAD>']] * max(0, self.max_length - len(word_ids))
        padded_tag_ids = tag_ids[:self.max_length] + [self.tag_to_id['O']] * max(0, self.max_length - len(tag_ids))
        
        # Create mask
        mask = [1] * min(len(word_ids), self.max_length) + [0] * max(0, self.max_length - len(word_ids))
        
        return {
            'word_ids': torch.tensor(padded_word_ids, dtype=torch.long),
            'tag_ids': torch.tensor(padded_tag_ids, dtype=torch.long),
            'mask': torch.tensor(mask, dtype=torch.bool),
            'length': len(word_ids)
        }
````

````python
# Trainer class for all models
class NERTrainer:
    def __init__(self, model, device, lr=0.001):
        self.model = model.to(device)
        self.device = device
        self.optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
        self.criterion = nn.CrossEntropyLoss(ignore_index=0)  # Ignore padding index
    
    def train_epoch(self, train_loader):
        self.model.train()
        total_loss = 0
        num_batches = 0
        
        for batch in train_loader:
            word_ids = batch['word_ids'].to(self.device)
            tag_ids = batch['tag_ids'].to(self.device)
            mask = batch['mask'].to(self.device)
            
            # Forward pass
            tag_scores = self.model(word_ids)
            
            # Reshape for loss calculation
            batch_size, seq_len, num_tags = tag_scores.shape
            tag_scores_flat = tag_scores.reshape(-1, num_tags)
            tag_ids_flat = tag_ids.reshape(-1)
            
            # Compute loss only on non-padded tokens
            loss = self.criterion(tag_scores_flat, tag_ids_flat)
            
            # Backward pass
            self.optimizer.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(self.model.parameters(), 5.0)  # Gradient clipping
            self.optimizer.step()
            
            total_loss += loss.item()
            num_batches += 1
        
        return total_loss / num_batches
    
    def evaluate(self, data_loader, id_to_tag):
        self.model.eval()
        all_true_tags = []
        all_pred_tags = []
        
        with torch.no_grad():
            for batch in data_loader:
                word_ids = batch['word_ids'].to(self.device)
                tag_ids = batch['tag_ids']
                mask = batch['mask']
                lengths = batch['length']
                
                # Get predictions
                tag_scores = self.model(word_ids)
                pred_tag_ids = torch.argmax(tag_scores, dim=2)
                
                # Convert to lists
                for i, length in enumerate(lengths):
                    true_seq = tag_ids[i][:length].tolist()
                    pred_seq = pred_tag_ids[i][:length].tolist()
                    
                    true_tags = [id_to_tag[tag_id] for tag_id in true_seq]
                    pred_tags = [id_to_tag[tag_id] for tag_id in pred_seq]
                    
                    all_true_tags.extend(true_tags)
                    all_pred_tags.extend(pred_tags)
        
        return all_true_tags, all_pred_tags
    
    def train(self, train_loader, val_loader, epochs=10):
        train_losses = []
        val_f1_scores = []
        
        for epoch in range(epochs):
            # Training
            train_loss = self.train_epoch(train_loader)
            train_losses.append(train_loss)
            
            # Validation
            true_tags, pred_tags = self.evaluate(val_loader, id_to_tag)
            report = classification_report(true_tags, pred_tags, output_dict=True, zero_division=0)
            val_f1 = report['macro avg']['f1-score']
            val_f1_scores.append(val_f1)
            
            print(f"Epoch {epoch+1}/{epochs}: Train Loss = {train_loss:.4f}, Val F1 = {val_f1:.4f}")
        
        # Plot training history
        self.plot_training_history(train_losses, val_f1_scores)
        
        return train_losses, val_f1_scores
    
    def plot_training_history(self, train_losses, val_f1_scores):
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
        
        # Plot loss
        ax1.plot(train_losses, 'b-', label='Training Loss')
        ax1.set_xlabel('Epoch')
        ax1.set_ylabel('Loss')
        ax1.set_title('Training Loss')
        ax1.legend()
        ax1.grid(True)
        
        # Plot F1 score
        ax2.plot(val_f1_scores, 'r-', label='Validation F1')
        ax2.set_xlabel('Epoch')
        ax2.set_ylabel('F1 Score')
        ax2.set_title('Validation F1 Score')
        ax2.legend()
        ax2.grid(True)
        
        plt.tight_layout()
        plt.show()
````

````python
# Utility functions for data loading
def load_conll_file(file_path: str):
    sentences = []
    current_sentence = []
    
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                if current_sentence:
                    sentences.append(current_sentence)
                    current_sentence = []
            else:
                parts = line.split()
                if len(parts) >= 2:
                    token = parts[0]
                    tag = parts[1]
                    current_sentence.append((token, tag))
        
        if current_sentence:
            sentences.append(current_sentence)
    
    return sentences
````

````python
def create_mappings(sentences):
    word_freq = Counter()
    tag_set = set()
    
    for sentence in sentences:
        for token, tag in sentence:
            word_freq[token] += 1
            tag_set.add(tag)
    
    sorted_words = [word for word, count in word_freq.most_common()]
    
    word_to_id = {'<UNK>': 0, '<PAD>': 1}
    id_to_word = {0: '<UNK>', 1: '<PAD>'}
    
    for idx, word in enumerate(sorted_words, 2):
        word_to_id[word] = idx
        id_to_word[idx] = word
    
    tag_list = sorted(list(tag_set))
    tag_to_id = {}
    id_to_tag = {}
    
    for idx, tag in enumerate(tag_list):
        tag_to_id[tag] = idx
        id_to_tag[idx] = tag
    
    return word_to_id, id_to_word, tag_to_id, id_to_tag
````

````python
def prepare_corpus(sentences, word_to_id, tag_to_id):
    corpus = []
    
    for sentence in sentences:
        word_ids = []
        tag_ids = []
        
        for token, tag in sentence:
            word_id = word_to_id.get(token, word_to_id['<UNK>'])
            tag_id = tag_to_id[tag]
            
            word_ids.append(word_id)
            tag_ids.append(tag_id)
        
        corpus.append({
            'words': word_ids,
            'tags': tag_ids
        })
    
    return corpus
````

````python
# Task 4: Comparative Evaluation
def compare_models(models_dict, test_loader, id_to_tag):
    """
    Compare multiple models and print comprehensive evaluation metrics.
    """
    results = {}
    
    for model_name, (model, trainer) in models_dict.items():
        print(f"\n{'='*50}")
        print(f"Evaluating {model_name}")
        print(f"{'='*50}")
        
        # Evaluate model
        true_tags, pred_tags = trainer.evaluate(test_loader, id_to_tag)
        
        # Classification report
        report = classification_report(true_tags, pred_tags, output_dict=True, zero_division=0)
        
        # Store results
        results[model_name] = {
            'precision_macro': report['macro avg']['precision'],
            'recall_macro': report['macro avg']['recall'],
            'f1_macro': report['macro avg']['f1-score'],
            'precision_weighted': report['weighted avg']['precision'],
            'recall_weighted': report['weighted avg']['recall'],
            'f1_weighted': report['weighted avg']['f1-score'],
            'accuracy': report['accuracy']
        }
        
        # Print detailed report
        print(classification_report(true_tags, pred_tags, zero_division=0))
        
        # Plot confusion matrix
        plot_confusion_matrix(true_tags, pred_tags, list(id_to_tag.values()), model_name)
    
    # Print comparative results
    print("\n" + "="*80)
    print("COMPARATIVE RESULTS")
    print("="*80)
    print(f"{'Model':<20} {'Macro F1':<10} {'Weighted F1':<12} {'Accuracy':<10}")
    print("-"*80)
    
    for model_name, metrics in results.items():
        print(f"{model_name:<20} {metrics['f1_macro']:.4f}     {metrics['f1_weighted']:.4f}       {metrics['accuracy']:.4f}")
    
    return results
````

````python

def plot_confusion_matrix(y_true, y_pred, labels, model_name):
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=labels, yticklabels=labels)
    plt.title(f'Confusion Matrix - {model_name}')
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.xticks(rotation=45)
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
````

````python
# Main execution

# Load and prepare data
print("Loading data...")
train_data = load_conll_file("conll_output/train.conll")
val_data = load_conll_file("conll_output/val.conll")
test_data = load_conll_file("conll_output/test.conll")

print("Creating mappings...")
word_to_id, id_to_word, tag_to_id, id_to_tag = create_mappings(train_data)

print("Preparing corpus...")
train_corpus = prepare_corpus(train_data, word_to_id, tag_to_id)
val_corpus = prepare_corpus(val_data, word_to_id, tag_to_id)
test_corpus = prepare_corpus(test_data, word_to_id, tag_to_id)

# Create datasets
train_dataset = NERDataset(train_corpus, word_to_id, tag_to_id)
val_dataset = NERDataset(val_corpus, word_to_id, tag_to_id)
test_dataset = NERDataset(test_corpus, word_to_id, tag_to_id)

# Create data loaders
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

vocab_size = len(word_to_id)
tagset_size = len(tag_to_id)

print(f"Vocabulary size: {vocab_size}")
print(f"Number of tags: {tagset_size}")
print(f"Training samples: {len(train_dataset)}")
print(f"Validation samples: {len(val_dataset)}")
print(f"Test samples: {len(test_dataset)}")

# Initialize models
models = {
    'Vanilla LSTM': VanillaLSTM(vocab_size, tagset_size, embedding_dim=128, hidden_size=256),
    'BiLSTM': BiLSTM(vocab_size, tagset_size, embedding_dim=128, hidden_size=256),
    '2-Layer BiLSTM': MultiLayerBiLSTM(vocab_size, tagset_size, embedding_dim=128, hidden_size=256, num_layers=2),
    '3-Layer BiLSTM': MultiLayerBiLSTM(vocab_size, tagset_size, embedding_dim=128, hidden_size=256, num_layers=3)
}

# Train and evaluate models
models_dict = {}

for model_name, model in models.items():
    print(f"\n{'='*50}")
    print(f"Training {model_name}")
    print(f"{'='*50}")
    
    trainer = NERTrainer(model, device, lr=0.001)
    train_losses, val_f1_scores = trainer.train(train_loader, val_loader, epochs=6)
    models_dict[model_name] = (model, trainer)

# Comparative evaluation
print("\n" + "="*80)
print("FINAL COMPARATIVE EVALUATION ON TEST SET")
print("="*80)

results = compare_models(models_dict, test_loader, id_to_tag)

# Print computational analysis
print("\n" + "="*80)
print("COMPUTATIONAL ANALYSIS")
print("="*80)

for model_name, (model, _) in models_dict.items():
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    
    print(f"{model_name}:")
    print(f"  Total parameters: {total_params:,}")
    print(f"  Trainable parameters: {trainable_params:,}")
    
    # Measure inference time
    model.eval()
    with torch.no_grad():
        start_time = torch.cuda.Event(enable_timing=True) if torch.cuda.is_available() else None
        end_time = torch.cuda.Event(enable_timing=True) if torch.cuda.is_available() else None
        
        if torch.cuda.is_available():
            start_time.record()
        else:
            start_time = time.time()
        
        # Run inference on a sample batch
        sample_batch = next(iter(test_loader))
        word_ids = sample_batch['word_ids'].to(device)
        _ = model(word_ids)
        
        if torch.cuda.is_available():
            end_time.record()
            torch.cuda.synchronize()
            inference_time = start_time.elapsed_time(end_time)
        else:
            end_time = time.time()
            inference_time = (end_time - start_time) * 1000  # Convert to milliseconds
        
        print(f"  Inference time (1 batch): {inference_time:.2f} ms")
    print()
````
